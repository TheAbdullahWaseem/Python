#Python Dictionaries

#Phase 1: Understanding
##1. Dictionaries Internals
####Dictionaries use a hash table internally.
####Each key is passed through a hash function → determines storage location.
###This is why:
####1. Lookup → O(1) (constant time)
####2. Insert/Delete → O(1) (average case)

```python
Student = {"Ali": 90, "Sana": 85, "Omar": 78}
print(Student["Ali"]) #O(1) Lookup
```

##2. Hashable & Immutable

###Dictionary keys must be:

####Immutable (cannot change)
####Hashable (has fixed hash value)
###Allowed Keys
####**int, float, str, tuple (if immutable inside)**
###Not Allowed
####**list, dict, set**

```python
# Valid
data = {
    "name": "Ali",
    101: "ID",
    (1, 2): "tuple key"}
# invalid
#data = {[1, 2]: "list key"} will raise error
```

##4. Safer Access Patterns


####Avoid errors when key is missing.
###Methods
####.get() (Recommended)

```python
Student = dict(Ali = 90, Hashir = 100, Azaan = 80 )
print(Student.get("Ali"))
print(Student.get("Abdullah", 0))
```

####Just retrieves value safely. If not exists returns 78
####Does NOT store anything

```python
Student['Faizan'] = Student.get("Faizan", 78)
```

####Checks if "Faizan" exists. If exists, gets existing value. If not uses 78
####Then assigns it back to dictionary.

```python
print(Student)
```

Here Abdullah doesn't exist but instead of giving error, it returns 0

####.setdefault()

```python
Student.setdefault("Sana", 95)
Student.setdefault("Hira", []).append("90")
Student.setdefault("Hira", []).append("92")
print(Student)
```

####Direct Access

```python
#Student['Noor']
#It will raise an error
print(Student.get("Noor", 0)) #It is safe access, no error
```

##5. Shallow vs Deep Copy


####**Shallow Copy**	References nested objects
####**Deep Copy**	Fully independent copy

```python
import copy
original = {"a": [1,2,3]}

#Shallow copy
shallow = original.copy()
shallow["a"][0] = 90
print(original)
```

Changed

```python
deep = copy.deepcopy(original)
deep["a"][0] = 0
print(original)
```

Unchanged original and deep changed

```python
print(deep)
```

##9. Sorting Dictionaries

####Dictionaries are insertion-ordered , but not automatically sorted.

###Sort by Keys

```python
data = {"b": 2, "a": 1, "c": 3}
sorted_by_keys = dict(sorted(data.items()))
sorted_by_keys
```

###Sort by Values

```python
sorted_by_values = dict(sorted(data.items(), key= lambda x: x[1]))
sorted_by_values
```

##Descending Order

```python
Descend = dict(sorted(data.items(), key= lambda x: x[1], reverse= True))
Descend
```

###Custom Sorting

```python
#Sort by length of key
datum = {"Ali": 90, "Abdullah":99, "Faizan": 94}
custom = dict(sorted(datum.items(), key= lambda x: len(x[0])))
custom
```

####sorted() returns list of tuples
####Convert back using dict()

##10. Reversing / Inverting Dictionaries

####Swap keys ↔ values

```python
data = {"a": 1, "b": 2}
inverted = {v: k for k, v in data.items()}
inverted
```

####If values are not unique, data will be lost


```python
data = {"a": 1, "b": 1}
data
```

###Safe Inversion (Handling duplicates)

```python
from collections import defaultdict
data = {"a": 1, "b": 1}

inverted = defaultdict(list)
for k, v in data.items():
  inverted[v].append(k)
inverted
```

#PHASE 3: Usage of Dictionaries

##14. Frequency Counters (Character / Word Count)


####Used to count occurrences very common in:

####Text analysis
####Log analysis
####Data preprocessing

###Character Frequency

```python
text = 'data analysis'

freq = {}
for char in text:
  if char != " ":
    freq[char] = freq.get(char, 0) + 1
print(freq)
```

###Word Frequency

```python
text = "python is easy and python is powerful"
words = text.split()

freq = {}
for word in words:
  freq[word] = freq.get(word, 0) + 1
print(freq)
```

##15. Grouping Data (Like SQL GROUP BY)


####Group items based on a key, similar to SQL:

```python
data = [
    ("Ali", "IT"),
    ("Sana", "HR"),
    ("Omar", "IT"),
    ("Ayesha", "HR")]
grouped = {}
for k, v in data:
  grouped.setdefault(v, []).append(k)
grouped
```

```python
from collections import defaultdict
grouping  = defaultdict(list)
for k, v in data:
  grouping[v].append(k)
grouping
```

heavily used in data aggregation and reporting systems.

##18. Fast Lookup Optimization

##23. ChainMap

####Combines multiple dictionaries into one logical view

```python
from collections import ChainMap
d1 = {"a": 1}
d2 = {"b": 2}
combined = ChainMap(d1, d2)
combined
```

###Priority Rule

```python
d1 = {"a": 1}
d2 = {"a": 2}

combined = ChainMap(d1, d2)
print(combined["a"])
```

First Dict value will be preferred
