# 1. List Comprehensions (Review)

A **list comprehension** provides a concise and readable way to create a new list by applying an expression to each element of an iterable.

Compared to traditional `for` loops, list comprehensions are often shorter, more expressive, and considered more Pythonic when the logic is simple.

## 1.1 Basic List Comprehension

**Syntax**

```python
[expression for item in iterable]
```

The expression is evaluated for every element in the iterable, and the results are collected into a new list.

````python
numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]

print(squares)
````

## 1.2 List Comprehension with a Condition

A condition can be added using an `if` clause to include only elements that satisfy a specific condition.

**Syntax**

```python
[expression for item in iterable if condition]
```

````python
numbers = range(1, 11)

even_numbers = [num for num in numbers if num % 2 == 0]

print(even_numbers)
````

## 1.3 Nested List Comprehension

A nested list comprehension uses multiple `for` clauses to iterate over more than one iterable.

**Syntax**

```python
[expression for x in iterable1 for y in iterable2]
```

````python
colors = ["Red", "Blue"]
sizes = ["S", "M", "L"]

products = [f"{color}-{size}" for color in colors for size in sizes]
print(products)
````

## 1.4 Flattening Nested Lists

A common use of nested list comprehensions is converting a nested list into a single flat list.

````python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

flattened = [item for row in matrix for item in row]
print(flattened)
````

## 1.5 When **Not** to Use List Comprehensions

List comprehensions are best suited for **simple transformations**.

Avoid them when:

- The expression becomes too complex.
- Multiple nested conditions reduce readability.
- The logic spans several steps.
- Readability is more important than writing fewer lines of code.

In such cases, a traditional `for` loop is often clearer and easier to maintain.

````python
# Good: Simple and readable
squares = [num ** 2 for num in range(5)]

print(squares)
````

````python
# Better written as a for loop due to complex logic
result = []

for number in range(20):
    if number % 2 == 0:
        if number % 3 == 0:
            result.append(number ** 2)

print(result)
````

## Professional Notes

- List comprehensions provide a concise and Pythonic way to create new lists.
- Use a basic comprehension for straightforward transformations.
- Add an `if` clause to filter elements during creation.
- Nested comprehensions are useful for iterating over multiple iterables and flattening nested lists.
- Prioritize readability over brevity. If a comprehension becomes difficult to understand, replace it with a regular `for` loop.

# 2. Dictionary Comprehensions (Review)

A **dictionary comprehension** provides a concise and Pythonic way to create a new dictionary by generating key-value pairs from an iterable.

Like list comprehensions, dictionary comprehensions improve code readability for simple dictionary transformations and filtering operations.

## 2.1 Basic Dictionary Comprehension

**Syntax**

```python
{key_expression: value_expression for item in iterable}
```

When iterating over an existing dictionary, `items()` is commonly used to access both keys and values.

````python
numbers = [1, 2, 3, 4, 5]

squares = {num: num ** 2 for num in numbers}

print(squares)
````

````python
students = {
    "Ali": 85,
    "Ahmed": 90,
    "Sara": 88
}

grades = {name: marks for name, marks in students.items()}

print(grades)
````

## 2.2 Dictionary Comprehension with Filtering

An `if` condition can be used to include only selected key-value pairs in the new dictionary.

**Syntax**

```python
{
    key: value
    for key, value in dictionary.items()
    if condition
}
```

````python
students = {
    "Ali": 85,
    "Ahmed": 90,
    "Sara": 72,
    "Ayesha": 95
}

top_students = {
    name:marks
    for name, marks in students.items()
    if marks >= 85
}

print(top_students)
````

## 2.3 Inverting a Dictionary (Swapping Keys and Values)

Dictionary comprehensions can be used to swap keys and values.

**Syntax**

```python
{value: key for key, value in dictionary.items()}
```

**Note:** This works correctly only when all values are unique. If duplicate values exist, later keys overwrite earlier ones.

````python
countries = {
    "PK": "Pakistan",
    "SA": "Saudi Arabia",
    "TR": "Turkey"
}

inverted = {
    value:key
    for key, value in countries.items()
}
print(inverted)
````

## 2.4 Counting with Dictionary Comprehensions

Dictionary comprehensions can generate summary dictionaries by calculating values during creation.

A common example is counting the occurrences of each unique element.

````python
text = "banana"

counts = {
    char:text.count(char)
    for char in set(text)
}

print(counts)
````

````python
numbers = [1, 2, 2, 3, 3, 3]

counts = {
    num:numbers.count(num)
    for num in set(numbers)
}

print(counts)
````

## Professional Notes

- Dictionary comprehensions provide a concise way to create new dictionaries.
- Use filtering to include only key-value pairs that satisfy a condition.
- Dictionary inversion is useful when values are unique and need to become keys.
- Dictionary comprehensions can also generate summary dictionaries, such as counting occurrences of elements.
- Keep dictionary comprehensions simple; if the logic becomes difficult to read, use a regular `for` loop instead.

# 3. Set Comprehensions (Review)

A **set comprehension** provides a concise and Pythonic way to create a new set by applying an expression to each element of an iterable.

Like regular sets, the resulting collection contains **only unique elements**. Duplicate values are automatically removed.

## 3.1 Basic Set Comprehension

**Syntax**

```python
{expression for item in iterable}
```

Each expression is evaluated, and the results are stored as unique elements in the new set.

````python
numbers = [1, 2, 3, 4, 5]

squares = {num ** 2 for num in numbers}

print(squares)
````

````python
words = ["python", "data", "analysis"]

uppercase = {word.upper() for word in words}

print(uppercase)
````

## 3.2 Deduplication Using Set Comprehensions

Since sets store only unique values, set comprehensions are commonly used to remove duplicate data while applying a transformation.

This is particularly useful during data cleaning.

````python
numbers = [1, 2, 2, 3, 3, 3, 4, 5]

unique_numbers = {num for num in numbers}

print(unique_numbers)
````

````python
names = [
    "Ali",
    "Ahmed",
    "Ali",
    "Sara",
    "Ahmed"
]

unique_names = {name.upper() for name in names}

print(unique_names)
````

## Professional Notes

- Set comprehensions create sets using a concise and readable syntax.
- Duplicate values are removed automatically because sets store only unique elements.
- Set comprehensions are commonly used for deduplication and simple data transformations.
- Since sets are unordered collections, the output order should not be relied upon.

# 4. Generator Expressions

A **generator expression** is a concise way to create a **generator object**, which produces values one at a time as they are needed.

Its syntax is very similar to a list comprehension, but instead of creating an entire list immediately, it creates a generator that can be iterated over.

Generator expressions are particularly useful when processing large amounts of data efficiently or when values only need to be used once.

## 4.1 What is a Generator Expression?

A generator expression is an expression that generates values **on demand** rather than storing all of them in memory at once.

It is created using **parentheses `()`** instead of square brackets `[]`.

Because a generator expression does not immediately create all values, it is generally more memory-efficient than a list comprehension when working with large datasets.

> **Note:** The concepts of lazy evaluation and the internal working of generators will be covered in detail in the **Iterators and Generators** notebook.

### Why Do Generator Expressions Exist?

Generator expressions were introduced to provide a concise way to process sequences without creating an entire list in memory.

They are especially useful when:

- Processing large datasets.
- Passing values directly to functions such as `sum()` or `max()`.
- The values only need to be used once.
- Memory efficiency is important.

### Basic Syntax

The general syntax of a generator expression is:

```python
(expression for item in iterable)
```

It closely resembles a list comprehension, with one important difference:

- **List comprehension:** uses square brackets `[]`
- **Generator expression:** uses parentheses `()`

### Generator Expression vs List Comprehension (Syntax)

| List Comprehension | Generator Expression |
|--------------------|----------------------|
| `[expression for item in iterable]` | `(expression for item in iterable)` |

The only visible difference is the type of brackets used, but the resulting objects are different.

### Simple Example

The following example creates a generator that produces the square of each number from 1 to 5.

````python
numbers = range(1, 6)

squares = (num ** 2 for num in numbers)

print(squares)
````

Notice that Python does **not** display all the generated values.

Instead, it displays a generator object.

To access the values, the generator must be iterated over, for example using a `for` loop.

````python
numbers = range(1, 6)

squares = (num**2 for num in numbers)

for value in squares:
  print(value)
````

### Using `type()`

The built-in `type()` function can be used to identify the object created by a generator expression.

````python
numbers = range(5)

generator = (num for num in numbers)

print(type(generator))
````

````python
numbers = range(5)

lst = [num for num in numbers]
gen = (num for num in numbers)

print(type(lst))
print(type(gen))
````

## Key Differences (So Far)

| Feature | List Comprehension | Generator Expression |
|---------|--------------------|----------------------|
| Brackets | `[]` | `()` |
| Returns | `list` | `generator` |
| Stores all values immediately | Yes | No |
| Can be indexed | Yes | No |
| Best for | Small to medium collections that need repeated access | Values that will be processed sequentially |

## Professional Notes

- Generator expressions use parentheses `()` instead of square brackets `[]`.
- A generator expression returns a **generator object**, not a list.
- Generator expressions are ideal when values only need to be processed sequentially.
- They are commonly used with built-in functions such as `sum()`, `max()`, and `min()`.
- The internal concepts that make generators memory-efficient—such as lazy evaluation and the iterator protocol—will be explored in detail in the **Iterators and Generators** notebook.

## 4.2 Basic Generator Expression Examples

Generator expressions can be used wherever values need to be produced sequentially.

Their syntax is nearly identical to list comprehensions, making them easy to learn if you are already familiar with comprehensions.

The following examples demonstrate some common use cases.

### Example 1: Squaring Numbers

A generator expression can apply an expression to every element of an iterable.

```python
(expression for item in iterable)
```

````python
numbers = range(1, 6)

squares = (num ** 2 for num in numbers)

for square in squares:
    print(square)
````

### Example 2: Filtering Even Numbers

Like list comprehensions, generator expressions can include an `if` condition.

**Syntax**

```python
(expression for item in iterable if condition)
```

````python
numbers = range(1, 11)

even_numbers = (
    num
    for num in numbers
    if num % 2 == 0
)

for number in even_numbers:
    print(number)
````

### Example 3: Creating a Generator from a String

Since strings are iterable, a generator expression can process one character at a time.

````python
word = "Python"

characters = (char for char in word)

for char in characters:
    print(char)
````

### Example 4: Applying a Transformation to Each Character

Generator expressions can transform values while iterating.

````python
word = "python"

uppercase = (
    char.upper()
    for char in word
)

for char in uppercase:
  print(char)
````

### Example 5: Iterating Over a Generator with a `for` Loop

The most common way to access the values produced by a generator expression is by using a `for` loop.

````python
numbers = (num for num in range(5))

for value in numbers:
    print(value)
````

### Generator Expressions Can Be Assigned to Variables

A generator expression is often assigned to a variable before it is used.

````python
names = (
    name.title()
    for name in ['ali', 'ahmad', 'sara']
)
for name in names:
  print(name)
````

## Professional Notes

- Generator expressions use the same structure as list comprehensions but are enclosed in parentheses `()`.
- They can perform transformations and filtering just like list comprehensions.
- Any iterable, such as lists, tuples, strings, or `range()` objects, can be used with a generator expression.
- A `for` loop is the most common way to iterate over the values produced by a generator expression.
- Generator expressions are particularly useful when values only need to be processed once.

## 4.3 Generator Expressions vs List Comprehensions

Generator expressions and list comprehensions share almost identical syntax and are often used for similar tasks. However, they differ in the type of object they create and how values are produced.

Choosing between them depends on how the resulting data will be used.

### Comparison

| Feature | List Comprehension | Generator Expression |
|----------|--------------------|----------------------|
| Syntax | `[]` | `()` |
| Returns | `list` | `generator` |
| Memory Usage | Higher | Lower |
| Value Creation | Creates all values immediately | Produces values as they are needed |
| Indexing | Supported | Not supported |
| Reusable | Yes | Typically consumed once |

### Syntax Difference

The syntax differs only in the brackets used.

**List Comprehension**

```python
squares = [num ** 2 for num in range(5)]
```

**Generator Expression**

```python
squares = (num ** 2 for num in range(5))
```

Although the syntax is nearly identical, the resulting objects are different.

### Returned Object

````python
numbers = range(5)

list_comp = [num for num in numbers]
gen_expr = (num for num in numbers)

print(type(list_comp))
print(type(gen_expr))
````

### Memory Usage

A list comprehension stores every generated value in memory immediately.

A generator expression does **not** store all values at once. Instead, it produces values one at a time when they are needed.

As a result, generator expressions generally require less memory, especially when working with large datasets.

> **Note:** The mechanism behind this behavior will be explained in detail in the **Iterators and Generators** notebook.

### Value Creation

Consider the following examples:

**List Comprehension**

```python
numbers = [num for num in range(5)]
```

The list is created immediately.

**Generator Expression**

```python
numbers = (num for num in range(5))
```

The generator produces each value only when it is requested.

This makes generator expressions suitable for processing data sequentially without creating an entire list first.

### Indexing

Lists support indexing because all elements are stored in memory.

Generators do not support indexing.

````python
numbers = [10, 20, 30]

print(numbers[0])
````

````python
numbers = (num for num in [10, 20, 30])

#print(numbers[0])

#TypeError: 'generator' object is not subscriptable
````

### Reusability

A list can be accessed multiple times because all of its elements are stored.

A generator expression is generally intended for a single pass. Once all values have been produced, the generator is exhausted.

````python
numbers = [1, 2, 3]

for num in numbers:
    print(num)

print("Again:")

for num in numbers:
    print(num)
````

````python
numbers = (num for num in [1, 2, 3])

print(type(numbers))

for num in numbers:
    print(num)

print("Again:")

for num in numbers:
    print(num)
````

## Which One Should You Choose?

Choose a **list comprehension** when:

- You need to access elements multiple times.
- You need indexing or slicing.
- You need the complete collection stored in memory.

Choose a **generator expression** when:

- You only need to process the values once.
- You are working with large datasets.
- You want a more memory-efficient solution.

## Professional Notes

- Generator expressions and list comprehensions have nearly identical syntax.
- A list comprehension creates and stores all values immediately.
- A generator expression produces values as they are needed instead of storing them all at once.
- Lists support indexing and repeated iteration, while generators are typically consumed once.
- The concepts of **lazy evaluation**, the **iterator protocol**, and the internal implementation of generators will be covered in detail in the **Iterators and Generators** notebook.

## 4.4 Using Generator Expressions with Built-in Functions

Generator expressions are commonly passed directly to Python's built-in functions.

This avoids creating an intermediate list and results in concise, readable code.

The pattern is:

```python
function(expression for item in iterable)
```

Since the generator expression is the only argument, the outer parentheses of the generator expression can be omitted.

### Using `sum()`

The `sum()` function calculates the total of all generated values.

````python
numbers = range(1, 6)

total = sum(num for num in numbers)

print(total)
````

````python
numbers = range(1, 6)

sum_of_squares = sum(num ** 2 for num in numbers)

print(sum_of_squares)
````

### Using `max()`

The `max()` function returns the largest generated value.

````python
numbers = [15, 42, 18, 73, 25]

largest = sum(num for num in numbers)

print(largest)
````

````python
words = ["Python", "Machine", "AI", "Data Science"]

longest_length = max(len(word) for word in words)

print(longest_length)
````

### Using `min()`

The `min()` function returns the smallest generated value.

````python
numbers = [15, 42, 18, 73, 25]

smallest = min(num for num in numbers)

print(smallest)
````

````python
words = ["Python", "Machine", "AI", "Data Science"]

shortest_length = min(len(word) for word in words)

print(shortest_length)
````

### Using `any()`

The `any()` function returns `True` if **at least one** generated value evaluates to `True`. Otherwise, it returns `False`.

It is commonly used for checking whether a condition is satisfied by any element in an iterable.

````python
scores = [75, 82, 91, 68]

has_high_score = any(score > 90 for score in scores)

print(has_high_score)
````

````python
numbers = [2, 4, 6, 8]

has_odd = any(num % 2 != 0 for num in numbers)

print(has_odd)
````

### Using `all()`

The `all()` function returns `True` only if **every** generated value evaluates to `True`.

If even one value evaluates to `False`, the result is `False`.

It is commonly used for validating that all elements satisfy a condition.

````python
ages = [20, 24, 31, 18]

all_adults = all(age >= 18 for age in ages)

print(all_adults)
````

````python
numbers = [2, 4, 6, 7]

all_even = all(num % 2 == 0 for num in numbers)

print(all_even)
````

### Why Use Generator Expressions with Built-in Functions?

When passing values directly to built-in functions such as `sum()`, `max()`, `min()`, `any()`, or `all()`, a generator expression is often preferred because the values are processed directly without first creating a separate list.

For example:

```python
sum(num ** 2 for num in range(10))
```

is generally preferred over:

```python
sum([num ** 2 for num in range(10)])
```

Both produce the same result, but the first approach avoids creating an unnecessary intermediate list.

## Professional Notes

- Generator expressions work seamlessly with many built-in functions.
- `sum()`, `max()`, and `min()` are commonly used for numerical calculations.
- `any()` checks whether at least one element satisfies a condition.
- `all()` checks whether every element satisfies a condition.
- When the generated values are used only once, passing a generator expression directly to a built-in function is considered clean and Pythonic.

## 4.5 When to Prefer a Generator Expression

Generator expressions are not intended to replace list comprehensions. Both have their own strengths, and choosing the appropriate one depends on how the data will be used.

As a general rule:

- Use a **generator expression** when values only need to be processed once.
- Use a **list comprehension** when you need to store and reuse the generated values.

### 1. When You Only Need to Iterate Once

If the generated values are processed a single time, a generator expression is usually the better choice.

**Example**

```python
total = sum(num ** 2 for num in range(100))
```

In this case, the values are generated only for calculating the sum and are not needed afterward.

### 2. When Working with Large Datasets

Generator expressions are well suited for processing large datasets because they generate values as they are needed instead of storing every value in memory.

This makes them a common choice in:

- Data Analysis
- Machine Learning
- File processing
- Data pipelines

> **Note:** The memory-saving behavior will be explained in detail in the **Iterators and Generators** notebook.

### 3. When Passing Values Directly to Built-in Functions

Generator expressions are ideal when values are immediately consumed by functions such as:

- `sum()`
- `max()`
- `min()`
- `any()`
- `all()`

**Preferred**

```python
sum(num ** 2 for num in range(100))
```

Instead of:

```python
sum([num ** 2 for num in range(100)])
```

The first approach is shorter and avoids creating an unnecessary intermediate list.

### 4. When You Do Not Need Indexing

Generator expressions should be used only when sequential processing is sufficient.

If you do not need to access elements by index or retrieve specific positions, a generator expression is often a suitable choice.

## When **Not** to Use a Generator Expression

### 1. When You Need to Reuse the Values

A generator expression is generally intended for a single pass.

If the generated values will be used multiple times, store them in a list instead.

```python
squares = [num ** 2 for num in range(10)]
```

### 2. When You Need Indexing or Slicing

Generator expressions do not support indexing or slicing.

If you need operations such as:

```python
values[0]
values[2:5]
```

use a list instead.

### 3. When You Need the Length

The `len()` function works with lists but not with generator objects.

```python
numbers = [num for num in range(10)]

len(numbers)
```

A generator expression does not know its total length in advance, so `len()` cannot be used directly on it.

````python
numbers = (num for num in range(10))

#len(numbers)
#TypeError: object of type 'generator' has no len()
````

## Practical Guidelines

Use a **generator expression** when:

- The values are processed only once.
- You are working with large datasets.
- You are passing values directly to functions such as `sum()`, `max()`, `min()`, `any()`, or `all()`.
- You do not require indexing or slicing.

Use a **list comprehension** when:

- You need to access elements by index.
- You need the length of the collection using `len()`.
- You need to reuse the generated values multiple times.
- You want to store all generated values in memory.

## Professional Notes

- Generator expressions are designed for sequential processing of values.
- They are especially useful when values are consumed immediately by another function.
- List comprehensions remain the better choice when you need indexing, slicing, repeated iteration, or the length of the collection.
- Choosing between a list comprehension and a generator expression depends on how the generated data will be be used rather than which syntax is shorter.

````python
numbers = [2,4,6,7,8,9]

total = sum(num for num in numbers)
print(total)
````

````python
iteration = (num for num in numbers)

for num in iteration:
  print(num)
````
