#Level 1: Core Concepts
##  1. Tuple Creation

A **tuple** is an ordered, immutable collection of elements.

###  Key Characteristics:
- Ordered (index-based)
- Immutable (cannot change after creation)
- Allows duplicate values
- Can store multiple data types

---

###  Creating Tuples

```python
t1 = (1,2,3)
print(t1)
```

```python
t2 = (1, "Abdullah", 95.5)
print(t2)
```

```python
t3 = 1,2,3
print(t3)
```

### Single Element Tuple

A single value inside parentheses is NOT a tuple unless you add a comma.

```python
t = (5)
print(type(t))
```

```python
t = (5,)
print(type(t))
```

##  2. Tuple Indexing & Slicing

Since tuples are ordered, we can access elements using **indexing and slicing**.

---

###  Indexing
- Starts from 0
- Negative indexing allowed

```python
t = 'Ali', 'Sana', 'Omar', 'Zara'
print(t[0])
print(t[-1])
print(t[0:2])
```

###  Slicing

Syntax:
tuple[start : end : step]

```python
t = 10,20,30,40,50
print(t[0:5:2])
print(t[1:4])
print(t[::2])
```

##  3. Immutability

Tuples are **immutable**, meaning:
- You CANNOT change elements
- You CANNOT add or remove elements

---

###  Not Allowed:
- Changing values
- Appending items
- Removing items

---

###  Why It Matters:
- Ensures **data safety**
- Makes tuples **faster than lists**
- Allows usage as **dictionary keys**

```python
t = (1,2,3)
#t[0] = TypeError: 'tuple' object does not support item assignment
```

###  Workaround (Convert → Modify → Convert Back)

```python
t = (1,2,3)
temp = list(t)
temp[0] = 99
t = tuple(temp)

print(t)
print(type(t))
```

1. count()

---

###  count()
Returns how many times a value appears

```python
t = (1,2,3,3,4,2,5)
print(t.count(2))
```

###  index()
Returns the index of the FIRST occurrence of a value

```python
t = (10, 20, 30, 20)
print(t.index(20))
```

##  5. Tuple Packing & Unpacking

###  Packing
Packing means putting multiple values into a tuple.

###  Unpacking
Unpacking means extracting values from a tuple into variables.

---

###  Why Important?
- Clean and readable code
- Used heavily in functions and data processing

```python
#Packing
student = (101, 'Ali', 95)

#Unpacking
id, name, marks = student

print(id)
print(name)
print(marks)
```

### Rule:
Number of variables must match number of values

```python
#id, name = student
#This will fail
```

##  6. Extended Unpacking (*args Style)

Extended unpacking allows you to capture **multiple values into one variable** using `*`.

---

###  Syntax:
a, *b, c = tuple

- a → first value
- c → last value
- b → remaining values (as a list)

```python
t= 1,2,3,4,5
a, *b, c = t

print(a)
print(b)
print(c)
```

```python
*a, b = (10,20,30,40)
print(a,b)

a, *b = (10,20,30,40)
print(a,b)
```

##  8. Tuples Containing Mutable Objects

Even though tuples are immutable, they can contain **mutable objects** like lists.

---

###  Important Concept:
- You cannot change the tuple itself
- BUT you can modify the object inside it

##  11. Tuples as Function Return Values

Functions in Python can return **multiple values**, and they are returned as a tuple.

---

###  Why Important?
- Clean way to return multiple results

##  12. Tuples as Function Arguments

Tuples can be passed to functions as arguments.

---

###  Two Ways:
1. Pass tuple as a single argument
2. Unpack tuple into multiple arguments

```python
def display(student):
  print(student)

t = (101, 'Ali', 95)
display(t)
```

###Unpacking into Function

```python
def display(id, name, marks):
  print(name, marks)

t = (101, 'Abdullah', 95)
result = display(*t)
result
```

##  13. Swapping Values Using Tuples

Python allows swapping values in a **single line** using tuple unpacking.

---

###  Syntax:
a, b = b, a

---

###  Why Important?
- Cleaner than traditional swapping
- No temporary variable needed

```python
a = 10
b = 20
print('a is:', a)
print('b is:', b)

a, b = b, a
print('a is now:', a)
print('b is now:', b)
```

Python internally creates a tuple:

(a, b) = (b, a)

Then unpacks it.

##  14. Iterating with enumerate() and Tuples

enumerate() adds an index to each element while looping.

---

###  Why Important?
- Cleaner loops
- Avoid manual index tracking

```python
names = ("Ali", "Sana", "Omar")

for name in enumerate(names):
  print(name)
```

###More Pythonic way

```python
names = ("Ali", "Sana", "Omar")

for index, name in enumerate(names):
  print(index, name)
```

###With Nested Tuples

```python
records = (
    (1, "Ali", 90),
    (2, "Sana", 95)
)
for index, (id, *val) in enumerate(records):
  print(index, val)
```

enumerate() returns tuples:
(index, value)

That’s why unpacking works perfectly here.

```python
for record in enumerate(records):
  index, *values = record
  print(index, values)
```

##  15. zip() and Tuples

zip() combines multiple iterables into **tuples**.

---

###  Syntax:
zip(list1, list2)

---

###  Output:
Iterator of tuples:
(value1, value2)

```python
names = ["Ali", "Sana", "Omar"]
marks = [90, 95, 85]

zipped = list(zip(names, marks))
print(zipped)
```

###Iterating with Zip

```python
for name, mark in zip(names, marks):
  print(name, mark)
```

```python
for name, mark in enumerate(zip(names, marks)):
  print(name, mark)
```

###Multiple Lists

```python
names = ["Ali", "Sana", "Omar"]
marks = [90, 95, 85]
ids = [1, 2, 3]

for id, name, mark in zip(ids, names, marks):
  print(name, mark)
```

###Unequal Lengths

```python
a = [1, 2, 3]
b = [10, 20]

print(list(zip(a, b))) #Stops at shortest
```

###Unzipping

```python
pairs = [(1, "Ali"), (2, "Sana")]
ids, names = zip(*pairs)
```

```python
print(ids)
print(names)
```

###This way, Tuples unpacked values in same list

```python
pairs = [(1, "Ali"), (2, "Sana")]

unpacked = list(zip(*pairs))
print(unpacked)
```

###This way, Tuples unpacked into different tuples

```python
pairs = [(1, "Ali"), (2, "Sana")]

id, name = list(zip(pairs))
print(id)
print(name)
```

##  17. Tuples as Dictionary Keys (Hashability)

Tuples can be used as dictionary keys because they are **hashable**.

---

###  What is Hashability?

A value is hashable if:
- It does NOT change (immutable)
- It has a fixed hash value

---

###  Lists cannot be keys
Because they are mutable

```python
#  Tuple as key
locations = {
    (34.015, 71.580): "Peshawar",
    (24.860, 67.010): "Karachi"
}

print(locations[(34.015, 71.580)])
```

```python
# This will fail
# d = {[1, 2]: "Invalid"}
```

A tuple is hashable ONLY IF all its elements are hashable.

Example:
- (1, 2, 3) →  valid
- (1, [2, 3]) →  invalid (list inside)

##  18. Immutability vs Mutability (Deep Understanding)

This is a **core Python concept**.

---

###  Immutable Objects:
- Cannot change after creation
- Examples: tuple, int, string

###  Mutable Objects:
- Can change after creation
- Examples: list, dict, set

```python
# Immutable example
x = 10
x = x + 5  # creates new object
print(x)
# Mutable example
l = [1, 2, 3]
l.append(4)  # modifies same object
print(l)
```

```python
t = (1, [2, 3])
t[1].append(4)
print(t)
```

 Tuple immutability means:
- Structure cannot change

 BUT:
- Objects inside can change if they are mutable

This is called:
 "Shallow immutability"

## 23. Named Tuples (collections.namedtuple)

Named tuples make tuples more readable by assigning names to fields.

---

### Problem with normal tuples:
t = (1, "Ali", 90)

What is t[1]? Not clear.

---

### Solution:
Use named tuples for clarity

```python
from collections import namedtuple

student = namedtuple("Student", ["id", "name", "marks"])
s = student(1, "Ali", 90)
print(s.name)
```

- Readable code
- Still immutable
- Works like both tuple and object

## Creating namedtuple Objects from Records (Clean Approach)

Instead of manually writing values, we use existing data (records) and convert them into namedtuple objects.

---

### Key Idea:
Use unpacking (*) to pass tuple values into namedtuple

```python

Student = namedtuple("Student", ["id", "name", "marks"])

records = [
    (1, "Ali", 90),
    (2, "Sana", 95),
    (3, "Omar", 85)
]

students = [Student(*record) for record in records]

for student in students:
    print(student.name)
```

## 24. Pattern-Style Unpacking

Pattern-style unpacking improves readability when working with structured data.

---

### Instead of:
record[0], record[1]

Use:
id, name = record

```python
records = [
    (1, "Ali", ("Math", 90)),
    (2, "Sana", ("English", 95))
]
for id, name, marks in records:
    print(name, marks)
```

```python
for id, name, (sub, marks) in records:
  print(name, sub, marks)
```

This makes code:
- Cleaner
- More readable
- Easier to maintain
