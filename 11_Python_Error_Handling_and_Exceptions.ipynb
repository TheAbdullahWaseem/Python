# 1. What is an Error?

An **error** is any problem that prevents a program from behaving as intended.

Errors may occur because:

- The code contains invalid syntax.
- An unexpected situation occurs while the program is running.
- The program contains incorrect logic.

In simple terms:

> **An error is any condition that causes a program to fail or produce incorrect results.**

Examples include:

- Misspelling Python syntax
- Dividing a number by zero
- Accessing a dictionary key that does not exist
- Writing incorrect business logic

Not every error is handled in the same way. Some errors stop the program before it even starts, while others occur only during execution.

---

# 2. What is an Exception?

An **exception** is a specific type of error that occurs **while a program is running** (runtime).

When Python encounters an unexpected situation during execution, it creates an **exception object** containing information about the problem.

If the exception is not handled, Python stops the program and displays an error message called a **traceback**.

Example:

```python
number = 10
result = number / 0
```

Output

```text
ZeroDivisionError: division by zero
```

Here:

- The program started successfully.
- Python reached an invalid operation.
- Python created a `ZeroDivisionError` exception.
- The exception was not handled.
- Program execution terminated.

> **Every exception is an error, but not every error is an exception.**

For example, a `SyntaxError` prevents the program from starting, so it is not considered a runtime exception.

# 6. Exception Propagation (High-Level Overview)

When an exception occurs, Python immediately stops executing the current statement.

It then looks for code that can handle the exception.

If no suitable handler is found, Python continues searching outward through the call stack.

This process is called **exception propagation**.

Consider the following example.

```python
def function_a():
    function_b()


def function_b():
    function_c()


def function_c():
    print(10 / 0)


function_a()
```

Execution flow

```text
function_a()

    ↓

function_b()

    ↓

function_c()

    ↓

ZeroDivisionError
```

Since no function handles the exception:

- `function_c()` terminates.
- Control returns to `function_b()`.
- `function_b()` also terminates.
- Control returns to `function_a()`.
- Finally, the program terminates with a traceback.

Later in this notebook, you will learn how `try` and `except` can stop this propagation and recover from the exception.

## 1.1 Basic `try` / `except` Block

### What is a `try` Block?

A `try` block contains code that **may raise an exception** during execution.

Python executes the statements inside the `try` block normally. If every statement executes successfully, Python skips the corresponding `except` block.

### What is an `except` Block?

An `except` block contains the code that should execute **only when an exception occurs** inside the associated `try` block.

Its primary purpose is to:

- Handle expected runtime errors.
- Prevent unexpected program termination.
- Display meaningful messages.
- Allow the program to continue executing.

### Basic Syntax

```python
try:
    # Code that may raise an exception
except:
    # Handle the exception
```

### Execution Rule

Python follows these rules:

1. Execute the `try` block.
2. If no exception occurs:
   - Skip the `except` block.
3. If an exception occurs:
   - Stop executing the remaining statements inside the `try` block immediately.
   - Jump to the matching `except` block.
   - Continue executing the program after the exception-handling block.

> **Important:** The `except` block does not prevent exceptions from occurring. It defines how your program should respond when an exception occurs.

````python
try:
    result = 20 / 5
    print(result)

except:
    print("An error occurred.")
````

````python
try:
    result = 20 / 0
    print(result)

except:
    print("Cannot divide by zero.")
````

````python
try:
    age = int("twenty")
    print(age)

except:
    print("Invalid age.")
````

````python
student = {
    "name": "Ali",
    "age": 20
}

try:
    print(student["grade"])

except:
    print("Grade does not exist.")
````

````python
numbers = [10, 20, 30]

try:
    print(numbers[5])

except:
    print("Index is out of range.")
````

````python
try:
  file = open('sales.csv')
  content = file.read()
except:
  print("file not found")
````

### Important Notes

- The `try` block contains code that might fail.
- The `except` block executes only if an exception is raised.
- As soon as an exception occurs, Python immediately stops executing the remaining statements inside the `try` block.
- If no exception occurs, the `except` block is ignored completely.

### Professional Notes

In professional applications, exception handling is commonly used when working with files, databases, APIs, user input, and external resources because these operations are inherently unpredictable.

At this stage, we are using a bare `except` block only to understand the mechanics of exception handling. In production code, catching specific exceptions is considered a best practice and will be covered next.

## 1.2 Program Execution Flow

To use exception handling effectively, it is important to understand **how Python executes a `try` and `except` block**.

Many beginners assume that Python executes every statement inside a `try` block, even if an exception occurs. This is incorrect.

The moment an exception is raised:

- Python immediately stops executing the current `try` block.
- Any remaining statements inside the `try` block are skipped.
- Python searches for a matching `except` block.
- If a matching `except` block is found, it executes that block.
- After the `except` block finishes, normal program execution resumes.

Understanding this execution flow is essential because it explains why some statements are never executed after an exception occurs.

````python
print("Program Started")

try:
    print("Step 1")
    print("Step 2")
    print("Step 3")

except:
    print("An error occurred.")

print("Program Finished")
````

### Explanation

No exception occurred inside the `try` block.

Therefore:

- Every statement inside the `try` block executed successfully.
- The `except` block was skipped.
- The program continued normally.

````python
print("Program Started")

try:
    print("Step 1")

    result = 10 / 0

    print("Step 2")

except:
    print("Division by zero detected.")

print("Program Finished")
````

### Explanation

Execution occurred in the following order:

1. Python entered the `try` block.
2. `"Step 1"` was printed successfully.
3. `10 / 0` raised a `ZeroDivisionError`.
4. Python immediately exited the `try` block.
5. `print("Step 2")` was **never executed**.
6. Control transferred to the `except` block.
7. After executing the `except` block, the program continued normally.

This demonstrates one of the most important rules of exception handling:

> Once an exception is raised, Python immediately stops executing the remaining statements inside the current `try` block.

````python
salary = 5000

try:
  salary = int(salary)

  print("Salary converted successfully")

  annual_salary = 5000 * 12

  print("Annual Salary", annual_salary)

except:
  print("Invalid Salary")
````

````python
salary = "Fifty Thousand"

try:
    salary = int(salary)

    print("Salary converted successfully.")

    annual_salary = salary * 12

    print("Annual Salary:", annual_salary)

except:
    print("Invalid salary.")
````

### Professional Example

Exception handling is especially useful when working with external resources.

If a configuration file is missing, we can notify the user instead of allowing the application to terminate unexpectedly.

````python
print("Application Started")

try:
    print("Reading configuration file...")

    config = open("config.json")

    print("Configuration loaded successfully.")

except:
    print("Configuration file not found.")

print("Application Finished")
````

### Key Takeaways

- Python executes statements inside the `try` block sequentially.
- If an exception occurs, execution stops immediately.
- Remaining statements inside the `try` block are skipped.
- Control transfers to the matching `except` block.
- After the `except` block completes, normal program execution resumes.

### Professional Notes

One of the most common mistakes beginners make is placing too many unrelated statements inside a single `try` block. In professional Python, the `try` block should be kept as small as practical, containing only the statements that are expected to raise an exception. This improves readability, debugging, and maintainability.

## 1.3 Catching Specific Exceptions

### Why Catch Specific Exceptions?

In the previous section, we used a **bare `except`** block:

```python
try:
    # Code that may fail
except:
    print("An error occurred.")
```

Although this works, it is **not recommended** for professional Python development because it catches **every exception**, regardless of its cause. As a result, it becomes difficult to determine what actually went wrong and can unintentionally hide programming errors.

Instead, Python allows us to catch **specific exception types**.

This enables our programs to:

- Handle only the exceptions we expect.
- Respond appropriately to different failure scenarios.
- Produce more meaningful error messages.
- Improve debugging and maintainability.

### General Syntax

```python
try:
    # Code that may raise an exception

except ExceptionType:
    # Handle the specific exception
```

For example:

```python
try:
    age = int(user_input)

except ValueError:
    print("Please enter a valid integer.")
```

Here, the `except` block executes **only** if a `ValueError` is raised. Any other exception will not be caught by this block.

> **Note:** In this section, we use different exception types to learn **how to catch them**. Each exception will be studied comprehensively later in **Section 3 – Essential Built-in Exceptions**.

---

### Why Is This Better Than a Bare `except`?

Suppose your code can raise a `ValueError`, but it also contains a programming mistake that raises a `NameError`.

Using a bare `except`:

- Both exceptions are caught.
- The real programming mistake may remain hidden.
- Debugging becomes much more difficult.

Using a specific exception:

- Only the expected exception is handled.
- Unexpected exceptions are still reported by Python.
- Bugs are easier to identify and fix.

This is one of the most important professional practices in exception handling.

````python
try:
    age = int("twenty")

except ValueError:
    print("Please enter a valid numeric age.")
````

### Explanation

The statement `int("twenty")` raises a `ValueError`.

Since the `except` block is specifically looking for a `ValueError`, Python executes that block and handles the exception gracefully.

We will study `ValueError` in detail later.

````python
try:
    result = "100" + 50

except TypeError:
    print("Cannot combine values of incompatible types.")
````

### Explanation

Python raises a `TypeError` because the `+` operator cannot add a `str` and an `int`.

Since the exception matches the specified type, the `except TypeError` block executes.

try:
    file = open("sales_report.csv")

except FileNotFoundError:
    print("The requested file was not found.")

````python
try:
    file = open("sales_report.csv")

except FileNotFoundError:
    print("The requested file was not found.")
````

### Explanation

If the file does not exist, Python raises a `FileNotFoundError`.

Instead of terminating the program, the exception is handled and an informative message is displayed.

````python
student = {
    "name": "Ali",
    "age": 21
}

try:
    print(student["grade"])

except KeyError:
    print("The requested key does not exist.")
````

### Explanation

The dictionary does not contain the key `"grade"`.

Python raises a `KeyError`, which is then caught by the corresponding `except` block.

````python
numbers = [10, 20, 30]

try:
    print(numbers[5])

except IndexError:
    print("The requested index is outside the list.")
````

### Explanation

The list contains only three elements.

Attempting to access index `5` raises an `IndexError`, which is handled by the matching `except` block.

````python
try:
    result = 100 / 0

except ZeroDivisionError:
    print("Division by zero is not allowed.")
````

### Explanation

Python cannot divide a number by zero.

It raises a `ZeroDivisionError`, which is handled by the corresponding `except` block.

### Comparison: Bare `except` vs Specific `except`

Consider the following examples.

Using a bare `except`:

```python
try:
    age = int("twenty")

except:
    print("Something went wrong.")
```

Using a specific exception:

```python
try:
    age = int("twenty")

except ValueError:
    print("Please enter a valid numeric age.")
```

The second version is preferred because it clearly communicates:

- which exception is expected,
- why it occurred,
- and how it is being handled.

This makes the code easier to read, debug, and maintain.

### Best Practices

- Catch the **most specific exception** possible.
- Avoid using a bare `except` in production code.
- Display clear and meaningful error messages.
- Handle only the exceptions your program is prepared to recover from.
- Allow unexpected exceptions to propagate so they can be identified and fixed during development.

### Professional Notes

Professional Python applications rarely use a bare `except` because it can unintentionally hide programming mistakes and make debugging significantly more difficult.

Specific exception handling makes code more reliable, easier to maintain, and easier for other developers to understand. It is the standard approach used in production software, data processing pipelines, web applications, and automation scripts.

## 1.4 Multiple `except` Clauses

### Why Do We Need Multiple `except` Clauses?

A single block of code may raise **different types of exceptions** depending on the input or the situation.

Instead of handling every error in the same way, Python allows us to define **multiple `except` clauses**, each responsible for handling a specific exception type.

This makes the program:

- More readable.
- Easier to debug.
- More informative.
- Easier to maintain.

### Syntax

```python
try:
    # Code that may raise different exceptions

except FirstException:
    # Handle first exception

except SecondException:
    # Handle second exception

except ThirdException:
    # Handle third exception
```

Python checks each `except` block **from top to bottom**.

As soon as it finds a matching exception type:

- that `except` block executes,
- all remaining `except` blocks are skipped.

> Only **one** `except` block executes for a single exception.

### Best Practice

Arrange exception handlers from **most specific** to **most general**.

````python
try:
  value = int(input("Enter a number: "))
  result = 100/value

except TypeError:
  print("Integer must be Entered")

except ValueError:
  print("Please enter a valid integer.")

except ZeroDivisionError:
  print("Division by zero is not allowed")
````

### Explanation

Notice that the same `try` block can raise different exceptions.

- Invalid input raises `ValueError`.
- Entering `0` raises `ZeroDivisionError`.

Python automatically executes the matching `except` block.

## 1.5 Catching Multiple Exception Types

Sometimes multiple exception types require **exactly the same handling**.

Instead of writing multiple `except` blocks with identical code, Python allows multiple exception types to be grouped together.

### Syntax

```python
try:
    ...

except (ExceptionType1, ExceptionType2):
    ...
```

Notice the parentheses.

The exception types are written as a tuple.

If **any** one of them occurs, the same handler executes.

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except (ValueError, ZeroDivisionError):
    print("Invalid input.")
````

````python
student = {
    "name": "Ali"
}

numbers = [10, 20, 30]

try:
    choice = input("Enter key or index: ")

    if choice == "key":
        print(student["grade"])

    else:
        print(numbers[5])

except (KeyError, IndexError):
    print("Requested data does not exist.")
````

### When Should You Use This?

Use this approach only when all exceptions should be handled in the **same way**.

If different exceptions require different actions, use separate `except` blocks instead.

## 1.6 Using `except Exception as e`

In many situations, you may want to know **what actually went wrong** instead of displaying a generic message.

Python allows you to capture the exception object using the `as` keyword.

### Syntax

```python
try:
    ...

except Exception as e:
    ...
```

Here,

- `Exception` catches almost all application-level exceptions.
- `e` is a variable that stores the exception object.

The variable name `e` is a convention.

You may choose another name if desired.

````python
try:
  number = int("Python")

except Exception as e:
  print(e)
````

# 2. The `else` and `finally` Clauses

In the previous section, we learned how the `try` and `except` statements allow us to handle runtime exceptions gracefully.

Python provides two additional optional clauses that make exception handling more structured and expressive:

- `else`
- `finally`

These clauses allow us to clearly separate:

- code that should run **only when no exception occurs**, and
- code that should **always execute**, regardless of whether an exception occurs.

Using `else` and `finally` appropriately improves code readability, simplifies resource management, and follows professional Python practices.

---

## 2.1 The `else` Clause

### What is the `else` Clause?

The `else` clause executes **only if the `try` block completes successfully without raising any exception**.

If an exception occurs, the `else` block is skipped entirely.

The `else` clause is useful for separating code that depends on the successful execution of the `try` block from the code that may raise exceptions.

### Syntax

```python
try:
    # Code that may raise an exception

except ExceptionType:
    # Handle the exception

else:
    # Executes only if no exception occurs
```

### Execution Rule

- If the `try` block succeeds → `else` executes.
- If an exception occurs → `except` executes and `else` is skipped.

Using `else` makes your code cleaner because it keeps only the potentially risky statements inside the `try` block.

````python
try:
  value = int(input("Enter a number: "))
  result = 100/value

except ZeroDivisionError:
  print("Don't divide by Zero")

else:
  print(result)
````

````python
try:
    number = int("25")

except ValueError:
    print("Invalid number.")

else:
    print("Conversion successful.")
````

````python
try:
    number = int("Python")

except ValueError:
    print("Invalid number.")

else:
    print("Conversion successful.")
````

### Explanation

In the first example:

- `int("25")` succeeds.
- No exception occurs.
- The `else` block executes.

In the second example:

- `int("Python")` raises a `ValueError`.
- The `except` block executes.
- The `else` block is skipped.

---

````python
try:
    file = open("employees.csv")

except FileNotFoundError:
    print("File not found.")

else:
    print("File opened successfully.")
````

### Why Use `else`?

Consider the following code.

```python
try:
    number = int(user_input)
    print("Conversion successful.")
except ValueError:
    print("Invalid input.")
```

The statement

```python
print("Conversion successful.")
```

cannot raise a `ValueError`.

Therefore, it doesn't belong inside the `try` block.

A cleaner approach is:

```python
try:
    number = int(user_input)

except ValueError:
    print("Invalid input.")

else:
    print("Conversion successful.")
```

Keeping only the statements that may fail inside the `try` block is considered a professional coding practice.

## 2.2 The `finally` Clause

### What is the `finally` Clause?

The `finally` clause contains code that **always executes**, regardless of whether an exception occurs.

It runs:

- if no exception occurs,
- if an exception occurs and is handled,
- even if a `return` statement is executed inside the `try` block.

Its primary purpose is **resource cleanup**.

### Syntax

```python
try:
    ...

except:
    ...

finally:
    ...
```

### Execution Rule

The `finally` block always executes before leaving the `try` statement.

````python
try:
    print("Inside try.")

except:
    print("Inside except.")

finally:
    print("Inside finally.")
````

````python
try:
    result = 10 / 0

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program completed.")
````

### Explanation

Even though an exception occurred,

the `finally` block still executed.

This is the guarantee provided by the `finally` clause.

It is therefore the ideal place for cleanup operations.

## 2.3 Resource Cleanup

Many programs acquire resources while executing.

Examples include:

- Opening files
- Connecting to databases
- Opening network connections
- Creating sockets
- Acquiring locks

These resources should always be released after use.

Failing to release them may lead to:

- memory leaks,
- locked files,
- exhausted database connections,
- reduced application performance.

The `finally` clause ensures that cleanup code always executes.

````python
file = None

try:
    file = open("employees.csv")
    print("Reading file...")

except FileNotFoundError:
    print("File not found.")

finally:
    if file is not None:
        file.close()

    print("Cleanup completed.")
````

### Explanation

If the file opens successfully:

- it is closed inside the `finally` block.

If opening the file fails:

- the `finally` block still executes.

This guarantees that cleanup code is always performed.

## 3.1 `ValueError`

### What is a `ValueError`?

A `ValueError` is raised when a function or operation receives an **argument of the correct data type but an invalid value**.

In other words, **the type is acceptable, but the value itself is not**.

For example:

- Passing a string that cannot be converted to an integer.
- Requesting the square root of a negative number (for functions that only accept non-negative values).
- Searching for an element that does not exist in a list using certain methods.
- Parsing an incorrectly formatted date.

`ValueError` is one of the most common exceptions encountered in professional Python development because many programs rely on external input such as user data, files, APIs, and datasets, where invalid values are common.

---

### When is a `ValueError` Raised?

Python raises a `ValueError` when:

- The data type is valid.
- The operation is supported.
- The supplied value is outside the acceptable range or format.

For example,

```python
int("25")
```

works because `"25"` is a valid numeric string.

However,

```python
int("Python")
```

raises a `ValueError` because `"Python"` cannot be interpreted as an integer.

Notice that the argument is still a `str`, which is an acceptable input type for `int()`. The problem is the **content** of the string, not its type.

---

### Understanding the Difference Between Type and Value

Consider the following examples.

| Expression | Result | Reason |
|------------|--------|--------|
| `int("100")` | ✅ Works | Valid string representation of an integer |
| `int("abc")` | ❌ `ValueError` | Correct type (`str`), invalid value |
| `int(100.5)` | ✅ Works | Valid numeric value |
| `int(None)` | ❌ `TypeError` | Invalid argument type |

This distinction is very important because beginners often confuse `ValueError` with `TypeError`.

---

### Common Situations That Raise `ValueError`

Some of the most common situations include:

- Converting invalid strings to numbers.
- Invalid user input.
- Incorrect date or time formats.
- Searching for values that do not exist.
- Mathematical operations requiring specific value ranges.
- Data cleaning and preprocessing.

We will explore several practical examples below.

````python
number = int("150")

print(number)
````

In this example, the string contains a valid integer representation, so the conversion succeeds without raising an exception.

````python
try:
    number = int("Python")

except ValueError:
    print("Invalid integer value.")
````

Although `"Python"` is a string, it cannot be converted into an integer.

Therefore, Python raises a `ValueError`.

````python
try:
    price = float("Twenty Five")

except ValueError:
    print("Invalid floating-point value.")
````

The `float()` function accepts strings, but only if they represent valid floating-point numbers.

Since `"Twenty Five"` is not a numeric value, a `ValueError` is raised.

````python
fruits = ["Apple", "Banana", "Orange"]

try:
    position = fruits.index("Mango")

except ValueError:
    print("The value was not found in the list.")
````

The `list.index()` method searches for a specific value.

## 3.2 `TypeError`

### What is a `TypeError`?

A `TypeError` is raised when an operation or function is applied to an **object of an inappropriate data type**.

In other words, **the value itself may be perfectly valid, but its data type is not compatible with the operation being performed.**

Unlike a `ValueError`, where the type is correct but the value is invalid, a `TypeError` indicates that Python cannot perform the requested operation because the involved data types are incompatible.

`TypeError` is one of the most common exceptions in Python and is frequently encountered while working with functions, operators, collections, object-oriented programming, and third-party libraries such as NumPy and Pandas.

---

### When is a `TypeError` Raised?

Python raises a `TypeError` when:

- An operation is applied to an unsupported data type.
- Incompatible data types are used together.
- A function receives an argument of the wrong type.
- A function is called with an incorrect number of arguments.
- A non-callable object is called like a function.
- An object that does not support a particular operation is used with that operation.

---

### Understanding the Difference Between `ValueError` and `TypeError`

These two exceptions are commonly confused.

| Expression | Exception | Reason |
|------------|-----------|--------|
| `int("abc")` | `ValueError` | `str` is an acceptable type, but the value cannot be converted to an integer. |
| `int(None)` | `TypeError` | `None` is not an acceptable type for `int()`. |
| `"100" + 50` | `TypeError` | Cannot add a `str` and an `int`. |
| `int("100")` | No Exception | Correct type and valid value. |

A simple rule to remember is:

- **Correct type, wrong value → `ValueError`**
- **Wrong type → `TypeError`**

---

### Common Situations That Raise `TypeError`

Some common situations include:

- Combining incompatible data types.
- Passing the wrong argument type to a function.
- Calling objects that are not callable.
- Passing an incorrect number of arguments to a function.
- Performing unsupported operations on objects.

These situations occur frequently in both beginner and professional Python programs.

## 3.3 `ZeroDivisionError`

### What is a `ZeroDivisionError`?

A `ZeroDivisionError` is raised when a program attempts to **divide a number by zero**.

In mathematics, division by zero is undefined. Python follows this mathematical rule and prevents such operations by raising a `ZeroDivisionError` instead of producing an invalid result.

This exception commonly occurs in calculations involving:

- User input
- Mathematical formulas
- Financial calculations
- Statistical analysis
- Scientific computing
- Data Analysis and Machine Learning pipelines

Whenever there is a possibility that a divisor (the denominator) may become zero, professional Python code should anticipate and handle this exception appropriately.

---

### When is a `ZeroDivisionError` Raised?

Python raises a `ZeroDivisionError` whenever the divisor in a division or modulo operation evaluates to zero.

Common operations include:

- Division (`/`)
- Floor Division (`//`)
- Modulo (`%`)

Examples:

```python
10 / 0
```

```python
25 // 0
```

```python
15 % 0
```

All of these operations raise a `ZeroDivisionError`.

---

### Why Does Python Raise This Exception?

Allowing division by zero would produce mathematically meaningless results.

Instead of returning an incorrect value, Python immediately stops the operation and raises an exception to prevent invalid computations from propagating through the program.

This behavior helps maintain the correctness and reliability of applications.

````python
try:
    result = 100 / 0

except ZeroDivisionError:
    print("Division by zero is not allowed.")
````

### Explanation

The divisor is `0`.

Since division by zero is mathematically undefined, Python raises a `ZeroDivisionError`, which is then handled by the corresponding `except` block.

````python
try:
    result = 25 // 0
    print(result)

except ZeroDivisionError:
    print("Cannot perform floor division by zero.")
````

### Explanation

The floor division operator (`//`) also requires a non-zero divisor.

Attempting to divide by zero raises the same exception.

````python
try:
    remainder = 15 % 0

except ZeroDivisionError:
    print("Cannot calculate the remainder when dividing by zero.")
````

### Explanation

The modulo operator (`%`) computes the remainder after division.

Since division by zero is impossible, the modulo operation also raises a `ZeroDivisionError`.

### A Common Real-World Scenario

One of the most common causes of a `ZeroDivisionError` is **user input**.

Users may unintentionally enter `0` as a divisor, causing the calculation to fail if the program does not validate the input.

````python
try:
  divisor = int(input("Enter a divisor: "))
  result = 100/divisor

except ZeroDivisionError:
  print("The divisor cannot be zero")

else:
  print(result)
````

### Explanation

This example demonstrates how multiple exception types can occur in the same program.

- Invalid numeric input raises a `ValueError`.
- Entering `0` raises a `ZeroDivisionError`.

Each exception is handled independently using its own `except` block.

## 3.5 `IndexError`

### What is an `IndexError`?

An `IndexError` is raised when a program attempts to access an **index that is outside the valid range** of a sequence.

- `tuple`
- `str`

If you attempt to access an index that does not exist, Python raises an `IndexError`.

An `IndexError` indicates:

- The sequence exists.
- The requested index does **not** exist.

This exception is very common when processing datasets, looping through collections, and working with lists returned by APIs or databases.

---

### When is an `IndexError` Raised?

Python raises an `IndexError` whenever you try to access an index that is outside the valid range of a sequence.

```text
0
1
2
```

Since index `5` does not exist, Python raises an `IndexError`.

---

### Why Does Python Raise This Exception?

Every sequence has a fixed number of elements.

When you request an index, Python first checks whether that position exists.

If it does:

- the corresponding element is returned.

Otherwise,

Python raises an `IndexError` because there is no element at that position.

Returning an arbitrary value could lead to incorrect program behavior and make bugs difficult to detect.

---

## 3.6 `AttributeError`

### What is an `AttributeError`?

An `AttributeError` is raised when a program attempts to **access or call an attribute or method that an object does not have**.

In Python, almost everything is an **object**, and every object has its own set of **attributes** and **methods**.

If you try to access an attribute or call a method that does not exist for that object, Python raises an `AttributeError`.

An `AttributeError` indicates that:

- The object itself exists.
- The requested attribute or method does **not** exist for that object.

This exception is very common in professional Python development, particularly when working with:

- Object-Oriented Programming (OOP)
- Third-party libraries
- Pandas
- NumPy
- Custom classes
- API objects

---

### What are Attributes and Methods?

Before understanding `AttributeError`, it is important to distinguish between **attributes** and **methods**.

An **attribute** is a variable that belongs to an object.

A **method** is a function that belongs to an object.

For example,

```python
name = "Python"

name.upper()
```

- `name` is a string object.
- `upper()` is one of the methods provided by the `str` class.

Since `upper()` exists for strings, the operation succeeds.

However,

```python

```

raises an `AttributeError` because strings do not have an `append()` method.

---

### When is an `AttributeError` Raised?

Python raises an `AttributeError` whenever:

- An attribute does not exist.
- A method does not exist.
- The object type does not support the requested attribute or method.

The `str` object provides the `upper()` method.

````python
try:
    name = "Python"

    name.append("!")

except AttributeError:
    print("String objects do not have an 'append()' method.")
````

This demonstrates that methods belong to specific object types.

A `PermissionError` is raised when a program attempts to access a file, directory, or other operating system resource **without having the necessary permissions**.

- Deleting a protected file.
- Accessing files owned by another user.
- Attempting operations that require administrator or root privileges.

---

### Why Does Python Raise This Exception?

Modern operating systems protect files and directories using permission systems.

Before allowing an operation such as reading, writing, deleting, or modifying a file, the operating system checks whether the current process has the required permissions.

If permission is denied, Python raises a `PermissionError`.

This protects important files from unauthorized access or accidental modification.

## 3.11 `StopIteration`

### What is `StopIteration`?

`StopIteration` is a built-in exception that signals that an **iterator has no more items to produce**.

Unlike exceptions such as `ValueError` or `TypeError`, `StopIteration` is **not used to report an error**.

Instead, it tells Python:

> "The iteration has finished. There are no more values available."

Python uses this exception internally whenever an iterator becomes exhausted.

---

### Before Understanding `StopIteration`

To understand this exception, you only need to know one idea.

An **iterator** is an object that returns one value at a time.

Each time Python requests the next value, the iterator either:

- returns the next item, or
- raises `StopIteration` if no items remain.

We will study iterators in detail later in this Python Mastery series.

---

### When is `StopIteration` Raised?

Python raises `StopIteration` whenever the next item is requested from an iterator, but the iterator has already produced all of its values.

This usually happens when calling the built-in `next()` function.

---

### Why Does Python Raise This Exception?

Imagine a list containing three elements.

Python retrieves them one by one.

Eventually, there are no elements left.

Instead of returning an invalid value,

the iterator raises `StopIteration` to indicate that iteration has completed.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

The iterator still has elements available, so each call to `next()` returns the next value.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
#print(next(iterator))
````

### Explanation

The list contains only three elements.

The first three calls to `next()` return those values.

The fourth call requests another value, but none remain.

Therefore, Python raises a `StopIteration` exception.

## How `for` Loops Handle `StopIteration`

One of the reasons most Python developers rarely encounter `StopIteration` is that `for` loops handle it automatically.

Internally, a `for` loop repeatedly calls `next()` on the iterator.

When `StopIteration` is raised, the loop catches the exception and terminates gracefully.

This process happens automatically and is completely hidden from the programmer.

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

Although no exception appears in the output, Python internally raises and handles `StopIteration` after printing the final element.

This is one of the reasons `for` loops are preferred over manually calling `next()`.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

try:
    while True:
        print(next(iterator))

except StopIteration:
    print("Iteration completed.")
````

In this example, we explicitly catch the `StopIteration` exception.

Once the iterator is exhausted, the exception is raised, caught by the `except` block, and the program continues normally.

## 4.2 Re-raising Exceptions

### What is Re-raising an Exception?

Re-raising an exception means **raising the same exception again after it has been caught**.

Instead of completely handling an exception inside an `except` block, a program may need to:

- Log the error.
- Perform cleanup operations.
- Release resources.
- Record debugging information.
- Notify the user.

After performing these tasks, the exception can be **re-raised** so that higher levels of the program are still aware that an error occurred.

In other words, the exception is **not suppressed**—it is simply allowed to continue propagating after some additional processing.

---

### Why Do We Re-raise Exceptions?

In professional software, different parts of an application have different responsibilities.

For example:

- A lower-level function may detect and log an error.
- A higher-level function may decide whether the program should recover or terminate.

Re-raising allows each part of the program to perform its own responsibilities without hiding the original error.

---

### Syntax

To re-raise the current exception, use a **bare `raise`** statement.

```python
try:
    ...
except SomeException:
    ...
    raise
```

Notice that `raise` is written **without specifying an exception type**.

This tells Python:

> "Raise the same exception that is currently being handled."

---

````python
try:
    number = int("abc")

except ValueError:
    print("Logging the error...")
    raise
````

### Explanation

The `ValueError` is raised when Python fails to convert `"abc"` into an integer.

The `except` block catches the exception and prints a log message.

The bare `raise` statement then re-raises the **same `ValueError`**, allowing it to continue propagating.

Without the final `raise`, the exception would be considered handled, and the program would continue executing.

````python
def process_age(age):
    try:
        if age < 0:
            raise ValueError("Age cannot be negative.")

    except ValueError:
        print("Invalid age detected.")
        raise


process_age(-5)
````

In this example, the function detects invalid input and records a message before re-raising the exception.

This allows another part of the program to decide how the error should ultimately be handled.

````python
def validate_salary(salary):
    try:
        if salary < 0:
            raise ValueError("Salary cannot be negative.")

    except ValueError:
        print("Validation failed.")
        raise


try:
    validate_salary(-10000)

except ValueError as error:
    print(f"Program terminated because: {error}")
````

### Program Execution Flow

This example demonstrates exception propagation across multiple levels.

1. `validate_salary()` raises a `ValueError`.
2. The function catches the exception.
3. A validation message is displayed.
4. The exception is re-raised using a bare `raise`.
5. The outer `try` block catches the same exception.
6. The program handles the error at a higher level.

This layered approach is very common in professional applications.

## Why Use a Bare `raise`?

When re-raising the current exception, always use:

```python
raise
```

instead of

```python
raise error
```

A bare `raise` preserves the original traceback, making debugging much easier.

It tells Python to continue propagating the exception exactly as it occurred.

This is the preferred and Pythonic approach.

````python
try:
    result = 10 / 0

except ZeroDivisionError:
    print("Cleaning up resources...")
    raise
````

## 4.3 Exception Chaining (Brief Introduction)

### What is Exception Chaining?

Sometimes, while handling one exception, another exception may occur.

In such situations, it is often useful to preserve information about the original exception while raising a new one.

This process is called **exception chaining**.

Exception chaining allows Python to record the relationship between exceptions, making debugging significantly easier.

Instead of seeing only the final exception, developers can also determine what originally caused the failure.

---

### Why is Exception Chaining Useful?

Large applications are often organized into multiple layers.

For example:

- File handling layer
- Data processing layer
- Business logic layer
- User interface layer

A low-level function may encounter a technical exception such as a `FileNotFoundError`.

A higher-level function may wish to present a more meaningful exception to the rest of the application.

Instead of losing the original exception, Python allows the new exception to reference it.

This provides much richer debugging information.

---

### Syntax

```python
raise NewException("message") from original_exception
```

The `from` keyword explicitly links the new exception to the original one.

---

### When Should You Use Exception Chaining?

Exception chaining is commonly used when:

- Converting low-level exceptions into application-specific exceptions.
- Creating reusable libraries.
- Building APIs.
- Developing large applications with multiple abstraction layers.

For most beginner programs, exception chaining is not required.

However, understanding the concept prepares you for professional software development.

````python
try:
    int("abc")

except ValueError as error:
    raise RuntimeError(
        "Unable to process user input."
    ) from error
````

ValueError: invalid literal for int() with base 10: 'abc'

The above exception was the direct cause of the following exception:

RuntimeError: Unable to process user input.

### Explanation

The original exception is a `ValueError`.

Instead of hiding it, the program raises a more meaningful `RuntimeError` while preserving the original cause.

The traceback now contains both exceptions, making debugging much easier.

---

### Important Note

Exception chaining is considered an advanced feature.

You do not need to use it in everyday Python scripts.

It becomes valuable when building reusable libraries, APIs, frameworks, or other large software systems where preserving debugging information is important.

## 6.1 The `with` Statement

## 6.3 Understanding `__enter__()`

### What is `__enter__()`?

The `__enter__()` method is one of the two special methods that make an object a **Context Manager**.

Whenever a `with` statement begins execution, Python automatically calls the object's `__enter__()` method.

Its primary responsibilities are to:

- Acquire or initialize the resource.
- Perform any setup required before the resource is used.
- Return the object (or another object) that will be assigned to the variable following the `as` keyword.

---

### When is `__enter__()` Called?

Consider the following statement:

```python
with ContextManager() as resource:
    ...
```

Python performs these steps:

1. Creates an instance of `ContextManager`.
2. Calls its `__enter__()` method.
3. Stores the value returned by `__enter__()` in `resource`.
4. Executes the code inside the `with` block.

---

### General Syntax

```python
class MyContextManager:

    def __enter__(self):
        ...
        return object
```

The returned object becomes available inside the `with` block.

---

### Execution Flow

```text
Create Context Manager Object
            │
            ▼
Call __enter__()
            │
            ▼
Return Resource
            │
            ▼
Assign to Variable After 'as'
            │
            ▼
Execute with Block
```

---

### What Should `__enter__()` Return?

The `__enter__()` method can return:

- `self`
- another object
- a file object
- a database connection
- a network session
- any object the programmer wants to use inside the `with` block

The returned object becomes the value of the variable after the `as` keyword.

````python
class SimpleContextManager:

    def __enter__(self):
        print("Entering the context...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting the context...")


with SimpleContextManager() as manager:
    print("Inside the with block.")
````

### Explanation

Let's examine the execution step by step.

Python executes:

```python
with SimpleContextManager() as manager:
```

Internally, Python performs approximately the following steps:

1. Create the object.

```python
manager_object = SimpleContextManager()
```

2. Call `__enter__()`.

```python
returned_object = manager_object.__enter__()
```

3. Assign the returned object.

```python
manager = returned_object
```

4. Execute the code inside the `with` block.

5. When the block finishes, Python automatically calls `__exit__()`.

Although Python performs these operations internally, understanding this sequence makes the behavior of the `with` statement much easier to understand.

````python
class ResourceManager:

    def __enter__(self):
        print("Opening resource...")
        return "Database Connection"

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing resource...")


with ResourceManager() as resource:
    print(resource)
````

### Explanation

Notice that `__enter__()` does **not** return `self`.

Instead, it returns a string.

Therefore,

```python
resource
```

contains:

```text
Database Connection
```

This demonstrates that `__enter__()` may return **any object**, depending on what should be available inside the `with` block.

## Professional Notes

Many built-in Python objects implement `__enter__()` internally.

For example, when opening a file with:

```python

```

the `open()` function returns a file object whose `__enter__()` method prepares the file and returns the file object itself.

This is why the variable `file` is immediately available inside the `with` block.

A useful way to remember `__enter__()` is:

> **Enter → Prepare → Return**

Whenever you see a `with` statement, think:

> "Python is preparing a resource before I begin using it."

This mental model makes the Context Manager Protocol much easier to understand.

## 6.4 Understanding `__exit__()`

### What is `__exit__()`?

The `__exit__()` method is the second method of the **Context Manager Protocol**.

Python automatically calls `__exit__()` whenever execution leaves a `with` block.

Unlike `__enter__()`, which prepares the resource, `__exit__()` is responsible for cleaning it up.

Typical cleanup tasks include:

- Closing files
- Closing database connections
- Releasing thread locks
- Closing network sockets
- Releasing operating system resources

One of the most important features of `__exit__()` is that it executes **whether the `with` block finishes normally or because of an exception**.

This guarantees that cleanup code always runs.

````python
class SimpleContextManager:

    def __enter__(self):
        print("Entering...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Cleaning up...")


with SimpleContextManager():
    print("Working...")
````

### The Parameters of `__exit__()`

The `__exit__()` method receives three arguments that describe any exception raised inside the `with` block.

```python
def __exit__(self, exc_type, exc_value, traceback):
```

#### `exc_type`

The class of the exception.

Example:

```text
ValueError
```

If no exception occurred, its value is `None`.

---

#### `exc_value`

The actual exception object containing the error message.

Example:

```text
ValueError("Invalid input")
```

If no exception occurred, its value is `None`.

---

#### `traceback`

Contains information about where the exception occurred.

This information is primarily useful for debugging and logging.

If no exception occurred, its value is `None`.

````python
class DemoContextManager:

    def __enter__(self):
        print("Opening resource...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print(f"Exception Type : {exc_type}")
        print(f"Exception Value: {exc_value}")
        print(f"Traceback      : {traceback}")
        print("Closing resource...")


with DemoContextManager():
    number = 10 / 0
````

### Why Did the Exception Still Appear?

Although `__exit__()` executed successfully, it **did not suppress the exception**.

By default, after `__exit__()` finishes, Python re-raises the exception, and normal exception propagation continues.

This behavior ensures that resources are cleaned up **without hiding errors**.

````python
class DemoContextManager:

    def __enter__(self):
        print("Entering...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Cleaning up...")
        return True


with DemoContextManager():
    print(10 / 0)

print("Program continues...")
````

### Returning `True` vs `False`

The return value of `__exit__()` determines whether an exception should continue propagating.

- `return False` (or no return statement)
  - The exception continues propagating.
  - This is the default and most common behavior.

- `return True`
  - Python considers the exception handled.
  - The exception is suppressed, and execution continues after the `with` block.

In professional software, returning `True` should be done only when the exception has been fully handled and suppressing it is the intended behavior.

### Program Execution Flow

```text

        │
        ▼
Call __enter__()
        │
        ▼
Execute Code
        │
        ▼
Exception Occurred?
      │         │
     No        Yes
      │         │
      └────┬────┘
           ▼
Call __exit__()
           │
           ▼
Return True?
      │         │
     Yes       No
      │         │
Continue   Re-raise Exception
```

### Steps to Create a Custom Context Manager

1. Create a class.
2. Implement the `__enter__()` method.
3. Implement the `__exit__()` method.
4. Use the class with the `with` statement.

---

### General Syntax

```python
class MyContextManager:

    def __enter__(self):
        # Acquire or prepare the resource
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        # Clean up the resource
        pass


with MyContextManager() as resource:
    # Use the resource
```

---

### Example

In this example, the context manager simulates opening and closing a database connection.

````python
class DatabaseConnection:

    def __enter__(self):
        print("Connecting to database...")
        return self

    def query(self):
        print("Executing SQL query...")

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing database connection...")


with DatabaseConnection() as db:
    db.query()
````

### Example with an Exception

Even if an exception occurs, `__exit__()` still executes, ensuring proper cleanup.

````python
class DatabaseConnection:

    def __enter__(self):
        print("Connecting...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Disconnecting...")


try:
    with DatabaseConnection():
        print("Running query...")
        raise ValueError("Database error")

except ValueError as error:
    print(error)
````

### Key Points

- `__enter__()` prepares the resource.
- `__exit__()` performs cleanup.
- Cleanup occurs whether or not an exception is raised.
- The `with` statement automatically calls these methods.

## 6.6 The `contextlib.contextmanager` Decorator

Writing a class for every context manager can become repetitive.

Python's built-in `contextlib` module provides the `@contextmanager` decorator, which allows you to create a context manager using a generator function instead of a class.

This approach is shorter and often easier to read for simple resource management tasks.

---

### General Syntax

```python
from contextlib import contextmanager

@contextmanager
def my_context():
    # Setup
    yield resource
    # Cleanup
```

The code before `yield` acts like `__enter__()`.

The code after `yield` acts like `__exit__()`.

---

### Example

````python
from contextlib import contextmanager


@contextmanager
def database_connection():
    print("Connecting...")
    yield
    print("Disconnecting...")


with database_connection():
    print("Executing query...")
````

### How It Works

```text
Before yield
        ↓
Acts like __enter__()
        ↓
yield
        ↓
Execute with Block
        ↓
After yield
        ↓
Acts like __exit__()
```

---

### When Should You Use It?

Use the `@contextmanager` decorator when:

- The context manager is simple.
- A full class would be unnecessary.
- You want cleaner and more concise code.

Use a class-based context manager when:

- The logic is more complex.
- Multiple methods or attributes are required.
- The context manager represents an object with its own state.

# 7. Best Practices for Exception Handling

By this point, you have learned how Python exceptions work and how to handle them effectively.

This final section focuses on the practices followed by professional Python developers when writing production-quality applications.

These practices improve code readability, simplify debugging, and make applications more reliable and maintainable.

---

## 7.1 Choose Appropriate Exception Types

Always raise or catch the exception that best describes the problem.

Using specific exception types makes your code easier to understand and allows errors to be handled more precisely.

### Good

```python
raise ValueError("Age cannot be negative.")
```

```python
raise FileNotFoundError("Dataset not found.")
```

### Avoid

```python
raise Exception("Something went wrong.")
```

Using generic exceptions makes debugging more difficult because they provide little information about the actual problem.

---

## 7.2 Avoid Bare `except`

A **bare `except`** catches **every** exception, including unexpected ones such as `KeyboardInterrupt` and `SystemExit`.

### Avoid

```python
try:
    result = 10 / 0

except:
    print("An error occurred.")
```

Instead, catch only the exceptions you expect.

### Preferred

```python
try:
    result = 10 / 0

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

If you truly need to catch most application errors, use:

```python
except Exception as error:
```

instead of a bare `except`.

---

## 7.3 Write Informative Error Messages

An error message should clearly explain what went wrong.

### Poor

```python
raise ValueError("Invalid.")
```

### Better

```python
raise ValueError("Age cannot be negative.")
```

### Even Better

```python
raise ValueError(
    "Employee age must be greater than or equal to zero."
)
```

Clear error messages reduce debugging time and make applications easier to maintain.

---

## 7.4 Catch Exceptions Only When You Can Handle Them

Do not catch an exception simply to prevent your program from crashing.

Catch an exception only if you can take meaningful action.

### Good Example

```python
try:
    file = open("sales.csv")

except FileNotFoundError:
    print("Please provide a valid dataset.")
```

Here, the program informs the user how to resolve the problem.

---

## 7.5 Let Exceptions Propagate When Appropriate

Sometimes the best decision is **not** to catch an exception.

Instead, allow it to propagate to the calling function, where it can be handled appropriately.

Example:

```python
def load_dataset(filename):
    return open(filename)
```

If the file does not exist, the exception naturally propagates to the caller.

This often results in cleaner and more reusable code.

---

## 7.6 Write Pythonic Exception Handling

Python emphasizes simplicity and readability.

Prefer:

```python
try:
    value = numbers[index]

except IndexError:
    print("Invalid index.")
```

instead of writing overly complicated exception-handling logic.

Keep `try` blocks as small as practical.

### Good

```python
try:
    result = numerator / denominator

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### Avoid

```python
try:
    # Hundreds of lines of code...
```

Smaller `try` blocks make it easier to identify which statement caused the exception.

---

## 7.7 Common Mistakes

### Catching Every Exception

```python
except Exception:
    pass
```

This silently ignores errors and makes debugging extremely difficult.

---

### Ignoring Error Messages

Always inspect or log the exception instead of hiding it.

### Better

```python
except Exception as error:
    print(error)
```

---

### Using Exceptions for Normal Program Flow

Exceptions should represent **unexpected or exceptional situations**, not ordinary program logic.

Avoid:

```python
try:
    number = numbers.pop()

except IndexError:
    print("List is empty.")
```

when a simple condition would be clearer:

```python
if numbers:
    number = numbers.pop()
else:
    print("List is empty.")
```

---

### Catching the Wrong Exception

Catch the exception you actually expect.

Avoid unnecessarily broad exception handling.

---

## 7.8 Production Best Practices

Professional Python applications typically follow these practices:

- Catch the most specific exception possible.
- Keep `try` blocks small and focused.
- Use `finally` or context managers for resource cleanup.
- Prefer `with` when working with files or other resources.
- Write clear, descriptive error messages.
- Create custom exceptions for application-specific problems.
- Allow exceptions to propagate when higher-level code should handle them.
- Never suppress exceptions unless there is a valid reason.
- Log exceptions in production applications instead of silently ignoring them.
- Treat exceptions as exceptional events, not as part of normal program logic.

---

## Summary

Professional exception handling is not about catching every error.

It is about:

- anticipating problems,
- choosing appropriate exception types,
- handling only the exceptions you can recover from,
- writing meaningful error messages,
- cleaning up resources safely,
- and allowing unexpected exceptions to propagate when appropriate.

Following these principles leads to Python code that is more reliable, easier to debug, and easier for others to maintain.

---

## Mentor's Tip

As a Data Analyst or Machine Learning Engineer, most of your exception handling will involve:

- validating user input,
- reading datasets,
- handling missing files,
- working with APIs,
- interacting with databases,
- and ensuring resources are closed properly using the `with` statement.

Focus on writing **clear, specific, and readable** exception handling rather than trying to catch every possible error. Well-written exception handling is a hallmark of professional Python code.

````python
# ❌ Bad Practice

try:
    age = int(input("Enter your age: "))
    result = 100 / age

except:
    print("Something went wrong.")
````

````python
# ✅ Good Practice

try:
    age = int(input("Enter your age: "))
    result = 100 / age

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Age cannot be zero.")

else:
    print(f"Result: {result}")

finally:
    print("Program finished.")
````
