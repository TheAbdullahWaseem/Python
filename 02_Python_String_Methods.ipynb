#1. split() — Breaking Text into Parts
The `.split()` method divides a string into multiple parts based on a separator and returns a list.

Syntax

`string.split(separator)`

If no separator is given, Python splits by whitespace.

##Example 1 — Basic Split

````python
text = "Ali,25,Data Analyst"
result = text.split(',')
print(result)
````

The string is split wherever a comma (",") appears.

Original string:

Ali,25,Data Analyst

After splitting:

['Ali', '25', 'Data Analyst']

##Example 2 — Real Dataset Processing

````python
for row in data:
  parts = row.split(',')
  print(parts)
````

Each row in the dataset is a single string.

Using `.split(",")`:
- Name becomes parts[0]
- Age becomes parts[1]
- Job becomes parts[2]

This is how raw CSV-like data is structured.

.split() is the FIRST step in converting unstructured text into structured data.

#2. .join() — Combining Lists into Strings

The `.join()` method combines elements of a list into a single string using a separator.

Syntax

`separator.join(list)`

````python
words = ["Python", "is", "powerful"]

result = " ".join(words)
print(result)
````

The list elements are joined using a space (" ").

List:

["Python", "is", "powerful"]

Result:

Python is powerful

````python
parts = ["Ali", "25", "Data Analyst"]

result = ",".join(parts)

print(result)
````

The list is converted back into a structured string.

This is useful when saving processed data back into CSV format.

.split() → breaks string into list  
.join() → converts list back into string

#3. .replace() — Replacing Text Values

The `.replace()` method replaces a part of a string with another value.

Syntax

`string.replace(old_value, new_value)`

````python
text = "I like Java"

result = text.replace('Java', 'Python')
print(result)
````

The word "Java" is replaced with "Python".

Original:

I like Java

Modified:

I like Python

````python
text = "Data-Scientist"

cleaned = text.replace("-", " ")
print(cleaned)
````

Hyphens are replaced with spaces to make text readable.

##Example 3 — Dataset Transformation

````python
data = [
    "Ali,25,Data Analyst",
    "Sara,30,Data Scientist"
]

cleaned_data = []

for row in data:
  row = row.replace("Data Analyst", "DA")
  row = row.replace('Data Scientist', 'DS')
  cleaned_data.append(row)

print(cleaned_data)
````

Each job title is standardized:

- Data Analyst → DA
- Data Scientist → DS

This is called text normalization in data processing.

These three methods are deeply connected:

- .split() → string → list  
- .join() → list → string  
- .replace() → modifies string content  

Together they form the foundation of text preprocessing pipelines.

````python
for row in data:
  print(row)
````

````python
for row in data:
  row = row.split(',')
  print(row)
  row = " ".join(row)
  print(row)
````

##Mini Professional Workflow

````python
row = "Ali,25,Data Analyst"

parts = row.split(',')

parts[2] = parts[2].replace('Data Analyst', 'DA')

clean_row = ",".join(parts)

print(clean_row)
````

This is a real-world ETL-style workflow:

1. Extract data using split()
2. Transform data using replace()
3. Load data back using join()

This pattern is used in:
- Data Engineering pipelines

#1. strip() — Remove Spaces from Both Sides

The `.strip()` method removes whitespace from BOTH the beginning and the end of a string.

````python
text = "   Python   "

cleaned  = text.strip()
print(cleaned)
````

The spaces from BOTH sides are removed.

Before:
"   Python   "

After:
"Python"

##Original String Does Not Change

````python
text = '  Data Science  '
text.strip()
print(text)
````

Strings are immutable in Python.

`.strip()` does NOT modify the original string.

It returns a NEW cleaned string.

##Cleaning Real Dataset

````python
for name in employee_data:
  print(name.strip())
````

Each name in the dataset is cleaned individually.

This is extremely common in:
- ETL pipelines
- Pandas preprocessing
- database cleaning
- machine learning preprocessing

`.strip()` ONLY removes whitespace from the edges.

It does NOT remove spaces inside the string.

##Internal Spaces Remain

````python
text = "   Machine Learning Engineer   "

print(text.strip())
````

Before:

"   Machine Learning Engineer   "

After:

"Machine Learning Engineer"

Internal spaces remain untouched.

##2 .lstrip() — Remove Spaces from Left Side

The `.lstrip()` method removes whitespace ONLY from the LEFT side of a string.

````python
text = "   Python"

cleaned = text.lstrip()
cleaned
````

Only the spaces from the LEFT side are removed.

Right-side spaces remain unchanged.

##Right Side Remains

````python
text = "   Python   "

print(text.rstrip())
````

Before:

`"   Python   "`

After using lstrip():

`"Python   "`

##Used when:
- text alignment matters
- right-side formatting must stay
- processing logs or reports
- parsing structured text files

##3. rstrip() — Remove Spaces from Right Side

The `.rstrip()` method removes whitespace ONLY from the RIGHT side of a string.

````python
text = "Python   "

cleaned = text.rstrip()

print(cleaned)
````

Used when:
- indentation must stay
- formatting is important
- processing exported reports
- handling fixed-width text files

````python
raw_data = [
    "   Ali   ",
    " Sara ",
    "Ahmed   ",
    "   Ayesha"
]

cleaned_data = []

for name in raw_data:
  name = name.strip()
  cleaned_data.append(name)

print(cleaned_data)
````

This is a realistic data cleaning pipeline.

Steps:
1. Read raw messy data
2. Remove unwanted whitespace
3. Store cleaned values

This workflow is heavily used in:
- Pandas
- SQL imports
- Machine Learning preprocessing
- Data Engineering

````python
text = "   ABDULLAH WASEEM   "

result = text.strip().title()

print(result)
````

Method chaining means applying multiple methods together.

Here:
- `.strip()` removes spaces
- `.title()` formats text professionally

This is very common in professional codebases.

#1. .find() — Locate Substring Index

The `.find()` method searches for a substring inside a string and returns the index position of its FIRST occurrence.

If the substring is NOT found, `.find()` returns `-1`.

````python
text = "Python Programming"

result = text.find('Program')
print(result)
````

Python searches for the substring `"Program"`.

The substring starts at index 7.

Index positions:

P y t h o n _ P r o g r a m
0 1 2 3 4 5 6 7

##Substring Not Found

````python
text = "Python Programming"

result = text.find("Java")

print(result)
````

The substring `"Java"` does not exist inside the string.

Therefore Python returns:
-1

Professionally, `-1` is commonly used to check whether something exists in text.

This is heavily used in:
- filtering systems
- search engines
- log analysis
- text parsing

````python
documents
````

````python
for doc in documents:
  print(doc.find('Python'))
````

# 6.3 Additional Searching Methods

Python provides several methods for locating substrings within a string. While these methods perform similar tasks, they differ in how they search and how they behave when the target substring is not found.

## 6.3.1 `rfind()`

The `rfind()` method searches for the **last occurrence** of a substring and returns its index.

string.rfind(substring)

````python
text = "banana"

print(text.rfind("a"))
print(text.rfind("na"))
````

````python
text = "Python"

print(text.rfind("Java"))
````

The `rindex()` method searches for the **last occurrence** of a substring.

# 6.4 Additional Cleaning Methods

Python provides additional string methods for splitting text in different ways. These methods are particularly useful when working with structured data, where only the first or last occurrence of a separator is important.

## 6.4.1 `rsplit()`

The `rsplit()` method works like `split()`, but it begins splitting from the **right side** of the string.

By default, it splits on all occurrences of the separator. It becomes especially useful when the `maxsplit` parameter is specified.

**Syntax**

```python
string.rsplit(separator, maxsplit)
```

````python
text = "apple,banana,orange"

print(text.rsplit(","))
````

````python
text = "apple,banana,orange,mango"

print(text.rsplit(",", 1))
````

````python
text = "one-two-three-four"

print(text.rsplit("-", 2))
````

### `split()` vs `rsplit()`

| Method | Direction |
|---------|-----------|
| `split()` | Left → Right |
| `rsplit()` | Right → Left |

Without `maxsplit`, both methods usually produce the same result.

The difference becomes important when limiting the number of splits.

## 6.4.2 `partition()`

The `partition()` method splits a string **only at the first occurrence** of a separator.

It always returns a tuple containing **three elements**:

1. Text before the separator.
2. The separator itself.
3. Text after the separator.

**Syntax**

```python
string.partition(separator)
```

````python
text = "Python-Programming"

print(text.partition("-"))
````

````python
email = "student@gmail.com"

print(email.partition("@"))
````

## 6.4.3 `rpartition()`

The `rpartition()` method behaves like `partition()`, but it splits at the **last occurrence** of the separator.

It also returns a tuple of three elements.

**Syntax**

```python
string.rpartition(separator)
```

````python
text = "2026-07-04"

print(text.rpartition("-"))
````

````python
path = "C:/Users/Abdullah/Documents/file.txt"

print(path.rpartition("/"))
````

###Separator Not Found

````python
text = "Python"

print(text.rpartition("@"))
````

### `partition()` vs `split()`

| Feature | `partition()` | `split()` |
|---------|---------------|-----------|
| Number of splits | Always one | Can perform multiple splits |
| Return type | Tuple | List |
| Includes separator | Yes | No |
| Best for | Extracting two parts around a separator | Breaking text into many parts |

## Professional Notes

- Use `rsplit()` when splitting should begin from the right, especially with `maxsplit`.
- Use `partition()` to separate text at the first occurrence of a delimiter.
- Use `rpartition()` to separate text at the last occurrence of a delimiter.
- Both `partition()` and `rpartition()` return a tuple containing the text before the separator, the separator itself, and the text after the separator.
- These methods are commonly used when processing file paths, URLs, email addresses, log entries, and other structured strings.

# 6.6 String Encoding

Computers do not understand text directly.

They only understand **binary data** (0s and 1s).

When we write text such as:

```text
Python
```

Python must convert that text into bytes before it can be stored in a file, sent over a network, or transmitted between programs.

This conversion process is called **encoding**.

Converting bytes back into readable text is called **decoding**.

## 6.6.1 `str` vs `bytes`

Python distinguishes between:

- `str`
- `bytes`

A **`str`** object represents human-readable text.

A **`bytes`** object represents raw binary data.

| Type | Represents |
|------|------------|
| `str` | Human-readable text |
| `bytes` | Binary data stored by computers |

````python
text = "Python"

print(text)
print(type(text))
````

````python
data = b"Python"

print(data)
print(type(data))
````

## 6.6.2 The `encode()` Method

The `encode()` method converts a string (`str`) into bytes.

**Syntax**

```python
string.encode(encoding)
```

If no encoding is specified, Python uses **UTF-8** by default.

````python
text = "Python"

encoded = text.encode()

print(encoded)
print(type(encoded))
````

````python
text = "Hello"

encoded = text.encode("utf-8")

print(encoded)
````

## 6.6.3 The `decode()` Method

The `decode()` method converts bytes back into a string.

**Syntax**

```python
bytes_object.decode(encoding)
```

The encoding used for decoding should match the encoding used during encoding.

````python
data = b"Python"
print(type(data))

decoded = data.decode()
print(type(decoded))
````

````python
text = "Machine Learning"

encoded = text.encode()

decoded = encoded.decode()

print(encoded)
print(decoded)
````

## 6.6.4 UTF-8

**UTF-8** is the most widely used text encoding standard in the world.

It can represent characters from virtually every writing system, including:

- English
- Arabic
- Urdu
- Chinese
- Japanese
- Korean
- Emoji

Because of its flexibility and compatibility, UTF-8 is the default encoding used by Python.

````python
text = "پاکستان"

encoded = text.encode("utf-8")

print(encoded)
````

````python
decoded = encoded.decode()
print(decoded)
````

## Why Do Encoding Errors Occur?

Encoding errors usually occur when text is encoded using one character encoding but decoded using another.

For example:

- Encoded using UTF-8
- Decoded using ASCII

Because the two encodings interpret bytes differently, Python cannot correctly reconstruct the original text.

````python
text = "Python"

encoded = text.encode("utf-8")

print(encoded.decode("utf-8"))
````

## Where Will You Use Encoding?

Although you may not use `encode()` and `decode()` every day, encoding is important when working with:

- Text files
- CSV files
- JSON files
- APIs
- Web scraping
- Network communication
- Databases

Many file-reading functions allow you to specify an encoding, for example:

```python
open("data.txt", encoding="utf-8")
```

Using the correct encoding helps prevent character corruption and decoding errors.

## Professional Notes

- `str` represents human-readable text.
- `bytes` represents binary data.
- Use `encode()` to convert a `str` into `bytes`.
- Use `decode()` to convert `bytes` back into a `str`.
- UTF-8 is Python's default and most widely used text encoding.
- Matching the encoding during both encoding and decoding is essential to avoid errors.
