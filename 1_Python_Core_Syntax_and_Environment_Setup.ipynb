# 1. Variables and Data Types

Variables and data types are the foundation of Python programming. Every Python program stores, manipulates, and processes data using variables and different data types.

Python is a dynamically typed language, meaning variables are not restricted to a single data type. Instead, a variable can refer to objects of different types during program execution.

A solid understanding of variables and data types is essential before learning control flow, functions, object-oriented programming, Data Analysis, or Machine Learning.

---

# 1.1 Variables

## What is a Variable?

A variable is a **name (identifier)** that refers to an object stored in memory.

Instead of repeatedly using literal values throughout a program, we assign them to meaningful names. This makes code easier to read, maintain, and modify.

Unlike many programming languages, Python does not require variables to be declared before use. A variable is created automatically when a value is assigned to it.

---

## Syntax

```python
variable_name = value
```

The assignment operator (`=`) assigns the object on the right-hand side to the variable on the left-hand side.

General form:

```python
name = "Abdullah"
age = 22
pi = 3.14159
```

---

## How Variables Work

When Python executes an assignment statement:

1. The object on the right-hand side is created (if it does not already exist).
2. The variable name is created.
3. The variable becomes a reference to that object.

For this reason, variables in Python do **not** store values directly. They store references to objects.

Understanding this concept will become important later when studying mutable and immutable objects, object identity, and memory management.

---

## Why Do We Use Variables?

Variables allow us to:

- Store information for later use.
- Reuse values without rewriting them.
- Write dynamic programs.
- Improve code readability.
- Simplify program maintenance.

Nearly every Python program uses variables, making them one of the most fundamental concepts in the language.

---

## Variable Naming Rules

Variable names (identifiers) must follow these rules:

- Must begin with a letter (`A-Z` or `a-z`) or an underscore (`_`).
- Remaining characters may contain letters, digits, and underscores.
- Cannot begin with a digit.
- Cannot contain spaces.
- Cannot contain special characters such as `@`, `#`, `%`, `$`, or `!`.
- Cannot be a Python keyword.
- Variable names are case-sensitive.


````python
name = "Abdullah"
age = 22
country = "Pakistan"

print(name)
print(age)
print(country)
````

## Valid Variable Names

The following are valid variable names because they follow Python's identifier rules.


````python
student_name = "Ali"

_marks = 95

price2026 = 199.99

total_sales = 150000

employee_id = 1001
````

## Invalid Variable Names

The following examples are invalid because they violate Python's naming rules.


````python
# Invalid examples (do not execute)
'''
2age = 20          # Starts with a digit

student-name = ""  # Hyphen is not allowed

first name = ""    # Spaces are not allowed

class = "Python"   # 'class' is a reserved keyword'''
````

## Variable Reassignment

A variable can refer to a different object at any time.

Reassigning a variable does not modify the original object. Instead, the variable is updated to reference a new object.

This behavior is possible because Python is a dynamically typed language.


````python
score = 75

print(score)

score = 90

print(score)
````

## Multiple Variable Assignment

Python allows assigning multiple values in a single statement.


````python
name, age, country = "Abdullah", 22, "Pakistan"

print(name)
print(age)
print(country)
````

The same object can be assigned to multiple variables in a single statement.


````python
x = y = z = 100

print(x)
print(y)
print(z)
````

# 1.2 Numeric Data Types

Numeric data types are used to store numerical values and perform mathematical operations.

Python provides four built-in numeric data types:

- `int` (Integer)
- `float` (Floating-Point Number)
- `complex` (Complex Number)
- `bool` (Boolean)

Although `bool` is primarily used to represent logical values (`True` and `False`), it is implemented as a subclass of the integer type in Python and therefore belongs to Python's numeric data types.

Choosing the appropriate numeric type is important for writing correct, efficient, and readable programs.

## 1.2.1 Integer (`int`)

An integer is a whole number without a decimal point.

Integers can be:

- Positive
- Negative
- Zero

Unlike many programming languages, Python integers have arbitrary precision. This means they are not limited to 32-bit or 64-bit storage and can grow as large as the available system memory allows.

### Syntax

```python
variable_name = integer_value
```

````python
age = 22

temperature = -10

students = 45

population = 240000000

print(age)
print(temperature)
print(students)
print(population)
````

Integers are commonly used to represent values that do not require fractional parts.

Examples include:

- Age
- Number of students
- Quantity of products
- Population
- Number of records in a dataset

## 1.2.2 Floating-Point Number (`float`)

A float represents a real number that contains a decimal point.

Floating-point numbers are used whenever precision beyond whole numbers is required.

Examples include measurements, percentages, probabilities, and financial calculations.

````python
price = 199.99

height = 5.9

accuracy = 98.75

pi = 3.141592653589793

print(price)
print(height)
print(accuracy)
print(pi)
````

Floating-point numbers are stored internally using the IEEE 754 floating-point standard.

Because of this representation, some decimal values cannot be represented exactly in binary.

As a result, floating-point arithmetic may occasionally produce results that appear slightly inaccurate.

This is a normal characteristic of floating-point computation and is common across most programming languages.

````python
print(0.1 + 0.2)

print(0.1 + 0.2 == 0.3)
````

The output may be surprising because floating-point numbers are approximations rather than exact decimal values.

Later, when working with scientific computing and Data Analysis libraries, we will learn techniques for handling floating-point precision appropriately.

## 1.2.3 Complex Number (`complex`)

A complex number consists of two parts:

- Real part
- Imaginary part

Python represents imaginary numbers using the suffix `j`.

General form:

```python
a + bj
```

where:

- `a` is the real part.
- `b` is the imaginary part.

````python
z1 = 3 + 4j

z2 = -2 + 7j

print(z1)

print(z2)
````

The real and imaginary components of a complex number can be accessed using the `.real` and `.imag` attributes.

````python
number = 5 + 8j

print(number.real)

print(number.imag)
````

Complex numbers are commonly used in fields such as:

- Electrical Engineering
- Signal Processing
- Physics
- Scientific Computing

They are less common in traditional Data Analysis but are fully supported by Python.

## 1.2.4 Boolean (`bool`)

A Boolean represents one of two logical values:

- `True`
- `False`

Boolean values are primarily used in comparisons, conditional statements, loops, filtering operations, and logical expressions.

Although `bool` is its own data type, Python internally treats:

- `True` as `1`
- `False` as `0`

This is why Boolean values can participate in arithmetic operations.

````python
is_student = True

is_logged_in = False

print(is_student)

print(is_logged_in)
````

````python
print(True + True)

print(True + False)

print(False + False)
````

Since `True` behaves like `1` and `False` behaves like `0`, the previous arithmetic operations produce integer results.

This behavior is frequently used internally by Python and many data-processing libraries.

````python
passed = True

marks = 85

print(passed)

print(marks < 50)

print(marks > 50)
````

Boolean values are fundamental to programming because they control decision-making.

They are extensively used in:

- Conditional statements (`if`, `elif`, `else`)
- Loops (`while`)
- Filtering data
- Comparisons
- Logical operations
- Data validation

In Data Analysis, Boolean values are used heavily for filtering rows, creating masks, handling missing values, and applying conditions to datasets.

# 1.3 String (`str`)

A string is a sequence of Unicode characters used to represent textual data.

Strings are one of the most frequently used data types in Python and play a central role in Data Analysis, Machine Learning, Web Development, Automation, and general-purpose programming.

Examples of textual data include:

- Names
- Addresses
- Email addresses
- Product names
- File paths
- Messages
- Customer reviews
- Column names in datasets

In Python, every piece of text is represented as a string.

## Creating Strings

Strings can be created using:

- Single quotes (`' '`)
- Double quotes (`" "`)
- Triple single quotes (`''' '''`)
- Triple double quotes (`""" """`)

Single and double quotes are functionally equivalent. The choice between them depends on readability and coding style.

Triple quotes are primarily used for multi-line strings and documentation strings (docstrings).

````python
first_name = "Abdullah"

last_name = 'Waseem'

print(first_name)

print(last_name)
````

````python
message = """
Python is an amazing programming language.

It is widely used in Data Analysis,
Machine Learning,
Artificial Intelligence,
and Automation.
"""

print(message)
````

## Strings are Immutable

Strings are **immutable**, which means their contents cannot be modified after they are created.

Any operation that appears to modify a string actually creates a new string object.

Understanding immutability is important because many Python data types behave differently depending on whether they are mutable or immutable. We will study this concept in greater depth later.

````python
language = "Python"

language = "Java"

print(language)
````

In the previous example, the original string `"Python"` was not modified.

Instead, the variable `language` was updated to reference a new string object containing `"Java"`.

## Escape Characters

Escape characters allow special characters to be included inside strings.

They begin with a backslash (`\`).

````python
print("He said, \"Python is powerful.\"")

print('It\'s a beautiful day.')

print("First Line\nSecond Line")

print("Name\tAge")
````

Some commonly used escape characters are:

| Escape Sequence | Description |
|-----------------|-------------|
| `\'` | Single quote |
| `\"` | Double quote |
| `\\` | Backslash |
| `\n` | New line |
| `\t` | Horizontal tab |

## Raw Strings

A raw string tells Python to treat backslashes as ordinary characters instead of escape characters.

Raw strings are created by placing the letter `r` before the opening quotation mark.

They are especially useful when working with Windows file paths and regular expressions.

````python
path = r"C:\Users\Abdullah\Documents\Python"
print(path)
````

Without a raw string, Python would interpret some backslashes as escape sequences.

Using raw strings avoids this problem and improves readability.

## String Length

The built-in `len()` function returns the number of characters in a string.

Every character, including spaces and punctuation, is counted.

````python
text = "Data Science"

print(len(text))
````

The `len()` function is one of the most frequently used built-in functions in Python and will be used extensively throughout Data Analysis and Machine Learning workflows.

## Empty String

A string can contain zero characters.

Such a string is called an **empty string**.

````python
empty = ""

print(empty)

print(len(empty))
````

An empty string is still a valid string object.

It is different from `None`, which represents the absence of a value. We will study `None` in the next section.

# 1.4 `None` Type

The `None` type represents the **absence of a value** or the **absence of an object**.

Python provides a single built-in object to represent this concept: `None`.

The data type of `None` is called `NoneType`.

Unlike other data types, there is only **one** `None` object in Python. Every variable assigned `None` refers to this same object.

`None` is commonly used to indicate that:

- A variable has not been assigned a meaningful value yet.
- A value is intentionally missing.
- A function does not return a value.
- An object does not currently exist.

Understanding `None` is important because it appears frequently in professional Python code, Data Analysis libraries, APIs, and Machine Learning workflows.

````python
value = None

print(value)
````

Although `None` is often described as "nothing" or "no value," it is actually a valid Python object.

It has its own data type (`NoneType`) and can be assigned to variables just like any other object.

````python
result = None

user = None

data = None

print(result)

print(user)

print(data)
````

## When Is `None` Used?

`None` is commonly used when a meaningful value is not yet available.

This allows a variable to exist before it receives its final value.

This is preferable to assigning an arbitrary value such as `0` or an empty string, which may have their own meanings.

````python
customer_email = None

print(customer_email)
````

In the previous example, `customer_email` exists, but no email address has been assigned yet.

Later in the program, the variable can be updated with an actual value.

````python
customer_email = None

customer_email = "abdullah@example.com"

print(customer_email)
````

## `None` Is Not the Same as Other "Empty" Values

`None` has a different meaning from values such as:

- `0`
- `0.0`
- `False`
- `""` (empty string)
- `[]` (empty list)
- `{}` (empty dictionary)

These values all represent actual objects with specific meanings.

`None` specifically represents the absence of a value.

````python
print(None)

print(0)

print(False)

print("")

print([])
````

Although some of these values may evaluate to `False` in Boolean expressions, they are not equal to `None` and should not be treated as interchangeable.

## Comparing with `None`

The recommended way to check whether a variable is `None` is to use the identity operator `is`.

```python
variable is None
```

Using `is` checks whether both operands refer to the same object.

Since there is only one `None` object in Python, this is the correct and recommended approach.

Avoid using the equality operator (`==`) when checking for `None`.

````python
value = None

print(value is None)
````

````python
value = 100

print(value is None)
````

## Professional Notes

- `None` represents the absence of a value.
- `None` is a built-in singleton object in Python.
- The data type of `None` is `NoneType`.
- `None` is different from `0`, `False`, an empty string, or an empty collection.
- The recommended way to compare with `None` is by using `is` or `is not`.
- `None` is widely used in functions, object-oriented programming, APIs, and Data Analysis libraries to represent missing or unavailable values.

# 1.5 Determining an Object's Type

Python is a dynamically typed language, which means the data type of an object is determined automatically at runtime.

In many situations, it is useful to determine the type of an object. For example, a program may need to:

- Validate user input.
- Process different data types differently.
- Debug unexpected behavior.
- Verify that function arguments have the expected type.
- Perform data validation in Data Analysis pipelines.

Python provides two built-in functions for this purpose:

- `type()`
- `isinstance()`

Although both functions are used to work with data types, they serve different purposes and should be used appropriately.

## 1.5.1 The `type()` Function

The `type()` function returns the type of an object.

### Syntax

```python
type(object)
```

The returned value is the class (type) of the object.

````python
age = 22

print(type(age))
````

````python
print(type(10))

print(type(3.14))

print(type(True))

print(type("Python"))

print(type(None))
````

Notice that the output is displayed as:

```python
<class 'int'>
```

rather than simply:

```python
int
```

This is because, in Python, data types themselves are objects known as **classes**.

At this stage, you only need to understand that `type()` tells us the type of an object. The concept of classes will be studied later in Object-Oriented Programming.

## Practical Example

````python
customer_name = "Abdullah"

customer_age = 22

salary = 45000.50

is_student = True

print(type(customer_name))

print(type(customer_age))

print(type(salary))

print(type(is_student))
````

The `type()` function is commonly used while learning Python, debugging programs, and inspecting unfamiliar objects.

However, in professional Python code, another function is often preferred when checking an object's type.

## 1.5.2 The `isinstance()` Function

The `isinstance()` function checks whether an object belongs to a specified type.

Instead of returning the type itself, it returns a Boolean value:

- `True`
- `False`

### Syntax

```python
isinstance(object, type)
```

````python
age = 22

print(isinstance(age, int))
````

````python
print(isinstance(3.14, float))

print(isinstance("Python", str))

print(isinstance(True, bool))

print(isinstance(None, type(None)))
````

If the object belongs to the specified type, `isinstance()` returns `True`.

Otherwise, it returns `False`.

````python
print(isinstance(100, str))

print(isinstance("Python", int))

print(isinstance(3.14, bool))
````

## Checking Multiple Types

The second argument of `isinstance()` can also be a tuple of types.

This allows an object to be checked against multiple possible data types in a single statement.

````python
number = 25
print(isinstance(number, (int, float)))
````

````python
value = 3.14

print(isinstance(value, (int, float)))
````

This feature is particularly useful when a function accepts more than one valid data type.

## `type()` vs `isinstance()`

Although both functions are related to data types, they have different purposes.

| `type()` | `isinstance()` |
|----------|----------------|
| Returns the type of an object. | Checks whether an object belongs to a specified type. |
| Returns a type object. | Returns `True` or `False`. |
| Mainly used for inspection and debugging. | Mainly used for type checking and validation. |
| Does not perform a Boolean check directly. | Directly returns a Boolean result. |

In professional Python code, `isinstance()` is generally preferred for checking an object's type because it is more flexible and works correctly with inheritance, which will be covered later in Object-Oriented Programming.

## Professional Notes

- Use `type()` when you want to know the exact type of an object.
- Use `isinstance()` when you want to verify that an object belongs to a particular type.
- `isinstance()` returns a Boolean value.
- `type()` returns the object's type.
- `isinstance()` can check multiple types by using a tuple.
- In professional Python code, `isinstance()` is generally preferred over `type()` for type checking.

# 1.6 Type Casting

Type casting is the process of converting an object from one data type to another.

Python provides several built-in functions for type conversion. The most commonly used are:

- `int()`
- `float()`
- `str()`

Type casting is useful when:

- Processing user input.
- Cleaning and transforming data.
- Performing mathematical operations.
- Preparing data for analysis and Machine Learning.
- Converting values into a desired format.

Not every conversion is valid. A conversion is only successful if the source object can be meaningfully converted into the target type.

## 1.6.1 The `int()` Function

The `int()` function converts a compatible object into an integer.

### Syntax

```python
int(x)
```

If the conversion is successful, an object of type `int` is returned.

````python
print(int(25.9))

print(int("100"))

print(int(True))

print(int(False))
````

When converting a floating-point number to an integer, the decimal portion is **discarded** rather than rounded.

This behavior is known as **truncation**.

````python
print(int(8.99))

print(int(8.01))

print(int(-5.8))
````

The string passed to `int()` must represent a valid integer.

Otherwise, Python raises a `ValueError`.

````python
# Valid

print(int("250"))

# Invalid

# print(int("Python"))

# print(int("25.5"))
````

## 1.6.2 The `float()` Function

The `float()` function converts a compatible object into a floating-point number.

### Syntax

```python
float(x)
```

The returned object has the data type `float`.

````python
print(float(25))

print(float("25"))

print(float("25.75"))

print(float(True))

print(float(False))
````

Unlike `int()`, the `float()` function accepts both integer strings and decimal strings.

If the string cannot be interpreted as a valid floating-point number, a `ValueError` is raised.

````python
# Valid

print(float("3.14159"))

# Invalid

# print(float("Python"))
````

## 1.6.3 The `str()` Function

The `str()` function converts an object into its string representation.

### Syntax

```python
str(object)
```

Unlike `int()` and `float()`, the `str()` function can convert almost any Python object into a string.

````python
print(str(100))

print(str(3.14))

print(str(True))

print(str(None))
````

````python
age = 22

message = "Age: " + str(age)

print(message)
````

Converting objects to strings is commonly required when displaying output, creating reports, writing log messages, or generating text-based files.

## Common Type Conversions

| Expression | Result | Type |
|------------|--------|------|
| `int(8.9)` | `8` | `int` |
| `float(8)` | `8.0` | `float` |
| `str(100)` | `"100"` | `str` |
| `int(True)` | `1` | `int` |
| `int(False)` | `0` | `int` |
| `float(True)` | `1.0` | `float` |
| `float(False)` | `0.0` | `float` |
| `str(True)` | `"True"` | `str` |
| `str(None)` | `"None"` | `str` |

## Invalid Type Conversions

Not every object can be converted into every data type.

Attempting an invalid conversion raises an exception.

Some common examples include:

| Expression | Reason |
|------------|--------|
| `int("Python")` | The string does not represent an integer. |
| `int("25.8")` | The string represents a floating-point number, not an integer. |
| `float("Hello")` | The string does not represent a valid number. |

When working with external data, always ensure that values are valid before attempting type conversion.

## Type Casting in Data Analysis

Type conversion is one of the most common tasks in Data Analysis.

Examples include:

- Converting numeric strings to integers or floating-point numbers.
- Preparing numerical data for calculations.
- Reading data from CSV files where numbers are stored as strings.
- Converting values before visualization.
- Preparing features for Machine Learning models.

Data cleaning and preprocessing frequently involve converting data from one type to another, making type casting an essential skill for every Data Analyst and Machine Learning Engineer.

## Professional Notes

- Type casting converts an object from one data type to another.
- `int()` converts compatible objects to integers.
- `float()` converts compatible objects to floating-point numbers.
- `str()` converts objects to their string representation.
- Invalid conversions raise exceptions.
- Type casting is extensively used in data cleaning, preprocessing, user input handling, and data transformation.

# 2. Operators

Operators are special symbols or keywords that perform operations on one or more operands.

An **operand** is the value or object on which an operator acts.

Python provides several categories of operators for performing arithmetic calculations, comparisons, logical evaluations, assignments, membership tests, and identity checks.

A solid understanding of operators is essential for writing expressions, implementing program logic, manipulating data, and performing calculations in Data Analysis and Machine Learning.

---

# 2.1 Arithmetic Operators

Arithmetic operators perform mathematical calculations on numeric operands.

Python supports the following arithmetic operators:

| Operator | Description |
|----------|-------------|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor Division |
| `%` | Modulus (Remainder) |
| `**` | Exponentiation |

````python
a = 10
b = 3

print(a + b)    # Addition

print(a - b)    # Subtraction

print(a * b)    # Multiplication

print(a / b)    # Division

print(a // b)   # Floor Division

print(a % b)    # Modulus

print(a ** b)   # Exponentiation
````

## Division (`/`)

The division operator always returns a value of type `float`, even when the mathematical result is a whole number.

````python
print(10 / 2)

print(type(10 / 2))
````

## Floor Division (`//`)

The floor division operator returns the quotient after discarding the fractional part and rounding down to the nearest whole number.

This behavior is particularly important when working with negative numbers.

````python
print(10 // 3)

print(15 // 4)

print(-10 // 3)
````

Notice the result of:

```python
-10 // 3
```

The result is `-4`, not `-3`.

Floor division rounds **toward negative infinity**, not toward zero.

## Modulus (`%`)

The modulus operator returns the remainder after division.

It is commonly used for:

- Checking whether a number is even or odd.
- Determining divisibility.
- Implementing cyclic operations.
- Working with indexing and hashing algorithms.

# 2.3 Logical Operators

Logical operators combine or modify Boolean expressions.

They are primarily used to evaluate multiple conditions and control the flow of a program.

Python provides three logical operators:

| Operator | Description |
|----------|-------------|
| `and` | Returns `True` if both expressions are `True`. |
| `or` | Returns `True` if at least one expression is `True`. |
| `not` | Reverses the Boolean value of an expression. |

## Logical `and`

The `and` operator returns `True` only when **both** expressions evaluate to `True`.

If either expression is `False`, the result is `False`.

````python
print(True and True)

print(True and False)

print(False and True)

print(False and False)
````

````python
age = 25

print(age >= 18 and age <= 30)

print(age > 30 and age < 40)
````

The `and` operator is commonly used when **multiple conditions must all be satisfied**.

## Logical `or`

The `or` operator returns `True` if **at least one** expression evaluates to `True`.

It returns `False` only when both expressions are `False`.

````python
print(True or True)

print(True or False)

print(False or True)

print(False or False)
````

````python
day = "Saturday"
print(day == 'Saturday' or day == 'Sunday')
print(day == "Monday" or day == "Tuesday")
````

The `or` operator is useful when **any one of multiple conditions** is sufficient.

## Logical `not`

The `not` operator reverses the Boolean value of an expression.

| Original Value | Result |
|---------------|--------|
| `True` | `False` |
| `False` | `True` |

````python
print(not True)

print(not False)
````

````python
is_logged_in = False

print(not is_logged_in)
````

The `not` operator is commonly used to check whether a condition is **not** satisfied.

## Short-Circuit Evaluation

Python evaluates logical expressions using **short-circuit evaluation**.

- For `and`, if the first expression evaluates to `False`, the second expression is **not evaluated**, because the overall result cannot become `True`.

- For `or`, if the first expression evaluates to `True`, the second expression is **not evaluated**, because the overall result is already `True`.

Short-circuit evaluation improves program efficiency and helps prevent unnecessary computations.

````python
x = 0

print(x != 0 and 10 / x > 1)
````

In the previous example, the second expression is never evaluated because the first expression is already `False`.

This prevents a division-by-zero error and demonstrates one of the practical benefits of short-circuit evaluation.

## Professional Notes

- Logical operators always evaluate Boolean expressions.
- `and` requires all conditions to be `True`.
- `or` requires at least one condition to be `True`.
- `not` reverses a Boolean value.
- Python uses short-circuit evaluation for `and` and `or`, improving both efficiency and safety.

# 2.6 Identity Operators

Identity operators compare the **identity** of two objects.

Unlike comparison operators, which compare **values**, identity operators compare whether two variables refer to the **same object in memory**.

Python provides two identity operators:

| Operator | Description |
|----------|-------------|
| `is` | Returns `True` if both variables refer to the same object. |
| `is not` | Returns `True` if the variables refer to different objects. |

## Value vs Identity

Understanding the difference between **value** and **identity** is essential.

- `==` compares whether two objects have the same value.
- `is` compares whether two variables refer to the same object.

Although these operators may sometimes produce the same result, they answer different questions.

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)

print(a is b)
````

In the previous example:

- `a == b` returns `True` because both lists contain the same values.
- `a is b` returns `False` because `a` and `b` refer to two different list objects.

````python
a = [1, 2, 3]

b = a

print(a == b)

print(a is b)
````

Here, both variables refer to the same list object.

Therefore, both the value comparison (`==`) and the identity comparison (`is`) return `True`.

## The `is not` Operator

The `is not` operator checks whether two variables refer to different objects.

````python
x = [10, 20]

y = [10, 20]

print(x is not y)
print(x != y)
````

## When Should `is` Be Used?

In everyday programming, `==` is used for comparing values.

The `is` operator should generally be used only when checking object identity.

A common and recommended use is checking for `None`.

```python
value is None
```

This is the standard practice recommended by the Python documentation.

````python
value = None

print(value is None)

print(value is not None)
````

## Professional Notes

- Identity operators compare objects, not values.
- `==` checks value equality.
- `is` checks object identity.
- `is not` checks whether two variables refer to different objects.
- Use `==` for comparing values.
- Use `is` and `is not` primarily when checking against `None` or when object identity is specifically required.

# 3.3 Formatted String Literals (f-Strings)

Formatted String Literals, commonly known as **f-strings**, provide a concise and readable way to embed expressions directly within strings.

Introduced in Python 3.6, f-strings are the recommended approach for string formatting because they are more readable, concise, and efficient than older formatting techniques.

They are extensively used in Data Analysis, Machine Learning, reporting, logging, and general Python development.

## Syntax

```python
f"Text {expression}"
```

or

```python
f'Text {expression}'
```

Any valid Python expression enclosed within curly braces (`{}`) is evaluated, and its result is inserted into the string.

````python
name = "Abdullah"
age = 22

print(f"My name is {name}.")

print(f"I am {age} years old.")
````

## Embedding Expressions

An f-string can evaluate any valid Python expression, not just variables.

````python
a = 10
b = 5

print(f"{a} + {b} = {a + b}")

print(f"{a} × {b} = {a * b}")

print(f"{a} / {b} = {a / b}")
````

Expressions inside the curly braces are evaluated before the string is displayed.

## Calling Functions Inside f-Strings

Function calls can also be placed inside an f-string.

````python
language = "python"

print(f"Length: {len(language)}")

print(f"Uppercase: {language.upper()}")
````

Since any valid Python expression can be used inside an f-string, built-in functions and object methods are commonly used for formatting output.

General form:

```python
f"{expression:format_specifier}"

````python
pi = 3.1415926535

print(f"{pi:.2f}")

print(f"{pi:.4f}")
````

In the format specifier:

- `.2f` displays two digits after the decimal point.
- `.4f` displays four digits after the decimal point.

Such formatting is frequently used in reports, dashboards, and Data Analysis projects.

````python
accuracy = 98.7564

print(f"Model Accuracy: {accuracy:.2f}%")
````

## Professional Notes

- Always prefix the string with `f` or `F`.
- Expressions are enclosed within `{}`.
- Any valid Python expression can be evaluated inside an f-string.
- f-strings are the recommended formatting method in modern Python.
- They provide excellent readability and performance.

# 4.1 What Is a Virtual Environment?

A **virtual environment** is an isolated Python environment that contains its own:

- Python interpreter
- Installed packages
- Package versions

Each virtual environment operates independently from other environments on the same computer.

This means that packages installed inside one environment do not affect any other project.

Instead of sharing one global collection of packages, every project can have its own isolated environment.

For example:

Project A may require:

- Pandas 2.2
- NumPy 2.0

while Project B may require:

- Pandas 1.5
- NumPy 1.24

Both projects can coexist on the same computer because each uses its own virtual environment.

## Why Is This Necessary?

Imagine you are working on two different projects.

Project A was created two years ago.

Project B is a brand-new Machine Learning project.

Both projects require different versions of the same library.

Without virtual environments, installing the newer version would replace the older version for every project.

As a result, Project A might stop working.

Virtual environments solve this problem by isolating each project's dependencies.

## Without a Virtual Environment

```
Computer
│
├── Python
│
└── Installed Packages
    ├── NumPy
    ├── Pandas
    ├── Matplotlib
    └── Requests
```

Every project shares the same packages.

Updating or removing a package affects every project that uses Python.

## With Virtual Environments

```
Computer
│
├── Project A
│   ├── Source Code
│   └── Virtual Environment
│       ├── NumPy 1.24
│       └── Pandas 1.5
│
├── Project B
│   ├── Source Code
│   └── Virtual Environment
│       ├── NumPy 2.0
│       ├── Pandas 2.2
│       └── Scikit-learn
│
└── Project C
    ├── Source Code
    └── Virtual Environment
        ├── Requests
        └── BeautifulSoup
```

Each project has its own isolated environment and package versions.

## Advantages of Virtual Environments

Virtual environments provide several important benefits:

- Package isolation between projects.
- Independent package versions for each project.
- Reduced dependency conflicts.
- Improved reproducibility.
- Easier collaboration with other developers.
- Cleaner project organization.
- Safer experimentation with new libraries.

For these reasons, virtual environments are considered a standard practice in professional Python development.

Suppose you develop a Machine Learning project today using the latest version of Scikit-learn.

One year later, you receive a bug report.

Without a virtual environment, your computer may already contain newer package versions.

The project might no longer behave exactly as it did when it was originally developed.

However, if the project uses its own virtual environment, you can recreate the original environment and obtain the same package versions, making debugging and maintenance much easier.

# 4.2 Creating a Virtual Environment

Python includes a built-in module named `venv` for creating virtual environments.

The `venv` module creates an isolated Python environment inside your project directory.

Each virtual environment contains its own:

- Python interpreter
- Installed packages
- Package management tools (`pip`)
- Configuration files

Since `venv` is included with the standard Python installation, no additional software is required.

For most Python projects, `venv` is the recommended and most widely used solution for managing virtual environments.

A virtual environment should be created **inside the project's root directory**.

A typical project structure looks like this:

```text
My_Project/
│
├── .venv/
├── main.py
├── requirements.txt
└── README.md
```

The `.venv` directory contains the virtual environment.

Everything outside `.venv` belongs to your project.

## What Happens Internally?

When the virtual environment is created, Python performs several tasks automatically:

- Creates an isolated copy of the Python environment.
- Installs a separate copy of `pip`.
- Creates directories for future package installations.
- Generates configuration files for the environment.

No third-party packages are installed at this stage.

The environment is initially almost identical to a fresh Python installation.

## Why Use the Name `.venv`?

Although the environment can have any name, such as:

```text
env
myenv
project_env
python_env
```

the name `.venv` is widely recommended because:

- It clearly identifies the directory as a virtual environment.
- Many IDEs, including Visual Studio Code, automatically detect it.
- The leading dot (`.`) indicates that it is an implementation detail of the project rather than source code.

## Common Mistakes

Avoid these common mistakes when creating virtual environments:

- Creating one virtual environment for multiple unrelated projects.
- Creating the virtual environment outside the project directory.
- Deleting files inside the `.venv` directory manually.
- Copying a virtual environment from one computer to another.
- Renaming files inside the `.venv` directory.

## Professional Notes

- Use the built-in `venv` module for creating virtual environments.
- Create one virtual environment for each project.
- Place the virtual environment inside the project's root directory.
- The `.venv` naming convention is recommended for consistency and IDE compatibility.
- The contents of the `.venv` directory are managed automatically and should not be edited manually.

# 4.3 Activating and Deactivating a Virtual Environment

After creating a virtual environment, it must be **activated** before it can be used.

Activating a virtual environment tells your operating system and terminal to use the Python interpreter and installed packages from that environment instead of the global Python installation.

While the environment is active:

- Running `python` uses the virtual environment's Python interpreter.
- Running `pip` installs packages into the virtual environment.
- Installed packages remain isolated from other projects.

Every time you begin working on a project, the first step should be to activate its virtual environment.

## Why Activation Is Necessary

Consider the following project structure:

```text
Python_Practice/
│
├── .venv/
├── main.py
└── requirements.txt
```

Although the virtual environment exists, your terminal still uses the **global Python installation** by default.

Until you activate the environment, commands such as:

```bash
python
```

and

```bash
pip
```

will continue using the global installation instead of the project's isolated environment.

## Before Activation

```text
Terminal
    │
    ▼
Global Python
    │
    ▼
Global Packages
```

All Python commands use the global environment.

## After Activation

```text
Terminal
    │
    ▼
Project Virtual Environment (.venv)
    │
    ▼
Project Python
    │
    ▼
Project Packages
```

All Python and `pip` commands now use the project's isolated environment.

## Activating the Environment (Windows)

If you are using Windows PowerShell, run:

```powershell
.venv\Scripts\Activate.ps1
```

If you are using Command Prompt (CMD), run:

```cmd
.venv\Scripts\activate.bat
```

After successful activation, your terminal prompt changes.

For example:

Before activation:

```text
PS C:\Python_Practice>
```

After activation:

```text
(.venv) PS C:\Python_Practice>
```

The prefix:

```text
(.venv)
```

indicates that the virtual environment is currently active.

## Activating the Environment (Linux / macOS)

On Linux or macOS, run:

```bash
source .venv/bin/activate
```

Although the command is different, the purpose is exactly the same: activate the project's virtual environment.

## Verifying the Active Python Interpreter

After activation, it is good practice to verify that Python is using the correct interpreter.

Run:

Windows:

```powershell
Get-Command python
```

Linux / macOS:

```bash
Get-Command python
```

The displayed path should point to the Python executable inside the `.venv` directory rather than the global Python installation.

For example, on Windows the output may resemble:

```text
C:\Python_Practice\.venv\Scripts\python.exe
```

This confirms that the virtual environment is active.

## Deactivating the Environment

When you finish working on the project, you can deactivate the virtual environment.

Run:

```text
deactivate
```

This command works on Windows, Linux, and macOS.

After deactivation:

- The `(.venv)` prefix disappears.
- The terminal returns to using the global Python installation.
- Future `python` and `pip` commands use the global environment until another virtual environment is activated.

## Common Mistakes

Avoid these common mistakes:

- Forgetting to activate the virtual environment before starting work.
- Installing packages before activating the environment.
- Assuming the environment activates automatically in every terminal.
- Ignoring the terminal prompt instead of checking for the `(.venv)` prefix.

## Professional Notes

- Creating a virtual environment does not activate it.
- Always activate the virtual environment before running Python or installing packages.
- Verify activation by checking for the `(.venv)` prefix or by confirming the Python interpreter path.
- Deactivate the environment when you finish working on the project.

# 4.4 Installing Packages with `pip`

Most Python projects depend on external libraries that are not included with the standard Python installation.

Examples include:

- NumPy
- Pandas
- Matplotlib
- Scikit-learn
- Requests

Python uses a package manager called **`pip`** to install, update, uninstall, and manage these third-party packages.

The name `pip` originally stood for **"Pip Installs Packages."**

When a virtual environment is activated, `pip` installs packages **only into that environment**, keeping each project's dependencies isolated.

## What Is a Package?

A **package** is a collection of Python modules that provides additional functionality.

For example:

- `numpy` provides numerical computing tools.
- `pandas` provides data analysis tools.
- `matplotlib` provides plotting capabilities.
- `requests` enables HTTP requests.

Instead of writing these libraries from scratch, developers install them using `pip`.

## Checking the Installed Version of `pip`

Before installing packages, you can verify that `pip` is available.

Run:

```bash
pip --version
```

Example output:

```text
pip 25.x.x from C:\...\site-packages\pip
```

This confirms that `pip` is installed and ready to use.

## Installing a Package

To install a package, use the following syntax:

```bash
pip install package_name
```

For example, to install NumPy:

```bash
pip install numpy
```

`pip` downloads the package from the Python Package Index (PyPI) and installs it into the active virtual environment.

## Installing a Specific Version

Sometimes a project requires a particular version of a package.

Use the `==` operator to specify the version:

```bash
pip install numpy==2.3.2
```

This installs exactly version `2.3.2` of NumPy.

## Viewing Installed Packages

To see all packages installed in the current environment, run:

```bash
pip list
```

This displays the package names and their installed versions.

## Displaying Information About a Package

To view detailed information about an installed package, use:

```bash
pip show package_name
```

Example:

```bash
pip show numpy
```

This displays information such as:

- Package name
- Version
- Installation location
- Dependencies

## Updating a Package

To upgrade an installed package to the latest available version:

```bash
pip install --upgrade package_name
```

Example:

```bash
pip install --upgrade numpy
```

## Uninstalling a Package

To remove a package from the active virtual environment:

```bash
pip uninstall package_name
```

Example:

```bash
pip uninstall numpy
```

`pip` will ask for confirmation before uninstalling the package.

## Common `pip` Commands

| Command | Purpose |
|---------|---------|
| `pip --version` | Display the installed version of `pip`. |
| `pip install package_name` | Install a package. |
| `pip install package==version` | Install a specific version of a package. |
| `pip list` | List installed packages. |
| `pip show package_name` | Display information about a package. |
| `pip install --upgrade package_name` | Upgrade a package. |
| `pip uninstall package_name` | Remove a package. |

# 4.5 The `requirements.txt` File

A `requirements.txt` file is a plain text file that lists all the Python packages required by a project, along with their versions.

Instead of manually remembering or documenting every installed package, developers use a `requirements.txt` file so that the project's dependencies can be recreated automatically on any computer.

This makes projects **portable**, **reproducible**, and **easy to share**.

## Why Is `requirements.txt` Important?

Consider the following scenario:

You develop a Data Analysis project using:

- NumPy
- Pandas
- Matplotlib

A month later, you send the project to another developer.

Without a `requirements.txt` file, they would not know:

- Which packages are required.
- Which versions were used.
- How to recreate your environment.

As a result, the project may fail to run or behave differently.

A `requirements.txt` file solves this problem by documenting the project's dependencies.

## Project Structure

A professional project typically looks like this:

```text
My_Project/
│
├── .venv/
├── main.py
├── requirements.txt
└── README.md
```

The `requirements.txt` file is placed in the project's root directory, alongside the source code.

## Creating a `requirements.txt` File

After installing the required packages, generate the file using:

```bash
python -m pip freeze > requirements.txt
```

The `pip freeze` command lists all installed packages and their exact versions.

The `>` operator redirects the output into a file named `requirements.txt`.

If your virtual environment contains NumPy, the generated file may look similar to:

```text
numpy==2.3.2
```

If additional packages are installed, they will also appear in the file, one package per line.

## Viewing the File

Open `requirements.txt` using Visual Studio Code.

Example:

```text
numpy==2.3.2
pandas==2.3.1
matplotlib==3.10.5
```

Each line specifies:

- Package name
- Exact version

## Installing Packages from `requirements.txt`

Suppose another developer downloads your project.

Instead of installing each package individually, they can install all required packages with a single command:

```bash
python -m pip install -r requirements.txt
```

Here:

- `install` tells `pip` to install packages.
- `-r` means "read the package list from the specified file."

`pip` automatically installs every package listed in the file.

## Typical Workflow

A professional workflow usually follows these steps:

```text
Create Virtual Environment
            │
            ▼
Activate Virtual Environment
            │
            ▼
Install Required Packages
            │
            ▼
Develop the Project
            │
            ▼
Generate requirements.txt
            │
            ▼
Share the Project
            │
            ▼
Another Developer Creates a Virtual Environment
            │
            ▼
Installs Packages Using requirements.txt
```

## Updating `requirements.txt`

Whenever new packages are installed or existing packages are updated, regenerate the file:

```bash
python -m pip freeze > requirements.txt
```

This ensures that the file accurately reflects the current state of the project's environment.

## `pip list` vs `pip freeze`

Although both commands display installed packages, they serve different purposes.

| Command | Purpose |
|---------|---------|
| `python -m pip list` | Displays installed packages in a human-readable table. |
| `python -m pip freeze` | Produces package entries suitable for a `requirements.txt` file. |

For creating `requirements.txt`, always use:

```bash
python -m pip freeze
```

## Common Mistakes

Avoid these common mistakes:

- Creating `requirements.txt` before installing project dependencies.
- Forgetting to update the file after adding new packages.
- Editing package versions manually without a specific reason.
- Assuming another developer has the same packages installed.

## Professional Notes

- Every Python project should include a `requirements.txt` file.
- Generate the file using `python -m pip freeze > requirements.txt`.
- Install dependencies using `python -m pip install -r requirements.txt`.
- Keep the file updated whenever project dependencies change.
- A `requirements.txt` file ensures that the project can be reproduced consistently on any machine.

# 4.6 Never Install Packages Globally

One of the most important habits in professional Python development is:

> **Never install project dependencies into the global Python environment.**

Instead, create and activate a virtual environment for every project before installing any packages.

This ensures that each project remains isolated, reproducible, and independent from every other project on your computer.

## What Is the Global Python Environment?

When Python is installed on a computer, it creates a **global environment**.

If you install a package without activating a virtual environment, the package is installed into this global environment.

As a result, every Python project on your computer can access that package.

## Why Is This a Problem?

Suppose you have two projects.

**Project A** requires:

- `numpy` version `1.26`

**Project B** requires:

- `numpy` version `2.3`

If both projects use the global Python environment, installing one version may replace the other.

This can cause one of the projects to stop working or produce unexpected results.

Virtual environments prevent this problem by giving each project its own isolated package installation.

## Global Installation (Not Recommended)

```text
Computer
│
├── Global Python
│
├── NumPy
├── Pandas
├── Matplotlib
└── Requests
│
├── Project A
├── Project B
└── Project C
```

All projects depend on the same package installations.

Changes made for one project can unintentionally affect every other project.

## Virtual Environment (Recommended)

```text
Computer
│
├── Project A
│   └── .venv
│       ├── NumPy 1.26
│       └── Pandas
│
├── Project B
│   └── .venv
│       ├── NumPy 2.3
│       └── Scikit-learn
│
└── Project C
    └── .venv
        ├── Requests
        └── BeautifulSoup
```

Each project has its own isolated dependencies.

Installing or updating packages in one project does not affect the others.

## The Professional Workflow

Every time you start a new Python project, follow this workflow:

1. Create the project folder.
2. Create a virtual environment.
3. Activate the virtual environment.
4. Install the required packages.
5. Develop the project.
6. Generate a `requirements.txt` file.

Following this workflow from the beginning helps avoid dependency conflicts and makes projects easier to maintain and share.

## Are Global Installations Ever Acceptable?

In general, project dependencies should **not** be installed globally.

However, there are a few exceptions.

Tools that are used across many projects, rather than being part of a specific project, may be installed globally.

Examples include:

- Git
- Visual Studio Code
- Python itself

Some developers also install certain command-line tools globally, but project-specific libraries such as `numpy`, `pandas`, and `scikit-learn` should always be installed inside a virtual environment.

## Common Mistakes

Avoid these common mistakes:

- Installing packages before activating the virtual environment.
- Using the same virtual environment for multiple unrelated projects.
- Assuming that packages installed globally are available in every virtual environment.
- Forgetting to include a `requirements.txt` file when sharing a project.

## Professional Notes

- Create a separate virtual environment for every project.
- Activate the virtual environment before installing packages.
- Install project dependencies only inside the active virtual environment.
- Keep project dependencies documented in `requirements.txt`.
- Avoid relying on the global Python environment for project development.

# 5.1 Python REPL (Interactive Shell)

The **Python REPL** is an interactive environment where Python code is executed immediately after it is entered.

REPL stands for:

- **R**ead
- **E**valuate
- **P**rint
- **L**oop

The REPL continuously performs these four steps:

1. **Read** the code entered by the user.
2. **Evaluate** the code.
3. **Print** the result (if any).
4. **Loop** back and wait for the next input.

This process continues until the user exits the interpreter.

## Why Use the REPL?

The REPL is ideal for:

- Experimenting with Python code.
- Testing small snippets.
- Verifying expressions.
- Learning new functions.
- Debugging simple problems.

Because code executes immediately, the REPL provides instant feedback without creating a Python file.

## Starting the REPL

Open a terminal or command prompt and run:

```bash
python
```

If Python is installed correctly, you should see output similar to:

```text
Python 3.12.x
>>>
```

The `>>>` prompt indicates that the Python REPL is ready to accept Python code.

## Executing Python Code

Once the REPL is running, you can enter Python statements directly.

Each statement is executed immediately after pressing **Enter**.



```
>>> print("Hello, Python!")

Hello, Python!

>>> 10 + 5

15

>>> name = "Abdullah"

>>> print(name)

Abdullah```



## Exiting the REPL

There are several ways to exit the Python REPL.

**Method 1 (Recommended):**

```python
exit()
```

**Method 2 (Windows):**

Press:

```text
Ctrl + Z
```

then press **Enter**.

**Method 3 (Linux/macOS):**

Press:

```text
Ctrl + D
```

## Advantages

- Immediate execution.
- Excellent for experimentation.
- Useful for testing small pieces of code.
- No need to create a Python file.
- Ideal for beginners and quick calculations.

## Limitations

The REPL is **not** suitable for large programs because:

- Code is not automatically saved.
- Organizing large projects is difficult.
- Reusing previous work is inconvenient.
- Version control is impractical.

For larger applications, Python scripts or Jupyter Notebooks are preferred.

## REPL vs Python Scripts

| Python REPL | Python Script |
|-------------|---------------|
| Interactive | File-based |
| Executes one statement at a time | Executes the entire file |
| Best for experimentation | Best for complete programs |
| Temporary | Saved permanently |

## Professional Notes

- The REPL is an interactive Python interpreter.
- It follows the Read–Evaluate–Print–Loop cycle.
- Use the REPL for quick experiments, testing expressions, and learning Python.
- Avoid using the REPL for large or long-term projects.
- Professional developers frequently use the REPL to verify ideas before incorporating them into their programs.

- Data processing pipelines
- Machine Learning applications
- Web applications
- APIs
- Desktop applications

```

Add the following code:

My_Project/
│

The source code is stored in Python files such as `main.py`.

Open the terminal inside the project's root directory.

Python reads the file, executes every statement from top to bottom, and displays the output.

## Example

Run:

```bash

```

Output:

## How Python Executes a Script

When you run:

```bash
python main.py

Python performs the following steps:

1. Opens the file.

# 5.3 Jupyter Notebooks

A **Jupyter Notebook** is an interactive web-based environment that allows you to write, execute, and document Python code in a single document.

Unlike a traditional Python script, a Jupyter Notebook combines:

- Executable Python code
- Formatted text
- Mathematical equations
- Tables
- Images
- Charts
- Visualizations

This makes Jupyter Notebooks one of the most popular tools for Data Analysis, Data Science, Machine Learning, and research.

## Why Jupyter Notebooks?

Data Analysis is an exploratory process.

A data analyst rarely writes an entire program from start to finish.

Instead, they typically:

1. Load data.
2. Inspect the data.
3. Clean the data.
4. Transform the data.
5. Visualize the results.
6. Build models.
7. Analyze the output.

A Jupyter Notebook allows each step to be performed independently while preserving all previous work.

## Notebook Structure

A notebook consists of a sequence of **cells**.

Each cell has a specific purpose.

The two most important cell types are:

- Code Cells
- Markdown Cells

## Code Cells

A **Code Cell** contains executable Python code.

When a code cell is run, the notebook sends the code to the Python kernel for execution.

The result is displayed immediately below the cell.

Example:

```python
name = "Abdullah"

print(f"Welcome, {name}!")
```

Output:

```text
Welcome, Abdullah!
```

## Markdown Cells

A **Markdown Cell** contains formatted documentation rather than executable code.

Markdown is commonly used for:

- Titles
- Headings
- Explanations
- Lists
- Tables
- Notes
- Documentation

Throughout this Python series, we are using Markdown extensively to create a professional notebook that explains every concept alongside the corresponding code.

## Execution Order

One important characteristic of Jupyter Notebooks is that **cells can be executed in any order**.

For example:

```text
Cell 1

x = 10
```

```text
Cell 2

print(x)
```

If Cell 2 is executed before Cell 1, Python raises an error because `x` has not yet been created.

For this reason, it is important to execute notebook cells in a logical order.

## The Python Kernel

A **kernel** is the engine that executes the notebook's Python code.

Every time you run a code cell:

Notebook Cell

↓

Python Kernel

↓

Execution

↓

Output

The kernel also stores all variables, functions, and imported libraries currently in memory.

## Restarting the Kernel

Sometimes you may want to clear the notebook's memory.

Restarting the kernel:

- Removes all variables.
- Removes imported modules from memory.
- Clears the current execution state.

After restarting the kernel, every required cell must be executed again.

## Advantages of Jupyter Notebooks

Jupyter Notebooks provide several important advantages:

- Interactive execution.
- Immediate feedback.
- Rich documentation using Markdown.
- Built-in visualization support.
- Easy experimentation.
- Excellent for presenting analyses.
- Widely used in Data Science and Machine Learning.

## Limitations

Although Jupyter Notebooks are extremely powerful, they are not ideal for every situation.

Some limitations include:

- Execution order can become confusing.
- Large software projects are difficult to organize.
- Merge conflicts in version control systems can be challenging.
- Production applications are typically developed as Python packages rather than notebooks.

## Jupyter Notebook vs Python Script

| Jupyter Notebook | Python Script |
|------------------|---------------|
| Interactive | File-based |
| Cell-by-cell execution | Executes the entire file |
| Combines code and documentation | Primarily code |
| Best for analysis and experimentation | Best for applications and automation |
| Saved as `.ipynb` | Saved as `.py` |

## When Should You Use Jupyter Notebooks?

Jupyter Notebooks are the preferred choice for:

- Learning Python
- Data Analysis
- Exploratory Data Analysis (EDA)
- Data Visualization
- Machine Learning experiments
- Research
- Teaching
- Creating technical reports

For software development, automation, APIs, and production applications, Python scripts are generally more appropriate.

## Professional Notes

- A Jupyter Notebook is an interactive document composed of cells.
- The two primary cell types are Code Cells and Markdown Cells.
- Code Cells execute Python code, while Markdown Cells provide documentation.
- The Python kernel executes code and maintains the notebook's state.
- Jupyter Notebooks are the standard environment for Data Analysis and Machine Learning workflows.

- Sessions may disconnect after periods of inactivity.
- The runtime is temporary, so manually installed packages must be reinstalled in a new session unless otherwise managed.
- Resource availability may vary depending on usage and account type.

- Machine Learning
- Deep Learning
- Educational notebooks

- Working from multiple devices
