# 5. *args (Variable Positional Arguments)

# 6. **kwargs (Variable Keyword Arguments)

## Concept

**kwargs allows a function to accept any number of keyword arguments.

They are stored as a dictionary.

```python
def profile(**info):
  print(info)

print(profile(name= 'Ali', age = 24, city = 'Lahore'))
```

```python
def show_details(**data):
  for key, value in data.items():
    print(key, value)
  return data


result = show_details(name = 'Sara', age = 22, job = 'Data Scientist')
print(result)
```

```python
def show_details(**data):
  result = []

  for key, value in data.items():
    result.append(f"{key}: {value}")

  return result

result = show_details(name = 'Sara', age = 22, job = 'Data Scientist')

for item in result:
  print(item)
```

## Explanation

- `**data` collects keyword arguments
- Stored as dictionary
- Very useful for flexible APIs

# Returning Functions (Higher-Order Functions — Deep Understanding)

## Concept

In Python, functions are first-class objects. This means:

- Functions can be assigned to variables
- Functions can be passed as arguments
- Functions can be returned from other functions

A function that returns another function is called a **higher-order function**.

---

## Why This is Important

Returning functions allows us to:

- Create dynamic behavior
- Build reusable logic
- Delay execution
- Customize function behavior at runtime

This concept is widely used in:

- Decorators
- Machine learning pipelines
- Data preprocessing workflows
- Frameworks and APIs

```python
def outer():
    def inner():
        return "Hello from inner function"
    return inner

func = outer()
print(func())
```

## Step-by-Step Execution

1. `outer()` is called
2. It does NOT execute `inner()`
3. It RETURNS the function `inner`
4. `func` now stores a function
5. `func()` calls that returned function

---

## Key Insight

`outer()` returns a function, not a value.

This is a critical mental shift.

## What Exactly is Returned?

When you write:

    return inner

You are returning the FUNCTION itself, not calling it.

Difference:

- `return inner` → returns function
- `return inner()` → returns result of function

```python
def outer():
    def inner():
        return "Hello"
    return inner

print(outer)       # function reference
print(outer())     # inner function
print(outer()())   # final output
```

## Real Use Case 1: Creating Customized Functions

We can generate functions dynamically based on input.

```python
def multiplier(x):
  def multiply(n):
    return x * n
  return multiply

double = multiplier(2)
triple = multiplier(3)

print(double(5))
print(triple(5))

print(multiplier(3)(5))
```

## Explanation

- `multiplier(2)` creates a function that multiplies by 2
- `multiplier(3)` creates a function that multiplies by 3

---

## Key Insight

- Data transformations

## Real Use Case 2: Function Factory Pattern

A function that creates other functions based on conditions.

```python
def operation(choice):
  def add(a, b):
    return a + b

  def multiply(a, b):
    return a * b

  if choice == 'summing': #Can use any name here "summing" or "add" etc
    return add
  else:
    return multiply

func = operation("summing")
print(func(10,20))

func = operation("multiply")
print(func(10, 5))
```

## Real Use Case 3: Delayed Execution

Returning functions allows execution to happen later.

```python
def greet_later(name):
    def message():
        return f"Hello {name}"
    return message

greet = greet_later("Abdullah")

print("Function created, not executed yet")
print(greet())
```

## Closures

A closure is when a returned function remembers variables from its outer function.

Even after the outer function has finished execution.

```python
def outer(x):
    def inner():
        return x
    return inner

func = outer(10)
print(func())
```

## Explanation

- `x` belongs to outer function
- Outer function finishes
- Inner function STILL remembers `x`

---

## Key Insight

This memory behavior is called a **closure**

Closures are used in:
- Decorators
- Caching systems
- Stateful functions

#1. Local vs Global Scope

## Concept

Scope defines where a variable can be accessed.

There are two basic types:

- Local Scope → inside a function
- Global Scope → outside all functions

```python
x = 10

def show():
  y = 5
  print("Inside Function:", x, y)

show()

print(x)
#print("Outside Function:", y) NameError: name 'y' is not defined
```

## Explanation

- `x` is global → accessible everywhere
- `y` is local → only inside function

---

## Key Insight

Local variables cannot be accessed outside their function.

#2. Modifying Global Variables

## Problem

You cannot modify a global variable directly inside a function.

```python
x = 10

def change():
  #x = x + 5  #Error: UnboundLocalError: cannot access
              #local variable 'x' where it is not associated with
              #a value
  print(x)

change()
```

## Why Error?

Python treats `x` inside function as local.

But it's used before assignment → error.

```python
x = 10

def change():
  x = 3
  x = x + 5
  print(x)

change()
```

# 3. global Keyword

## Concept

The `global` keyword allows you to modify a global variable inside a function.

```python
x = 10

def change():
  global x
  print(x)
  x = x + 5
  print("Inside:", x)

change()
print("Outside:", x)
```

## Explanation

- `global x` tells Python to use global variable
- Now modification works

---

## Warning

Avoid excessive use of global variables — bad practice in large systems.

# 4. Enclosing Scope & nonlocal Keyword

## Concept

When a function is inside another function:

- Inner function can access outer function variables
- But cannot modify them directly

```python
def outer():
  x = 10

  def inner():
    x = x + 5 #Error

  inner()
#outer()
```

##Fix using nonlocal

```python
def outer():
  x = 10

  def inner():
    nonlocal x
    x = x + 5
    print(x)
  inner()
  print(x)

outer()
```

## Explanation

- `nonlocal x` refers to variable in enclosing scope
- Allows modification

---

## Key Insight

`nonlocal` is used for nested functions, not global variables.

# 5. LEGB Rule (Name Resolution)

## Concept

Python searches variables in this order:

L → Local  
E → Enclosing  
G → Global  
B → Built-in

```python
x = 'Global'

def outer():
  x = "Enclosing"


  def inner():
    x = 'Local'
    print(x)
  inner()

outer()
```

## Output

Local

---

## Explanation

Python found `x` in Local scope first, so it stopped searching.

---

## Key Insight

Python stops searching as soon as it finds a match.

##Understand the Terms (Global, Enclosing, Local)

These come from the LEGB rule:

|Level|	Meaning|	Where it exists
|-----|--------|----------|
|L (Local)|	Inside current function|	inner()|
|E (Enclosing)|	Outer function|	outer()|
|G (Global)|	Outside all functions|	file level|
|B (Built-in)|	Python built-ins|	len, sum, etc|

###LEGB Demonstration

```python
x = "Global"

def outer():
    x = "Enclosing"

    def inner():
        print(x)

    inner()

outer()
```

#Built-in Scope

```python
print(len([1,2,3]))
```

## Explanation

- `len` is from built-in scope
- Last level of LEGB

# 7. Shadowing Variables

## Concept

When a local variable has the same name as a global variable, it hides (shadows) it.

```python
x = 10

def show():
    x = 5
    print("Inside:", x)

show()
print("Outside:", x)
```

## Explanation

- Local `x` shadows global `x`
- Does not affect global value

# 1. Pure Functions

## Concept

A pure function is a function that:

1. Always produces the same output for the same input
2. Does NOT modify any external data (no side effects)

---

## Key Properties

- Deterministic (predictable)
- No dependency on external state
- Does not modify input data

```python
def add(a, b):
    return a + b

print(add(10, 5))
print(add(10, 5))
```

## Explanation

- Same input → same output
- No external variables used
- No data modified

This is a pure function.

##Example 2 — Pure Function (Data Context)

```python
def calculate_average(marks):
  return sum(marks)/len(marks)

data = [80, 90, 100]
print(calculate_average(data))
```

## Rule

A pure function should NOT change the input data.

##Bad Example (Not Pure)

```python
def add_item(lst):
    lst.append(100)   # modifies original list
    return lst

data = [1, 2, 3]
print(add_item(data))
print(data)   # changed!
```

## Why This is NOT Pure

- It modifies external data
- Output depends on previous state

This creates bugs in real projects.

#2. Side Effects

## Concept

A side effect is any change a function makes outside its scope.

---

## Examples of Side Effects

- Modifying a global variable
- Changing input data (lists, dictionaries)
- Printing to console
- Writing to files

```python
x = 10

def change():
  global x
  x = x + 5
  return x

print(change())
print(x)
```

## Explanation

- Function modifies global variable
- Output depends on external state

This is a side effect.

##Example — Hidden Bug

```python
def process_data(data):
    data.append(0)
    return data

data = [1, 2, 3]

process_data(data)
process_data(data)

print(data)
```

## Problem

Each function call modifies the same list.

This creates unexpected results.

---

## Insight

This is why side effects are dangerous in data work.

##How to Avoid Side Effects
### Correct Approach

```python
def process_data(data):
  new_data = data.copy()
  new_data.append(0)
  return new_data

data = [1 ,2 ,3]

result1 = process_data(data)
result2 = process_data(data)

print(data)
print(result1)
print(result2)
```

## Why This is Better

- Original data is unchanged
- Each function call is independent
- No hidden bugs

---

## This is closer to a pure function

# Why This Matters in Data Analysis

- You often transform datasets
- You reuse functions
- You run operations multiple times

---

## If your functions have side effects:

- Data gets corrupted

- Debugging becomes very hard

---

## If your functions are pure:

- Safe transformations
- Reusable logic
- Predictable results

- Create new data

#9. Real Data Analyst Example

```python
def clean_data(data):
  try:
    return [int(x) for x in data if x is not None]
  except ValueError:
    print('Invalid data found')
    return []


data = [12,13, None, 11, 34, None, 33]
invalid = [12,13, None, 11, 34, None, 33, 'sd']

print(clean_data(data))
print(clean_data(invalid))
```

```python
def clean_data(data):
  cleaned = []

  for x in data:
    try:
      if x is not None:
        cleaned.append(int(x))
    except ValueError:
      continue

  return cleaned

print(clean_data([12,2,3,4, None, 45, None, 445]))
```

- Handles errors per element
- Does not stop entire process
- More robust for real datasets
