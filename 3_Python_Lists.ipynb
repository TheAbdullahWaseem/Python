# Phase 1 — Foundations

## 1. What is a List?

A list is a built-in Python data structure used to store an ordered collection of elements.

Key properties:
- Ordered
- Mutable
- Allows duplicate values
- Can store heterogeneous data (different data types)

---

## Formal Definition

A list is a **dynamic array** that stores references to objects in contiguous memory locations, allowing efficient indexing and flexible resizing.

---

## Why Lists Exist

Lists solve the fundamental problem of storing and managing multiple values under a single variable.

Without lists:
- Each value must be stored in a separate variable
- Operations on collections become inefficient and unscalable

With lists:
- Data can be grouped logically
- Operations like iteration, filtering, and transformation become possible

---

## Lists vs Tuple vs Set (Conceptual Motivation)

- Lists: Flexible, dynamic, frequently changing data
- Tuples: Fixed, immutable, safe data
- Sets: Unique elements, fast lookup

Lists exist primarily to support **dynamic and ordered data manipulation**.

---

## Example

A list representing transaction values:

transactions = [1200, 500, 3000, 1200]

## 3. Internal Working of Lists (Deep Understanding)

### Dynamic Array Concept

Python lists are implemented as dynamic arrays.

This means:
- The size is not fixed
- The list grows and shrinks as needed

Unlike static arrays (in lower-level languages), Python handles memory management automatically.

---

### Contiguous Memory

List elements are stored in contiguous memory locations.

Implications:
- Fast access via indexing
- Efficient iteration
- Cache-friendly performance

---

### Indexing Mechanism (O(1))

Accessing an element by index is constant time.

Reason:
- Python calculates the memory location directly using the index
- No traversal is required

This is a major performance advantage.

---

### Why Resizing Happens

When a list reaches its capacity and a new element is added:
- Python allocates a larger memory block
- Copies existing elements
- Adds the new element

This operation is expensive (O(n)), but does not happen frequently.

---

### Over-Allocation Strategy

Python uses over-allocation to optimize performance.

Instead of increasing size by 1:
- It increases capacity by a larger factor

Result:
- Fewer resizing operations
- append() becomes amortized O(1)

###Amortized Meaning

If we perform many append operations:

- Most operations → O(1)
- Few operations → O(n)

Average per operation → O(1)
###Think of a notebook:

- You write on empty pages → fast
- When full → you buy a new notebook and copy → slow

But copying happens rarely.

---

## Conceptual Insight

Although resizing is expensive, it happens rarely enough that average performance remains efficient.

````python
import sys

lst = []

for i in range(20):
  lst.append(i)
  print(f"Length: {len(lst)}, Size: {sys.getsizeof(lst)} bytes")
````

### Tuple

### Set

- Does not allow duplicates

Use when:

## Decision-Making Framework

Choose based on problem requirements:

- Need flexibility → List
- Need immutability → Tuple
- Need uniqueness and speed → Set

---

## Performance Perspective

- Lists: O(n) membership testing
- Sets: O(1) membership testing
- Tuples: faster than lists for fixed data

Understanding these trade-offs is essential for writing efficient code.

````python

lst = [1, 2, 2, 3]
tup = (1, 2, 2, 3)
st = {1, 2, 2, 3}

print("List:", lst)
print("Tuple:", tup)
print("Set:", st)
````

## 2. Copying Lists

### copy()

Creates a shallow copy.

```python
new_list = lst.copy()

###list()
``new_list = list(lst)``
###Slicing Copy
``new_list = lst[:]``
###Key Insight

-All three methods create a new list, but:

-Only copy top-level elements
Nested objects are still shared

````python
lst = [1, 2, 3]

copy1 = lst.copy()
copy2 = list(lst)
copy3 = lst[:]

lst[0] = 999

print(copy1, copy2, copy3)
````

## 3. Shallow vs Deep Copy

### Shallow Copy

Copies only the outer list.

Nested objects remain shared.

---

### Deep Copy
``import copy``

Copies everything recursively.

---

### Problem with Nested Lists

```python
lst = [[1, 2], [3, 4]]
copy_lst = lst.copy()

###Key Insight
- Shallow copy → faster, but risky for nested data
- Deep copy → safe, but more expensive

````python
import copy

lst = [[1, 2], [3, 4]]

shallow = lst.copy()
deep = copy.deepcopy(lst)

lst[0][0] = 999

print("Original:", lst)
print("Shallow:", shallow)
print("Deep:", deep)
````

## 4. Internal Behavior and Time Complexity

### append()

- Amortized O(1)
- Adds element at the end

---

### insert()

- O(n)
- Elements shift to the right

---

### delete (del, remove)

- O(n)
- Elements shift to fill gap

---

### Why Shifting Happens

Lists are stored in contiguous memory.

When inserting or deleting:
- Elements must move to maintain order

---

### Conceptual Insight

Contiguous memory gives:
- Fast access (O(1))
- Slow modification in middle (O(n))

````python
lst = [1, 2, 3, 4]

lst.insert(1, 100)
print("After insert:", lst)

del lst[2]
print("After delete:", lst)
````

## 5. List Flattening

Flattening means converting nested lists into a single list.

---

### Example

```python
[[1, 2], [3, 4]] → [1, 2, 3, 4]

````python
nested = [[1, 2], [3, 4], [5, 6]]

for sublist in nested:
  for item in sublist:
    print(item)

#OR Comprehension

flat = [item for sublist in nested for item in sublist]
print("Flattened:", flat)
````

## 6. Transposing Lists (Matrix)

Transposing swaps rows and columns.

---

### Example

```python
[
 [1, 2, 3],
 [4, 5, 6]
]

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

transpose = list(zip(*matrix))
print("Transposed:", transpose)
````

## 7. Chunking Lists

Chunking splits a list into smaller parts.

---

### Example

```python
[1, 2, 3, 4, 5, 6] → [[1, 2], [3, 4], [5, 6]]

````python
lst = [1, 2, 3, 4, 5, 6]
chunks = [lst[i:i+2] for i in range(0, len(lst), 2)]
print("Chunks:", chunks)
````

## 8. Sliding Window Technique

Sliding window processes subsets of fixed size.

---

### Example

```python
[1, 2, 3, 4] → windows of size 3:
[1,2,3], [2,3,4]

````python
lst = [1, 2, 3, 4, 5]

window_size = 3

windows = [lst[i:i+window_size] for i in range(len(lst) - window_size + 1)]
print("Windows:", windows)
````

````python
stack = []
stack.append(10)
stack.append(20)
stack.append(30)

print("Stack:", stack)

print("Pop:", stack.pop())
print("After pop:", stack)
````

## 2. Lists as Queue (FIFO)

Queue follows:

First In, First Out (FIFO)

---

### Using List

```python
queue.append(10)
queue.pop(0)

###Problem
- pop(0) is O(n)
- All elements shift left
###Why This is Inefficient

Lists use contiguous memory:

- Removing first element requires shifting all elements
###Conclusion

- Lists are NOT ideal for queues

````python
queue = [10, 20, 30]

queue.append(40)
print("Queue:", queue)

queue.pop(0)
print("After dequeue:", queue)
````

## 3. When to Use deque

deque (double-ended queue) is optimized for:

- Fast insertion/removal from both ends

---

### Import

```python
from collections import deque

###Advantages
- append() → O(1)
- popleft() → O(1)
###Use Case

Use deque when:

- You need queue behavior
- Frequent insertions/removals from front
###Key Insight

deque solves the inefficiency of lists for queue operations

````python
from collections import deque

queue = deque([10, 20, 30])

queue.append(40)
print("Queue:", queue)

queue.popleft()
print("After Deque:", queue)
````

## 4. Functional Tools

###filter()

###List Comprehension vs map/filter

List comprehension is more Pythonic:
```

###Key Insight
- map/filter → functional style
- list comprehension → readable and preferred

````python
lst = [1, 2, 3, 4]

result_map = list(map(lambda x: x * 2, lst))
print("map:", result_map)

result_filter = list(filter(lambda x: x % 2 == 0, lst))
print("filter:", result_filter)

comp = [x * 2 for x in lst]
print("comprehension:", comp)
````

###Final Insight

- filter() → selection
