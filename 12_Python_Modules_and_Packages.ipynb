# Part 1 — Understanding Modules

## What is a Module?

A **module** is a single Python file (`.py`) that contains Python code. This code can include:

- Variables
- Constants
- Functions
- Classes
- Executable statements
- Documentation

In simple terms, a module is a container that groups related functionality into a single file.

Instead of writing every function inside one large Python file, developers divide related functionality into separate modules. This makes code easier to understand, maintain, reuse, and test.

For example:

```
calculator.py
```

may contain mathematical functions, while

```
data_cleaning.py
```

contains functions for cleaning datasets.

Python treats each `.py` file as an independent module.

---

## Why Modules Exist

Imagine building a Data Analysis project containing:

- Data loading
- Data cleaning
- Feature engineering
- Visualization
- Machine Learning models
- Model evaluation
- Utility functions

If all of this code were written inside one file, it would quickly become difficult to:

- Read
- Debug
- Maintain
- Reuse
- Test

Modules solve this problem by dividing large programs into smaller, focused components.

Instead of one file with thousands of lines:

```
project.py
```

we organize the project as:

```
project/
│
├── data_loader.py
├── preprocessing.py
├── visualization.py
├── model.py
├── evaluation.py
└── main.py
```

Each module has a clear responsibility.

This follows an important software engineering principle called **Separation of Concerns**, where each module is responsible for one logical task.

---

## Advantages of Modular Programming

Using modules provides several important advantages.

### 1. Code Reusability

Functions written once can be imported and reused in multiple projects.

Instead of rewriting code repeatedly, you simply import it.

---

### 2. Better Organization

Large applications become easier to navigate because related code stays together.

Instead of searching through thousands of lines, you know exactly where each functionality belongs.

---

### 3. Easier Maintenance

If a bug exists inside one module, you only modify that module.

Other modules remain unchanged.

---

### 4. Better Collaboration

Professional software is developed by teams.

Different developers can work on different modules simultaneously.

Example:

Developer A

```
preprocessing.py
```

Developer B

```
model.py
```

Developer C

```
visualization.py
```

Without interfering with one another.

---

### 5. Easier Testing

Modules can be tested independently.

Rather than testing an entire application, developers verify one module at a time.

This significantly improves software reliability.

---

### 6. Improved Readability

Smaller files are much easier to understand.

Professional Python projects usually contain many small modules instead of one enormous file.

---

### 7. Namespace Isolation

Every module has its own namespace.

Variables inside one module do not automatically interfere with variables inside another module.

This reduces naming conflicts.

---

## How Python Finds and Loads Modules

When Python encounters an import statement, such as

```python
import math
```

it performs several steps internally.

```
Your Code
     │
     ▼
Python sees import statement
     │
     ▼
Checks whether module is already loaded
     │
     ├──────────────► Yes
     │                  │
     │                  ▼
     │         Reuse existing module
     │
     ▼
No
     │
     ▼
Search for module
     │
     ▼
Read module source
     │
     ▼
Compile (if necessary)
     │
     ▼
Execute module
     │
     ▼
Create module object
     │
     ▼
Store inside sys.modules
     │
     ▼
Program continues
```

This process happens automatically every time Python imports a module.

Let's understand each step.

---

### Step 1 — Check `sys.modules`

Python first checks an internal dictionary called `sys.modules`.

This dictionary stores every module that has already been imported during the current program execution.

If the requested module already exists there, Python simply reuses it.

The module is **not executed again**.

This is one reason imports are very efficient.

---

### Step 2 — Search for the Module

If the module has not been imported before, Python searches for it.

It searches directories listed inside `sys.path`.

We'll study `sys.path` shortly.

---

### Step 3 — Read the Module

Python locates the module file.

Examples:

```
math
numpy
pandas
my_module.py
```

It reads the module's contents.

---

### Step 4 — Compile (If Needed)

Python compiles the source code into bytecode.

Python may also create a cached `.pyc` file inside the `__pycache__` directory to speed up future imports.

This compilation is handled automatically.

---

### Step 5 — Execute the Module

Every statement at the top level of the module is executed exactly once.

For example,

```python
print("Loading module...")
```

will execute when the module is imported for the first time.

Functions and classes are also created during this execution.

---

### Step 6 — Create a Module Object

After execution, Python creates a **module object**.

This object contains everything defined inside the module.

For example,

```
math
```

contains

- `sqrt`
- `sin`
- `cos`
- `pi`
- `e`

Each of these becomes an attribute of the module object.

---

### Step 7 — Cache the Module

The module object is stored inside

```python
sys.modules
```

Future imports simply reuse this object.

The module is **not executed again** unless explicitly reloaded.

---

## The Module Search Path (`sys.path`)

When Python cannot find a module in memory, it searches directories listed in

```python
sys.path
```

`sys.path` is simply a list of directories.

Python searches these directories from top to bottom.

Example search process:

```
import my_module

        │
        ▼

Directory 1

        │
        ▼

Directory 2

        │
        ▼

Directory 3

        │
        ▼

Standard Library

        │
        ▼

Site Packages

        │
        ▼

Found?
```

The first matching module is imported.

---

### What Does `sys.path` Usually Contain?

Typical entries include:

- The current working directory
- Python installation directories
- Standard Library location
- Third-party packages (`site-packages`)
- Directories added using environment variables
- Custom paths added by the programmer

Python searches these locations in order.

---

### Why is `sys.path` Important?

Understanding `sys.path` helps explain many common issues, such as:

- `ModuleNotFoundError`
- Importing the wrong module
- Circular import problems
- Package organization
- Virtual environments

Professional developers frequently inspect `sys.path` when debugging import-related issues.

---

## Internal Relationship

```
project/
│
├── main.py
├── helper.py
│
└── utils.py
```

Running

```python
import helper
```

causes Python to

```
Check sys.modules

        │

Not found

        │

Search directories in sys.path

        │

Find helper.py

        │

Execute helper.py

        │

Create helper module object

        │

Store in sys.modules

        │

Return module object
```

---

## Best Practices

- Keep each module focused on a single responsibility.
- Use descriptive module names.
- Avoid creating very large modules.
- Organize related functionality together.
- Prefer reusable functions over duplicated code.
- Never rely on modifying `sys.path` unless absolutely necessary.

---

## Common Mistakes

- Placing all code in one large file.
- Creating modules with unrelated functionality.
- Assuming importing a module executes it every time.
- Confusing a module with a package.
- Ignoring import errors without understanding the search path.

---

## Professional Notes

Professional Python applications often consist of dozens or even hundreds of modules. Frameworks such as NumPy, Pandas, Scikit-learn, TensorFlow, and Django are all built by organizing functionality into carefully designed modules and packages. Mastering modular programming is therefore essential for writing scalable and maintainable software.

---

## Mentor's Tip

As you progress toward Data Analysis and Machine Learning, you will rarely write everything in a single notebook or script. You'll create reusable utilities for loading data, cleaning datasets, feature engineering, visualization, and model evaluation. Developing the habit of organizing your code into well-designed modules from the beginning will make your projects easier to maintain, debug, and extend as they grow in complexity.

# Part 2: Importing Modules in Python

## Introduction

One of Python's greatest strengths is its rich ecosystem of modules. Instead of writing every function yourself, you can reuse code written by Python's developers, third-party libraries, or even your own modules.

The `import` statement is the mechanism that allows one Python file to access code defined in another module.

Whether you're working with the Python Standard Library, third-party libraries like `pandas` and `numpy`, or your own code, understanding the import system is essential for writing clean, reusable, and maintainable programs.

---

# 1. `import module`

## What is `import module`?

The simplest way to use another module is:

```python
import module_name
```

This imports the **entire module** into your program.

However, **it does not copy the module's contents into your current namespace**. Instead, Python creates a **module object**, binds it to the module name, and you access its contents using the dot (`.`) operator.

Example:

```python
import math
```

Now `math` refers to the module object.

```
math
│
├── sqrt()
├── ceil()
├── floor()
├── pi
├── e
├── sin()
├── cos()
└── ...
```

---

## Why use `import module`?

This approach clearly indicates where every function or variable comes from.

Instead of writing:

```python
sqrt(25)
```

you write:

```python
math.sqrt(25)
```

This improves:

- Readability
- Maintainability
- Debugging
- Avoiding naming conflicts

---

## Syntax

```python
import module_name
```

---

## Internal Working

When Python executes:

```python
import math
```

it:

1. Checks `sys.modules`.
2. Searches the module if not already loaded.
3. Loads and executes the module.
4. Creates a module object.
5. Stores the object in memory.
6. Binds the name `math` to that module object.

Your variable does **not** contain every function individually.

Instead:

```
math
      │
      ▼
 Module Object
      │
      ├── sqrt
      ├── pi
      ├── ceil
      └── floor
```

---

## Advantages

- Prevents name collisions.
- Makes code self-documenting.
- Easy to identify the source of functions.
- Recommended by PEP 8.

---

## Best Practices

- Prefer `import module` when you use multiple functions from the same module.
- Use the module prefix (`module.function()`).
- Avoid importing unused modules.

---

## Common Mistakes

Forgetting the module prefix.

```python
import math

sqrt(25)
```

Raises:

```
NameError
```

Correct:

```python
math.sqrt(25)
```

---

## Professional Notes

Most professional Python code favors `import module` because it makes dependencies explicit and reduces ambiguity.

````python
import math

print(math.sqrt(64))
print(math.ceil(3.1))
print(math.floor(5.5))
print(math.pi)
````

````python
import random

print(random.randint(1, 10))
print(random.choice(["Apple", "Banana", "Orange"]))
````

# 2. `from module import name`

## What is it?

Sometimes you only need one or a few objects from a module.

Instead of importing the entire module namespace, Python allows you to import specific objects.

Syntax:

```python
from module import name
```

Example:

```python
from math import sqrt
```

Now `sqrt()` can be used directly.

---

## Internal Working

When Python executes:

```python
from math import sqrt
```

Python still loads the **entire module** if it has not already been imported.

It does **not** load only the `sqrt()` function.

Instead:

1. Import module.
2. Create module object.
3. Retrieve `sqrt`.
4. Bind `sqrt` directly into your current namespace.

Result:

```
Current Namespace

sqrt ─────────► math.sqrt
```

Notice:

The module object still exists.

Only the reference to `sqrt` is placed in your namespace.

---

## Advantages

- Less typing.
- Cleaner code when using one or two functions repeatedly.
- Useful in mathematical computations.

---

## Disadvantages

The source of the function becomes less obvious.

Compare:

```python
math.sqrt(25)
```

vs

```python
sqrt(25)
```

The first immediately tells the reader where `sqrt` comes from.

---

## Best Practices

Use this style when:

- importing one or two objects
- the imported names are widely recognized
- readability remains clear

Examples:

```python
from pathlib import Path
from datetime import datetime
```

---

## Common Mistakes

Importing too many names.

Bad:

```python
from math import sqrt, floor, ceil, sin, cos, tan, factorial
```

This pollutes your namespace.

Sometimes using

```python
import math
```

is cleaner.

````python
from math import sqrt

print(sqrt(49))
````

````python
from math import pi

radius = 5
area = pi * radius ** 2

print(area)
````

````python
from datetime import datetime

current_time = datetime.now()

print(current_time)
````

# 3. `from module import *`

## What is it?

Python also allows importing **every public object** from a module.

Syntax:

```python
from module import *
```

Example:

```python
from math import *
```

Now every public function and constant becomes available directly.

```
sqrt()
sin()
cos()
log()
pi
e
...
```

without using the module name.

---

## How does it work?

Python imports the module.

Then every public name is copied into your namespace.

```
Before

Current Namespace

↓

After

sqrt
sin
cos
tan
pi
e
log
factorial
...
```

---

## Why is this dangerous?

Imagine two modules both define

```python
calculate()
```

Example:

```
module_a.calculate()

module_b.calculate()
```

Using

```python
from module_a import *
from module_b import *
```

Which `calculate()` will Python use?

The second import overwrites the first.

This creates difficult bugs.

---

## Namespace Pollution

Your namespace becomes crowded.

Instead of a few variables, you suddenly have hundreds.

Large libraries like NumPy or Pandas expose hundreds or thousands of names.

Wildcard imports make it difficult to know where functions originated.

---

## Readability Problem

Consider:

```python
mean(data)
```

Where did `mean()` come from?

- statistics?
- numpy?
- pandas?
- scipy?

Nobody knows without searching.

Now compare:

```python
statistics.mean(data)
```

or

```python
np.mean(data)
```

Much clearer.

---

## PEP 8 Recommendation

PEP 8 explicitly discourages wildcard imports except in rare situations.

Professional projects almost never use:

```python
from module import *
```

---

## When is it acceptable?

Very rare cases:

- Interactive Python shell.
- Quick experimentation.
- Some package initialization files (`__init__.py`).

Not recommended for production code.

````python
from math import *

print(sqrt(81))
print(pi)
print(log(10))
````

# 4. Import Aliases (`import module as alias`)

## What is an Alias?

Python allows assigning another name to a module during import.

Syntax:

```python
import module as alias
```

Example:

```python
import numpy as np
```

Here,

```
np
```

is simply another name that refers to the NumPy module.

```
np
 │
 ▼
NumPy Module
```

---

## Why do we use aliases?

Some module names are:

- long
- frequently used
- difficult to type repeatedly

Instead of

```python
import numpy

numpy.array(...)
numpy.mean(...)
numpy.max(...)
```

we write

```python
import numpy as np

np.array(...)
np.mean(...)
np.max(...)
```

The functionality remains exactly the same.

---

## Internal Working

Python imports the module normally.

Then instead of binding

```
numpy
```

it binds

```
np
```

Both names point to the same module object if both are created.

---

## Standard Aliases

Some aliases have become industry standards.

| Module | Standard Alias |
|---------|----------------|
| `numpy` | `np` |
| `pandas` | `pd` |
| `matplotlib.pyplot` | `plt` |
| `seaborn` | `sns` |

Although aliases are optional, using these standard aliases improves readability because every Python developer recognizes them immediately.

---

## Best Practices

Use aliases:

- when module names are long
- when the alias is an industry standard
- when it improves readability

Avoid meaningless aliases like:

```python
import math as x
```

Good aliases should remain descriptive.

---

## Professional Notes

In Data Analysis and Machine Learning, you will almost always see:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

These aliases are so widely adopted that using different names is discouraged unless there is a compelling reason.

````python
import math as m

print(m.sqrt(100))
print(m.pi)
````

````python
# Professional Data Analysis style

import numpy as np
import pandas as pd

numbers = np.array([10, 20, 30, 40])

data = pd.DataFrame({
    "Sales": numbers
})

print(data)
````

| Import Style              | Namespace | Recommended | Typical Use                                |
| ------------------------- | --------- | ----------- | ------------------------------------------ |
| `import module`           | Clean     | ✅ Yes       | General-purpose programming                |
| `from module import name` | Moderate  | ✅ Yes       | Importing a few specific objects           |

| `import module as alias`  | Clean     | ✅ Yes       | Data Analysis, ML, commonly used libraries |


# 5. Absolute Imports and Relative Imports

## Introduction

Until now, we have imported built-in modules such as `math`, `random`, and `datetime`.

```python
import math
from datetime import datetime
```

These modules are part of Python's Standard Library.

However, professional projects rarely consist of a single Python file. Instead, they contain many files organized into **packages**.

Example:

```
sales_project/
│
├── main.py
├── data/
│   ├── loader.py
│   └── cleaner.py
│
├── models/
│   ├── train.py
│   └── predict.py
│
└── utils/
    ├── helpers.py
    └── logger.py
```

Now suppose `train.py` needs a function from `helpers.py`.

How should it import that function?

Python provides two approaches:

- **Absolute Imports**
- **Relative Imports**

Understanding **when** and **why** to use each is more important than memorizing their syntax.

---

# Why Do We Need Different Import Styles?

Imagine you are writing a book.

You can refer to a chapter in two ways.

### Method 1 — Full Address

```
Book
 └── Chapter 5
      └── Section 2
```

You write:

```
Book → Chapter 5 → Section 2
```

This is similar to an **Absolute Import**.

---

### Method 2 — Relative Address

If you are already inside Chapter 5, you simply write

```
Section 2
```

or

```
Go back one chapter.
```

This is similar to a **Relative Import**.

Python supports both because each is useful in different situations.

---

# Absolute Imports

## What is an Absolute Import?

An **Absolute Import** specifies the complete path to the module starting from the project's top-level package.

Instead of saying

> "Go to the module next to me."

you say

> "Start from the project's root and follow this complete path."

---

## Example Project

```
company/
│
├── main.py
│
├── employees/
│   ├── manager.py
│   └── payroll.py
│
└── utilities/
    └── logger.py
```

Suppose

```
manager.py
```

needs

```
logger.py
```

Using an absolute import:

```python
from utilities.logger import log_message
```

Python reads this as:

```
Start from the project root

↓

Find "utilities"

↓

Find logger.py

↓

Import log_message
```

Notice that the location of `manager.py` is irrelevant.

Python follows the complete path from the project's root.

---

## Visualization

```
Project Root
│
├── employees
│      │
│      └── manager.py
│
└── utilities
       │
       └── logger.py
```

Absolute Import

```
Project Root
     │
     ▼
utilities
     │
     ▼
logger
     │
     ▼
log_message
```

---

## Syntax

```python
from package.module import name
```

or

```python
import package.module
```

---

## Advantages

### 1. Very Readable

Anyone reading

```python
from utilities.logger import log_message
```

immediately knows where the function comes from.

---

### 2. Easier to Maintain

Suppose you open a project after six months.

You immediately know the module's location.

---

### 3. Works Well in Large Projects

Large Python libraries like

- NumPy
- Pandas
- Django
- Scikit-learn

primarily use absolute imports.

---

### 4. Preferred by PEP 8

Python officially recommends absolute imports whenever practical.

---

## Disadvantages

The import statement can become longer.

Example

```python
from project.utilities.data.cleaning.helpers import remove_nulls
```

Although long, it is still explicit and easy to understand.

---

# Relative Imports

## What is a Relative Import?

A relative import specifies the module's location **relative to the current module**, not from the project root.

Instead of giving the full path, you tell Python

> "Start from my current location."

---

## Why Relative Imports Exist

Suppose

```
employees/
│
├── manager.py
└── payroll.py
```

Both files are inside the same package.

Rather than writing

```python
from company.employees.payroll import calculate_salary
```

inside `manager.py`

you can simply write

```python
from .payroll import calculate_salary
```

The dot means

> "Look inside my current package."

---

# Understanding the Dot (`.`)

This is the part that confuses many students.

The dot **does not mean the current folder on your operating system**.

It means:

> **The current Python package.**

Remember this distinction.

---

## One Dot (`.`)

```
from .module import something
```

Meaning

```
Current Package

↓

module.py
```

---

## Two Dots (`..`)

```
from ..utilities import helper
```

Meaning

```
Current Package

↑

Parent Package

↓

utilities
```

---

## Three Dots (`...`)

```
from ...config import settings
```

Meaning

```
Current Package

↑

Parent

↑

Grandparent

↓

config
```

Although Python allows this, using many dots usually indicates that the project structure should be improved.

---

# Example

```
project/
│
├── employees/
│   ├── manager.py
│   └── payroll.py
│
└── utilities/
    └── logger.py
```

Inside

```
manager.py
```

Same package:

```python
from .payroll import calculate_salary
```

Go to parent package:

```python
from ..utilities.logger import log_message
```

---

# How Python Resolves Relative Imports

Suppose Python executes

```python
from .payroll import calculate_salary
```

Python internally thinks

```
Current package?

↓

employees

↓

Find payroll.py

↓

Import calculate_salary
```

Notice that Python must know

- current package
- parent package

Without package information, relative imports cannot work.

---

# Why Relative Imports Sometimes Fail

Many beginners encounter

```
ImportError:
attempted relative import with no known parent package
```

Example:

```
project/
│
└── employees/
      ├── manager.py
      └── payroll.py
```

Running

```
python manager.py
```

and writing

```python
from .payroll import calculate_salary
```

causes an error.

Why?

# 6. The `__name__ == "__main__"` Pattern

## Introduction

As you begin organizing your code into multiple modules and packages, you'll often encounter the following statement:

```python
if __name__ == "__main__":
    ...
```

This pattern appears in countless professional Python projects, open-source libraries, tutorials, and automation scripts.

Although the syntax is simple, understanding **why it exists** and **how it works internally** is essential for writing reusable and maintainable Python code.

---

# Why Does This Pattern Exist?

Imagine you have written a Python file that contains several utility functions.

Example:

```python
def add(a, b):
    return a + b
```

Now consider two different situations.

### Situation 1

You execute the file directly.

```text
python calculator.py
```

---

### Situation 2

Another Python file imports it.

```python
import calculator
```

Should Python behave the same way in both situations?

**No.**

When the file is executed directly, it should run the program.

When the file is imported, it should simply provide its functions, classes, and variables without automatically executing the program.

Python distinguishes between these two situations using the special variable `__name__`.

---

# What is `__name__`?

`__name__` is a **special built-in variable** that Python automatically creates for every module.

You never create it yourself.

Python assigns a value to it before executing the module.

Every Python module has its own `__name__` variable.

---

# How Does Python Assign `__name__`?

There are only two important cases.

## Case 1 — The File is Executed Directly

Suppose you run

```text
python app.py
```

Python assigns

```python
__name__ = "__main__"
```

to that file.

---

## Case 2 — The File is Imported

Suppose another file imports it.

```python
import app
```

Now Python assigns

```python
__name__ = "app"
```

Notice that the value is the **module name**, not `"__main__"`.

---

# Visualizing the Difference

### Running Directly

```
python calculator.py

        │

        ▼

__name__ = "__main__"
```

---

### Importing

```
main.py

      │

import calculator

      │

      ▼

__name__ = "calculator"
```

This single difference determines whether the code inside the `if` block executes.

---

# Understanding the Pattern

The pattern is simply an `if` statement.

```python
if __name__ == "__main__":
    ...
```

Python asks one question:

> "Is this file the program that the user started?"

If the answer is **Yes**, execute the code.

If the answer is **No**, skip it.

Nothing magical happens beyond this.

---

# Internal Working

Suppose Python starts executing

```
calculator.py
```

Internally, it behaves like this.

```
Start Program

      │

Assign __name__

      │

Is file executed directly?

      │

 ┌────┴────┐
 │         │
Yes        No
 │         │
 │         │
__name__   __name__
="__main__"="calculator"
 │         │
 │         │
Execute    Skip
if block   if block
```

---

# Why is This Useful?

Suppose you create a module containing reusable functions.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

You also want to test these functions.

You could write

```python
print(add(10, 5))
```

But there is a problem.

Whenever another file imports this module,

```python
import calculator
```

the `print()` statement will also execute.

That is usually **not** what we want.

Instead, we place testing code inside

```python
if __name__ == "__main__":
```

Now the functions are available when imported, while the test code runs only when the file is executed directly.

---

# Professional Structure

A professional module often looks like this.

```
Imports

↓

Constants

↓

Classes

↓

Functions

↓

if __name__ == "__main__":
        Testing
        Demo
        Entry Point
```

This keeps reusable code separate from executable code.

---

# Common Uses

The `if __name__ == "__main__":` block is commonly used for:

- Testing functions
- Demonstrating module usage
- Running example code
- Creating command-line programs
- Defining the application's entry point
- Preventing unwanted execution during imports

---

# Real-World Example

Imagine you write a module that loads CSV files.

Other projects should be able to import it without automatically reading files or printing output.

However, while developing the module, you may want to test it.

The `__main__` block allows you to include testing code without affecting other programs that import your module.

This pattern is widely used in Data Analysis, Machine Learning, automation scripts, and production applications.

---

# Best Practices

- Place executable code inside the `if __name__ == "__main__":` block.
- Keep reusable functions and classes outside the block.
- Use the block for testing, demonstrations, or as the application's entry point.
- Avoid writing important business logic directly inside the block; instead, call well-defined functions.

---

# Common Mistakes

### Mistake 1

Writing all program logic directly inside the `if __name__ == "__main__":` block.

Instead, define functions and call them from the block.

---

### Mistake 2

Putting reusable code inside the block.

Anything inside the block cannot be accessed simply by importing the module.

---

### Mistake 3

Thinking that `__main__` is a function or keyword.

It is simply the string value assigned to the special variable `__name__` when a module is executed directly.

---

# Professional Notes

Almost every mature Python codebase uses the `if __name__ == "__main__":` pattern. It promotes modularity by separating reusable definitions from executable behavior. As your projects grow, this pattern allows the same file to serve both as an importable module and as a standalone program.

---

# Mentor's Tip

As a professional developer, get into the habit of writing reusable functions first and using the `if __name__ == "__main__":` block only to start your program or demonstrate how the module works. This approach makes your code easier to test, reuse, and maintain.

````python
print(__name__)
````

Running Code Only When Executed Directly

````python
def greet():
  print("Welcome to Python!")

if __name__ == '__main__':
  greet()
````

````python
greet()
````

A Professional Module

````python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def main():
    print("Addition:", add(10, 5))
    print("Subtraction:", subtract(10, 5))


if __name__ == "__main__":
    main()
````

# `random.seed()`

## What is `seed()`?

`seed()` initializes Python's pseudo-random number generator.

In simple terms, it determines where the sequence of random numbers begins.

If the same seed value is used, Python generates exactly the same sequence of random numbers every time.

---

## Why is This Important?

Suppose you are training a Machine Learning model.

Your algorithm randomly selects training samples.

Tomorrow you run the same notebook again.

Without fixing the seed, the random selections may differ, producing different results.

Using a fixed seed ensures your experiments are reproducible.

---

## Syntax

```python
random.seed(value)
```

The value is usually an integer.

---

## Internal Working

Think of the random number generator as a very long predetermined sequence.

```
Seed = 42

↓

0.639...
↓

0.025...
↓

0.275...
↓

0.223...
```
The numbers 42, 20, 100, or 999 in random.seed() are not special numbers. They are simply starting values (called seeds) that tell Python where to begin generating its pseudo-random sequence.


Whenever you use

```python
random.seed(42)
```

Python starts from the same point in the sequence.

Therefore, the generated values are identical across runs.

---

## Best Practices

Use a fixed seed when:

- Developing software
- Debugging
- Training machine learning models
- Writing research code
- Creating reproducible experiments

Avoid fixing the seed in production systems where genuine unpredictability is desired.

---

# What is `sys.path`?

`sys.path` is a **list of directory paths** that Python searches whenever it needs to import a module.

Think of it as Python's **search list**.

When Python encounters an import statement, it checks each directory in `sys.path` one by one until it finds the requested module.

---

# Internal Working

Suppose your code contains:

```python
import math
```

Python internally performs a search similar to this:

```
Need to import "math"

        │

        ▼

Directory 1

Found?

No

        │

        ▼

Directory 2

Found?

No

        │

        ▼

Directory 3

Found?

Yes

        │

        ▼

Import Module

        │

        ▼

Program Continues
```

If Python reaches the end of the search list without finding the module, it raises:

```text
ModuleNotFoundError
```

---

# Why Does Python Need `sys.path`?

Imagine Python had no search path.

Whenever you imported a module, Python would have no idea where to look.

Instead, Python maintains a list of locations that it searches automatically.

This makes importing modules simple and consistent.

---

# What Does `sys.path` Contain?

Typically, `sys.path` contains directories such as:

- The current working directory
- Python's Standard Library
- Installed third-party packages (for example, packages installed using `pip`)
- Other directories configured by your environment

The exact contents vary depending on your operating system and Python installation.

```

# Viewing the Search Path

You can inspect the directories Python searches by printing `sys.path`.

````python
import sys

print(sys.path)
````

# Understanding the Search Order

Python searches the directories in `sys.path` **from top to bottom**.

The first matching module is imported.

Suppose `sys.path` contains:

```
1. Current Project

2. Standard Library

3. Third-Party Packages
```

Now your code executes:

```python
import math
```

Python performs the following search:

```
Current Project

↓

Is there math.py?

No

↓

Standard Library

↓

Is there math.py?

Yes

↓

Import math

↓

Stop Searching
```

Once Python finds the module, it immediately stops searching.

This search order is extremely important because the first matching module always wins.

# Why Does `ModuleNotFoundError` Occur?

Consider the following statement:

```python
import employee
```

Python checks every directory in `sys.path`.

```
Directory 1

↓

Not Found

↓

Directory 2

↓

Not Found

↓

Directory 3

↓

Not Found

↓

End of Search
```

Since the module cannot be located, Python raises:

```text

No module named 'employee'
```

This is not because the `import` statement is incorrect.

It simply means Python could not find the requested module in any of its search locations.

````python
try:
  import employee

except ModuleNotFoundError as MNF:
  print(MNF)
````

```

Datetime Object

```

```

↓

```

## Why Do We Need `strptime()`?

Suppose a CSV file contains:

```

Python reads this value as a string.

- How many days separate two dates?

To perform these operations, we first convert the string into a `datetime` object.

````python
from datetime import datetime

date_string = "2026-07-09"

parsed_date = datetime.strptime(date_string, "%Y-%m-%d")

print(parsed_date)
print(type(parsed_date))
````

Year

Month
```

```
09

Day
```

Once every part matches successfully, Python creates a `datetime` object.

````python
f = date.today()
print(f)
s = "2026-07-09"
print(s)
print(type(s))
convertion = datetime.strptime(s, "%Y-%m-%d")
print(convertion)
print(type(convertion))
````

````python
from collections import Counter

counter1 = Counter({
    "A": 5,
    "B": 2
})

counter2 = Counter({
    "A": 3
})

print(counter1 - counter2)
````

````python
from collections import Counter

c1 = Counter({
    "A": 5,
    "B": 2
})

c2 = Counter({
    "A": 3,
    "B": 4
})

print(c1 & c2)
````

````python
from collections import Counter

c1 = Counter({
    "A": 5,
    "B": 2
})

c2 = Counter({
    "A": 3,
    "B": 4
})

print(c1 | c2)
````

# Lesson 3 (Part 1) — `defaultdict`

## What is `defaultdict`?

`defaultdict` is a specialized dictionary provided by Python's `collections` module.

It behaves almost exactly like a normal `dict`, with one important difference:

> If you access a key that does not exist, `defaultdict` automatically creates it using a predefined default value instead of raising a `KeyError`.

This makes your code cleaner and eliminates many repetitive checks.

---

## Why Do We Need `defaultdict`?

Consider a normal dictionary.

```python
student_marks = {
    "Ali": 85,
    "Ahmed": 91
}
```

Accessing an existing key works normally.

```python
student_marks["Ali"]
```

However,

```python
student_marks["Sara"]
```

raises

```
KeyError
```

because `"Sara"` does not exist.

In many programs, we actually want Python to create the missing key automatically.

That is exactly what `defaultdict` does.

---

## Internal Working

A normal dictionary behaves like this:

```
Dictionary

↓

Key Exists?

↓

Yes

↓

Return Value

↓

No

↓

KeyError
```

A `defaultdict` changes the second branch:

```
defaultdict

↓

Key Exists?

↓

Yes

↓

Return Value

↓

No

↓

Create Default Value

↓

Store It

↓

Return It
```

This automatic key creation is controlled by something called the **default factory**.

---

## What is a Default Factory?

A **default factory** is simply a callable (usually a function or class) that creates the default value for missing keys.

Examples:

| Default Factory | Default Value |
|-----------------|---------------|
| `int` | `0` |
| `list` | `[]` |
| `set` | `set()` |
| `str` | `""` |

Whenever a missing key is accessed, Python calls the factory to create the value automatically.

````python
from collections import defaultdict
````

# Creating a `defaultdict`

## Syntax

```python

```

where `default_factory` is a callable such as:

- `int`
- `list`
- `set`
- `str`

Notice that we pass the callable itself, **not its return value**.

# Lesson 1 — Fundamentals of `deque`

## What is a `deque`?

`deque` stands for **Double-Ended Queue**.

It is a specialized sequence provided by Python's `collections` module that allows **fast insertion and removal of elements from both the left and right ends**.

Unlike a normal `list`, which is optimized mainly for operations at the end, a `deque` is optimized for operations at **both ends**.

---

## Why Does Python Provide `deque`?

Suppose you have a queue of customers.

```
Ali

Ahmed

Sara

John
```

New customers join at one end.

The first customer leaves from the other end.

This is a classic **queue**.

Using a normal `list`, removing the first customer is inefficient because every remaining element must be shifted one position to the left.

`deque` was designed specifically to solve this problem efficiently.

---

## What Does "Double-Ended" Mean?

A normal queue allows operations at one end for insertion and the other for removal.

A `deque` allows operations at **both** ends.

```
Left End                    Right End

← Append Left        Append →

← Pop Left              Pop →
```

This flexibility makes `deque` useful for implementing both queues and stacks.

# `list` vs `deque`

Although both can store ordered collections of data, they are optimized for different operations.

| Operation | `list` | `deque` |
|-----------|--------|---------|
| Append at end | ✅ Fast | ✅ Fast |
| Remove from end | ✅ Fast | ✅ Fast |
| Insert at beginning | ❌ Slow | ✅ Fast |
| Remove from beginning | ❌ Slow | ✅ Fast |
| Random indexing | ✅ Very Fast | ✅ Supported, but not its strength |

### Why is removing from the beginning slow for a `list`?

Consider:

```
Index

0   1   2   3

↓

A   B   C   D
```

Removing `A` results in:

```
B   C   D
```

Now Python must shift:

- `B`
- `C`
- `D`

one position to the left.

This shifting makes the operation slower as the list grows.

A `deque` avoids this shifting by using a different internal implementation.

# Internal Working

Conceptually, you can think of a `deque` as allowing efficient operations from both ends.

```
appendleft()

↓

[A] [B] [C]

↓

append()

↓

[A] [B] [C] [D]
```

Likewise,

```
popleft()

↓

[B] [C]

↓

pop()

↓

[B]
```

> **Note:** Internally, Python uses a more sophisticated implementation than a simple list. You don't need to know the implementation details to use `deque` effectively, but it's important to understand that it is optimized for efficient operations at both ends.

# Assignment vs Copying

One of the most common misconceptions is that assigning one variable to another creates a new object.

It does **not**.

Consider the following code:

```python
list_b = list_a
```

Many beginners think Python copies the list.

In reality, Python only copies the **reference** to the object.

Both variables now point to the **same object in memory**.

```
list_a
   │
   ▼
+----------------+
| [10, 20, 30]   |
+----------------+
   ▲
   │
list_b
```

Any modification made through one variable is visible through the other because there is only one underlying object.

````python
list_a = [10, 20, 30]

list_b = list_a

list_b.append(40)

print("list_a:", list_a)
print("list_b:", list_b)
````

Notice that `list_a` also changed.

This is because no copy was created. Both variables reference the same list.

# What is a Shallow Copy?

A **shallow copy** creates a **new outer object**, but it does **not** recursively copy the objects contained inside it.

In other words:

- The top-level container is new.
- Nested objects are still shared between the original and the copy.

This distinction becomes important when working with nested data structures.

````python
import copy

list_a = [10, 20, 30]

list_b = copy.copy(list_a)

print("list_a:", list_a)
print("list_b:", list_b)

print(id(list_a))
print(id(list_b))
````

````python
import copy

list_a = [10, 20, 30]

list_b = copy.copy(list_a)

list_b.append(40)

print(list_a)
print(list_b)
````

````python
import copy

marks = [
    [90, 85],
    [78, 88]
]

marks_copy = copy.copy(marks)

marks_copy[0].append(95)

print("Original:", marks)
print("Copy:", marks_copy)
````

# Why Isn't a Shallow Copy Always Enough?

In the previous lesson, we learned that a shallow copy creates:

- A new outer object ✅
- Shared inner objects ❌

This becomes a problem when your data contains nested mutable objects.

For example:

```
Student Records

↓

Students

↓

Subjects

↓

Marks
```

If you modify a nested object in the copied structure, the original data also changes.

Sometimes this is acceptable.

Often, it is not.

This is where **deep copying** becomes necessary.

# What is a Deep Copy?

A **deep copy** creates a completely independent copy of an object.

Unlike a shallow copy, it **recursively copies every nested object**.

This means:

- A new outer object is created.
- Every nested list, dictionary, set, or other mutable object is also copied.
- The copied structure shares **no mutable objects** with the original.

Changes made to the deep copy do **not** affect the original object.

````python
import copy
````

# Syntax

```python
copy.deepcopy(object)
```

`deepcopy()` traverses the entire object hierarchy and creates new copies of all nested mutable objects.

````python
import copy

marks = [
    [90, 85],
    [78, 88]
]

marks_copy = copy.deepcopy(marks)

print(marks)
print(marks_copy)
````

````python
import copy

marks = [
    [90, 85],
    [78, 88]
]

marks_copy = copy.deepcopy(marks)

marks_copy[0].append(95)

print("Original:", marks)
print("Copy:", marks_copy)
````

# Internal Working

A shallow copy behaves like this:

```
Original

↓

Outer List

↓

Shared Inner Lists
```

```
Original

┌──────────────┐
│   List       │
└──────────────┘
      │
      ▼
   [90, 85]

      ▲
      │
┌──────────────┐
│ Shallow Copy │
└──────────────┘
```

Both outer lists reference the same inner list.

---

A deep copy behaves differently:

```
Original

↓

Outer List

↓

New Inner Lists
```

```
Original

┌──────────────┐
│   List       │
└──────────────┘
      │
      ▼
   [90, 85]


Deep Copy

┌──────────────┐
│   List       │
└──────────────┘
      │
      ▼
   [90, 85]
```

The inner lists are completely independent.

````python
import copy

marks = [
    [90, 85],
    [78, 88]
]

marks_copy = copy.deepcopy(marks)

print(id(marks[0]))
print(id(marks_copy[0]))
````

# Comparison

| Feature | Shallow Copy | Deep Copy |
|----------|--------------|-----------|
| New outer object | ✅ | ✅ |
| New nested objects | ❌ | ✅ |
| Shares nested mutable objects | ✅ | ❌ |
| Independent copy | Partially | Completely |
| Speed | Faster | Slower |
| Memory usage | Lower | Higher |

The deeper and more complex the object, the greater the difference between the two approaches.

````python
import copy

original_dataset = [
    ["Ali", 25],
    ["Sara", 30]
]

working_dataset = copy.deepcopy(original_dataset)

working_dataset[0][1] = 26

print("Original:", original_dataset)
print("Working:", working_dataset)
````

# Performance Considerations

A deep copy recursively duplicates every nested object.

Because of this:

- It takes more time.
- It consumes more memory.

Use `deepcopy()` only when complete independence is required.

For simple or flat objects, a shallow copy is usually sufficient and more efficient.

# Choosing Between Assignment, Shallow Copy, and Deep Copy

| Situation | Recommended Approach |
|-----------|----------------------|
| Just another reference | Assignment (`=`) |
| Flat list or dictionary | `copy.copy()` |
| Nested mutable objects | `copy.deepcopy()` |
| Need a completely independent copy | `copy.deepcopy()` |

Ask yourself:

> **"Will I modify nested mutable objects?"**

If the answer is **yes**, consider using `deepcopy()`.

# Best Practices

- Use assignment (`=`) only when sharing the same object is intentional.
- Use `copy.copy()` for simple containers or when shared nested objects are acceptable.
- Use `copy.deepcopy()` for nested mutable data that must remain independent.
- Understand your data structure before choosing a copying strategy.
- Avoid unnecessary deep copies in performance-critical code.

# Common Mistakes

# `time.time()`

The `time()` function returns the **current Unix timestamp**.

## What is a Unix Timestamp?

A Unix timestamp is the number of seconds that have elapsed since:

```
January 1, 1970
00:00:00 UTC
```

Python returns this value as a floating-point number, where the decimal part represents fractions of a second.

## Syntax

```python

```

````python
import time

current_time = time.time()

print(current_time)
````

To measure how long a block of code takes:

1. Record the start time.
2. Execute the code.
3. Record the end time.
4. Subtract the two values.

```
Start

↓

Run Code

↓

End

↓

End − Start
```

````python
import time

start_time = time.time()

total = 0

for number in range(1_000_000):
  total += number

end_time = time.time()

execution_time = end_time - start_time

print(f"Execution Time: {execution_time:.6f} seconds")
````

The same approach works for individual functions.

This is a common practice when optimizing reusable code.

````python
import time

def calculate_sum():
    total = 0

    for number in range(1_000_000):
        total += number

start = time.time()

calculate_sum()

end = time.time()

print(f"Execution Time: {end - start:.6f} seconds")
````

Performance measurement is often used to compare different implementations of the same task.

The goal is not just to produce the correct result, but to determine **which implementation is more efficient**.

````python
import time

# Method 1
start = time.time()

numbers = []

for number in range(100_000):
    numbers.append(number)

end = time.time()

print(f"Append Method: {end - start:.6f} seconds")

# Method 2
start = time.time()

numbers = [number for number in range(100_000)]

end = time.time()

print(f"List Comprehension: {end - start:.6f} seconds")
````

Although `time.time()` is simple and widely used, it was designed to report the **current wall-clock time**.

Its precision and behavior can be affected by adjustments to the system clock (for example, if the operating system synchronizes with a time server).

For accurate performance measurements, Python provides a better tool:

`time.perf_counter()`

# `time.perf_counter()`

`perf_counter()` returns a **high-resolution timer** specifically designed for measuring short durations.

Compared with `time.time()`:

- Higher precision
- Better for benchmarking
- Recommended for performance testing

## Syntax

```python
time.perf_counter()
```

````python
import time

start = time.perf_counter()

total = sum(range(1_000_000))

end = time.perf_counter()

print(f"Execution Time: {end - start:.6f} seconds")
````

## Professional Recommendation

For benchmarking and measuring execution time, prefer:

```python
time.perf_counter()
```

instead of

```python
time.time()
```

unless you specifically need the current wall-clock time.

# `time.process_time()`

`process_time()` measures the amount of **CPU time** consumed by the current process.

Unlike `time.time()` and `time.perf_counter()`, it **does not include time spent waiting** (such as sleeping or waiting for input/output).

This makes it useful when you're interested only in how much CPU work your code performed.

## Syntax

```python
time.process_time()
```

````python
import time

start = time.process_time()

total = sum(range(1_000_000))

end = time.process_time()

print(f"CPU Time: {end - start:.6f} seconds")
````

# Choosing the Right Timer

| Function | Best Used For |
|-----------|---------------|
| `time.time()` | Current wall-clock time and simple timing |
| `time.perf_counter()` | Benchmarking and measuring execution time |
| `time.process_time()` | Measuring CPU time only |

### General Recommendation

For most performance measurements:

```python
time.perf_counter()
```

is the preferred choice.

# Best Practices

- Use `time.perf_counter()` for benchmarking.
- Measure the smallest code block that you want to optimize.
- Run benchmarks multiple times because execution times naturally vary.
- Compare implementations under similar conditions.
- Focus on optimizing code only after identifying real performance bottlenecks.

### Mistake 2

Optimizing code before verifying that it is actually a performance bottleneck.

Correctness should come before optimization.

---

### Mistake 3

Using `time.time()` for precise benchmarking when `time.perf_counter()` is more appropriate.

A module should have **one primary responsibility**.

For example:

```
email.py
```

Should contain:

- Send emails
- Email validation
- Email templates

It should **not** contain:

- Database functions
- Image processing
- Machine Learning code

Keeping modules focused follows the **Single Responsibility Principle (SRP)**, a widely accepted software design principle.

## What is a Circular Import?

A circular import occurs when two modules depend on each other during import.

Example:

```
employee.py

↓

imports payroll.py

↓

payroll.py

↓

imports employee.py
```

This creates a dependency loop that can cause import errors because Python may try to use a module before it has finished loading.

For example:

```
employee.py

```python
from payroll import calculate_salary
```

payroll.py

```python
from employee import Employee
```

Python cannot fully initialize one module before the other requests it.

### How to Avoid It

- Design modules with clear responsibilities.
- Avoid unnecessary dependencies between modules.
- Move shared functionality into a separate module (for example, `utils.py` or `common.py`) if multiple modules need it.

As a beginner, you don't need to master circular imports yet, but you should recognize why they occur.

## Best Practices

- Keep each module focused on one responsibility.
- Use descriptive module names.
- Avoid duplicate code by creating reusable modules.
- Keep related functions together.
- Import only what you need.
- Test modules independently when possible.
- Use the `if __name__ == "__main__":` pattern for module testing.
- Refactor large modules into smaller ones as projects grow.

# What is a Package?

A **package** is a collection of related Python modules organized inside a directory.

You already know that:

- A **module** is a single Python file (`.py`).
- A **package** is a directory that contains one or more related modules.

Think of it like this:

```
Module

↓

One Python file
```

```
Package

↓

Folder

↓

Multiple related modules
```

A package helps organize code into logical groups instead of placing every module in the same directory.

As projects become larger, simply creating more modules is not enough.

Imagine building an e-commerce application.

Without packages, your project might look like this:

```
project/

customer.py

order.py

payment.py

invoice.py

database.py

authentication.py

email.py

product.py

report.py

analytics.py

shipping.py

...
```

As the number of modules grows, the project becomes difficult to navigate.

Packages solve this problem by grouping related modules together.

Instead of:

```
project/

customer.py

employee.py

salary.py

attendance.py

department.py

...
```

You can organize them like this:

```
project/

employees/

    employee.py

    salary.py

    attendance.py

customers/

    customer.py

orders/

    order.py

payments/

    payment.py
```

Each folder represents a package.

Each package contains related modules.

# Module vs Package

| Module | Package |
|----------|----------|
| Single `.py` file | Directory containing related modules |
| Contains functions, classes, variables | Contains modules and sometimes subpackages |
| Smallest reusable unit | Higher-level organization |
| Example: `math.py` | Example: `numpy` |

Relationship:

```
Package

↓

Modules

↓

Functions

↓

Statements
```

A package is simply a higher level of organization.

# `__init__.py`

Traditionally, a package contains a special file named:

```
__init__.py
```

It is located inside the package directory.

Example:

```
employees/

│

├── __init__.py

├── employee.py

├── payroll.py

└── attendance.py
```

This file belongs to the package itself—not to any individual module.

Historically, `__init__.py` told Python:

> "This directory should be treated as a package."

Beginning with Python 3.3, Python introduced **namespace packages**, so a directory can act as a package even without an `__init__.py` file in many situations.

However, `__init__.py` is still widely used because it provides useful functionality beyond simply marking a package.

`__init__.py` is an ordinary Python file.

It can contain:

- Imports
- Variables
- Constants
- Functions
- Initialization code

For example:

```python
print("Employees package loaded.")
```

or

```python
company_name = "ABC Analytics"
```

When the package is imported, the code inside `__init__.py` executes.

One common use of `__init__.py` is to make selected objects available directly from the package.

Example:

```
employees/

├── __init__.py

├── payroll.py
```

`payroll.py`

```python
def calculate_salary():
    ...
```

`__init__.py`

```python
from .payroll import calculate_salary
```

Now users can write:

```python
from employees import calculate_salary
```

instead of:

```python
from employees.payroll import calculate_salary
```

This creates a cleaner public interface for the package.

## Best Practice

Although modern Python allows packages without `__init__.py` in many cases, most professional projects still include it because it:

- Clearly indicates that the directory is intended to be a package.
- Allows package-level initialization.
- Lets you control what is exposed when the package is imported.
- Improves compatibility with some tools and older codebases.

For these reasons, you'll commonly see an `__init__.py` file in production projects.

# Best Practices

- Group related modules into the same package.
- Choose descriptive package names.
- Keep packages focused on a single area of responsibility.
- Include an `__init__.py` file unless you have a specific reason not to.
- Keep `__init__.py` lightweight. Avoid placing complex logic inside it.
