# 1. Boolean Logic & Conditions (Core — Must Master)

Boolean logic is the **foundation of decision-making in Python**.

## 1.1 Boolean Values

### Definition
A Boolean is a data type that represents one of two values:
- True
- False

These are built-in constants in Python.

### Key Idea
Every condition in Python evaluates to either:
- **True** → condition satisfied  
- **False** → condition not satisfied

````python
x = 10

print(x > 5)
print(x < 5)
````

### Mental Model
Think of Booleans as answers to questions:
- Is 10 greater than 5? → True
- Is 10 less than 5? → False

---

## 1.2 Comparison Operators

Comparison operators are used to compare values and return Boolean results.

### Operators

| Operator | Meaning |
|----------|--------|
| == | Equal to |
| != | Not equal to |
| >  | Greater than |
| <  | Less than |
| >= | Greater than or equal to |
| <= | Less than or equal to |

### Important Note
- `=` is used for assignment  
- `==` is used for comparison

````python
a = 10
b = 20

print(a == b)   # False
print(a != b)   # True
print(a > 5)    # True
print(b <= 20)  # True
````

### Real-World Relevance
Used in:
- Filtering rows in datasets
- Comparing values during analysis
- Creating conditions for feature engineering

---

## 1.3 Chained Comparisons

Python allows writing comparisons in a clean and readable format.

### Syntax
Instead of writing:

````python
x = 15
print(10 < x and x < 20)
````

You can write:

````python
x = 12
print(10 < x < 20)
````

### Why This Matters
- Cleaner and more readable
- Common in real-world data filtering
- Reduces logical errors

---

## 1.4 Logical Operators

Logical operators combine multiple conditions.

### Operators

| Operator | Meaning |
|----------|--------|
| and | True if both conditions are True |
| or  | True if at least one condition is True |
| not | Reverses the condition |


````python
age = 25
income = 50000

print(age > 18 and income > 30000)
print(age < 18 or income > 30000)
print(not(age > 18))
````

### Short-Circuit Evaluation (Critical Concept)

Python evaluates conditions efficiently:

- In `and`: stops if first condition is False  
- In `or`: stops if first condition is True  

This prevents unnecessary computation and errors.

````python
x = 0

if x != 0 and (10 / x > 1):
  print("Valid")
````

### Why This Matters
- Prevents runtime errors (like division by zero)
- Improves performance
- Essential in real-world pipelines

## 1.5 Truthy & Falsy Values (Very Important)

Python automatically treats some values as False.

### Falsy Values

- False  
- None  
- 0, 0.0  
- "" (empty string)  
- [] (empty list)  
- {} (empty dictionary)  
- set()  

Everything else is considered Truthy.


````python
data = []
if data:
  print("Has data")

else:
  print("No data")
````

### Why This Matters in Data Work

Instead of writing:

````python
if len(data) > 0:
  print("Has data")

else:
  print("No data")
````

````python
data =[1,2,3]
if len(data) > 0:
  print("Has Data")
````

### Benefits
- Cleaner code
- More Pythonic
- Faster and readable

---

## 1.6 Identity vs Equality (Important Distinction)

### Equality (==)
Checks if values are equal

You should write:



````python
a = [1, 2]
b = [1, 2]

print(a == b)   # True
````

### Identity (is)
Checks if both variables refer to the same object in memory

````python
print(a is b)
````

### Best Practice
Use `is` when checking against None

````python
x = None
if x is None:
  print("No value")
````

# 2. Conditional Statements (if / elif / else)

Conditional statements allow your program to **make decisions based on conditions**.  
They are the core of building **logic-driven systems**, especially in data analysis, where you often need to **categorize, filter, and transform data**.

---

## 2.1 Basic `if` Statement

### Definition
The `if` statement executes a block of code **only if a condition is True**.

### Syntax
``if condition:``

    # code runs if condition is True

### Example

````python
sales = 600

if sales > 500:
  print("High Sales")
````

### Key Insight
- The block runs **only when condition is True**
- If False → Python skips it

---

## 2.2 `if-else` Statement

### Definition
Adds an alternative block when the condition is False.

### Syntax

```python
if condition:
    # runs if True
else:
    # runs if False

### Example

````python
sales = 300

if sales > 500:
    print("High sales")
else:
    print("Low sales")
````

### Use Case
- Binary decisions (Yes/No, High/Low, Pass/Fail)

---

## 2.3 `if-elif-else` (Multi-Condition Logic)

### Definition
Used when you have **multiple conditions to evaluate**.

### Syntax
```python
if condition1:
    pass
elif condition2:
    pass
elif condition3:
    pass
else:
    pass

### Important Rule
- Python checks conditions **top to bottom**
- Stops at the **first True condition**

---

### Example (Categorization)

````python
marks = 75

if marks >=90:
  print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
````

### Real-World Insight
This is used in:
- Grading systems
- Customer segmentation
- Risk classification

---

## 2.4 Multi-Condition Logic (Combining Conditions)

You can combine conditions using logical operators.

---

### Example

````python
age = 30
income = 40000

if age > 18 and income > 30000:
    print("Eligible")
````

### Key Insight
- Use `and`, `or`, `not` to build complex conditions
- Keep conditions **readable and simple**

---

## 2.5 Nested Conditions (Use Carefully)

### Definition
A condition inside another condition.

---

### Example

````python
age = 65

if age >= 18:
  if age >= 60:
    print("Senior")
  else:
    print("Adult")
````

### Why Limited Use?

- Deep nesting reduces readability
- Hard to debug and maintain

---

### Better Approach (Refactored)

````python
age = 65

if age >= 60:
    print("Senior")
elif age >= 18:
    print("Adult")
````

### Best Practice
- Avoid deep nesting
- Prefer flat, readable logic

---

## 2.6 Ternary Expressions (Clean One-Liners)

### Definition
A concise way to write simple `if-else` conditions in one line.

---

### Syntax
``value_if_true if condition else value_if_false``

### Example

````python
sales = 700

label = 'High' if sales > 500 else "Low"
print(label)
````

### When to Use
- Simple conditions
- Assigning values

### When NOT to Use
- Complex logic
- Multiple conditions

# 3.5 Loop Patterns (VERY IMPORTANT FOR DATA WORK)

---

# Pattern 1: Accumulator Pattern

## Definition
Used to calculate totals, sums, or averages.


````python
numbers = [10, 20, 30, 40]

total = 0

for num in numbers:
    total += num

print(total)
````

---

# Pattern 2: Counting Pattern

## Definition
Used to count occurrences of a condition.


````python
numbers = [1, 2, 2, 3, 3, 3]

count = 0

for x in numbers:
  if x == 3:
    count += 1

print(count)
````

---

# Pattern 3: Filtering Pattern

## Definition
Used to select specific elements from data.


````python
numbers = [10, 15, 20, 25, 30]

filtered = []

for num in numbers:
    if num > 20:
        filtered.append(num)

print(filtered)
````

---

# Pattern 4: Searching Pattern

## Definition
Used to check if a value exists.


````python
numbers = [5, 10, 15, 20]

target = 15
found = False

for num in numbers:
  if num == target:
    found = True
    break

print(found)
````

# 4. Loop Control Statements

Loop control statements allow you to **change the normal flow of loops**.  
Instead of simply repeating code, you can **stop, skip, or handle special conditions efficiently**.

These are extremely important in:
- Data cleaning
- Searching in datasets
- Machine learning training loops
- Error handling in iteration

---

# 4.1 `break` Statement

## Definition
`break` is used to **terminate a loop completely**.

Once `break` is executed:
- The loop stops immediately
- Control moves outside the loop

---

## Example

````python
for i in range(10):
    if i == 5:
        break
    print(i)
````

---

## Key Insight
- Stops entire loop
- Used when target is found
- Saves computation time

---

## Real-World Example (Search)


````python
numbers = [2, 4, 6, 8, 10]

for num in numbers:
  if num == 6:
    print("Found")
    break
````

---

# 4.2 `continue` Statement

## Definition
`continue` is used to **skip the current iteration** and move to the next one.

---

## Example

````python
for i in range(5):
    if i == 2:
        continue
    print(i)
````

---

## Key Insight
- Skips only one iteration
- Loop continues normally afterward

---

## Real-World Example (Skipping Bad Data)


````python
data = [10, None, 20, None, 30]

for value in data:
  if value == None:
    continue
  print(value)
````

---

# 4.3 `pass` Statement

## Definition
`pass` is a **do-nothing placeholder** in Python.

It is used when:
- Syntax requires a block
- You haven’t written logic yet

---

## Example

````python
for i in range(5):
  if i == 2:
    pass
  print(i)
````

---

## Key Insight
- Does nothing
- Used for structure planning
- Useful during development

---

## Real-World Usage


````python
for item in data:
  if item is None:
    pass # logic will be added later
````

---

# 4.4 `else` with Loops (VERY IMPORTANT)

## Definition
The `else` block runs **only if the loop completes normally (no break)**.

---

## Syntax
```python
for item in sequence:
    if condition:
        break
else:
    # runs if loop was NOT broken

---

## Example (Search Logic)


````python
numbers = [1, 3, 5, 7]

target = 4

for num in numbers:
    if num == target:
        print("Found")
        break
else:
    print("Not Found")
````

---

## Key Insight
- `else` runs only if **no break occurs**
- Very useful for search problems

---

## Real-World Analogy
Think of it like:
- "I searched everything and did NOT find it"

---

---

# 5. Comprehensions (Very Important — Must Master)

Comprehensions are a **Pythonic way to write loops in a compact, readable, and efficient form**.

They are heavily used in:
- Data analysis
- Data cleaning
- Feature engineering
- Machine learning pipelines
- Technical interviews

Instead of writing multi-line loops, comprehensions let you express logic in a **single line**.

---

# 5.1 List Comprehension

## Definition
A list comprehension is a **short way to create lists using a loop and optional condition**.

---

## Syntax
``[expression for item in iterable]``

---

## Example (Basic)


````python
numbers = [1, 2, 3, 4, 5]

squares = [n**2 for n in numbers]
squares
````

---

## Equivalent Loop Version


````python
numbers = [1, 2, 3, 4, 5]

squares = []
for n in numbers:
    squares.append(n**2)

print(squares)
````

---

## Key Insight
- Same logic
- But **cleaner and faster**
- Widely used in data science

---

# 5.2 Conditional List Comprehension

## Definition
Adds filtering logic inside comprehension.

---

## Syntax
``[expression for item in iterable if condition]``

````python
numbers = [1, 2, 3, 4, 5, 6]

evens = [x for x in numbers if x % 2 == 0]
print(evens)
````

---

## Key Insight
- Combines loop + condition in one line
- Very common in data filtering

---

# 5.3 Dictionary Comprehension

## Definition
Used to create dictionaries in a compact way.

---

## Syntax
``{key_expression: value_expression for item in iterable}``

---

## Example


````python
numbers = [1, 2, 3, 4]

squares_dict = {x:x**2 for x in numbers}
squares_dict
````

---

## Real-World Use Case
- Creating feature mappings
- Encoding data
- Fast transformations

---

# 5.4 Set Comprehension

## Definition
Creates a set using comprehension syntax.

---

## Syntax
``{expression for item in iterable}``


---

## Example


````python
from enum import UNIQUE
numbers = [1, 2, 2, 3, 4, 4]

unique_numbers = {n**2 for n in numbers}
unique_numbers
````

---

## Key Insight
- Automatically removes duplicates
- Useful for unique feature extraction

---

# 5.5 Conditional Expressions in Comprehensions

## Definition
Allows inline decision-making inside comprehension.

---

## Syntax
``[true_value if condition else false_value for item in iterable]``

---

## Example (Labeling Data)


````python
numbers = [10, 15, 20, 25]
labels = ["High" if x > 15 else "Low" for x in numbers]
labels
````

---

## Key Insight
- Enables classification in one line
- Very common in feature engineering

---

# 5.6 Nested Comprehensions (Advanced)

## Definition
Comprehension inside another loop (matrix-like structures).

---

## Example


````python
matrix = [[1, 2], [3, 4], [5, 6]]

flat = [num for row in matrix for num in row]
flat
````

---

## Key Insight
- Used for flattening data
- Powerful but should be used carefully for readability

# 7. Short-Circuit Logic (High Value)

Short-circuit logic is a powerful concept in Python where **logical expressions stop evaluating as soon as the result is determined**.

This leads to:
- Faster execution
- Safer code
- Cleaner conditions

---

# 7.1 What is Short-Circuiting?

## Definition
Short-circuiting means:
> Python **does not evaluate the full expression** if the result is already known.

---

## Key Idea

- In `and` → stops early if condition becomes False  
- In `or` → stops early if condition becomes True  

---

# 7.2 Short-Circuit with `and`

## Rule
If the **first condition is False**, Python does NOT evaluate the second.

---

## Example

````python
x = 0

if x != 0 and (10 / x > 1):
    print("Valid")
````

---

## Explanation

- `x != 0` → False  
- Python stops immediately  
- `(10 / x)` is NEVER executed  

---

## Why This Matters
- Prevents runtime errors (like division by zero)
- Makes code safer

---

# 7.3 Short-Circuit with `or`

## Rule
If the **first condition is True**, Python does NOT evaluate the second.

---

## Example

````python
x = 10

if x > 5 or (10 / 0):
    print("Condition met")
````

---

## Explanation

- `x > 5` → True  
- Python stops evaluation  
- `(10 / 0)` is NEVER executed  

---

## Key Insight
- Saves computation
- Avoids unnecessary operations

---

# 7.4 Short-Circuit Return Values (Advanced Insight)

Logical operators return values, not just True/False.

---

## Example

````python
a = 0
b = 10

print(a or b)   # 10
print(b and a)  # 0
````

---

## Rule

- `and` → returns first False OR last value  
- `or` → returns first True OR last value  

---

## Real Use Case


````python
user_input = ""

name = user_input or "Default Name"
print(name)
````

---

## Insight
- Very useful for setting default values
- Common in data preprocessing

---

# 7.5 Writing Efficient Conditions

## Bad Practice
```python
if len(data) > 0 and process(data):
    pass

---

## Better Practice
```python
if data and process(data):
    pass

---

## Why Better?
- Uses truthy/falsy
- Cleaner and faster
- Avoids unnecessary function calls

---

# 7.6 Real-World Applications

---

## 1. Preventing Errors

````python
x = 0

if x != 0 and 10 / x > 1:
    print("Safe")
````

##2. Default Values

````python
user_name = None
name = user_name or "Guest"
print(name)
````

##3. Data Pipeline Optimization

````python
data = []

if data and len(data) > 5:
    print("Process data")
````

##7.7 Common Mistakes
- Not understanding evaluation order
- Writing risky expressions (division before checking)
- Ignoring short-circuit benefits
##7.8 Best Practices
- Always check risky conditions first
- Use short-circuiting for safety
- Keep conditions readable
- Use it for default values
##Summary
- `and` stops when False is found
- `or` stops when True is found
- Prevents unnecessary computation
- Helps avoid runtime errors
- Widely used in data pipelines and real-world logic

# 9.1 Guard Clauses (Early Exit Pattern)

## Definition
A guard clause is a condition that **exits a function early** if something is wrong or unnecessary.

---

## Problem (Without Guard Clause)


````python
def process_data(data):
    if data is not None:
        if len(data) > 0:
            print("Processing data")
````

---

## Improved (With Guard Clause)


````python
def process_data(data):
    if data is None:
        return
    if len(data) == 0:
        return

    print("Processing data")
````

---

## Key Insight
- Handles edge cases first
- Reduces nesting
- Improves readability

---

## Real-World Example (Data Pipeline)


````python
def clean_data(data):
    if not data:
        return []

    cleaned = []

    for value in data:
        if value is None:
            continue
        cleaned.append(value)

    return cleaned
````

````python
values = [12,None,12,34,45]
value = []
print(clean_data(values))
print(clean_data(value))
````

##9.2 Sentinel Loops
###Definition

A sentinel loop continues until a special value (sentinel) is encountered.

````python
data = [10, 20, 30, -1, 50]

for value in data:
    if value == -1:  # sentinel value
        break
    print(value)
````

---

## Key Insight
- Sentinel = signal to stop
- Common in data streams and input processing

---

## While Loop Version


````python
while True:
    user_input = input("Enter number (-1 to stop): ")

    if user_input == "-1":
        break

    print("You entered:", user_input)
````

##9.3 Flag Variables
###Definition

A flag variable is used to track a condition (True/False) during execution.

````python
numbers = [1, 3, 5, 7]

found = False

for num in numbers:
    if num == 5:
        found = True
        break

if found:
    print("Found")
else:
    print("Not Found")
````

---

## Key Insight
- Helps track state
- Useful when logic continues after loop

---

# 9.4 When to Use Each Pattern

| Pattern | Use Case |
|--------|--------|
| Guard Clause | Input validation, early exit |
| Sentinel Loop | Unknown-length input, stopping condition |
| Flag Variable | Tracking search or condition |

---

# 9.5 Real-World Applications

---

## 1. Clean Data Pipeline


````python
def process_pipeline(data):
    if not data:
        return []

    result = []

    for value in data:
        if value < 0:
            continue
        result.append(value * 2)

    return result
````

````python
values = [2,4,6]
print(process_pipeline(values))
````

##2. Searching in Dataset

````python
users = ["Ali", "Ahmed", "Sara"]

target = "Omar"
found = False

for user in users:
    if user == target:
        found = True
        break

if not found:
    print("User not found")
````

##3. Input Processing System

````python
while True:
    data = input("Enter value (stop to exit): ")

    if data == "stop":
        break

    if not data:
        continue

    print("Processed:", data)
````
