# 1. Excel Fundamentals

## What Excel Files Are

Excel files are structured data files used to store information in a tabular format. They are widely used in business environments for organizing, analyzing, and reporting data.

Excel files can contain:
- structured tables
- calculations
- multiple sheets
- formatted reports

They are one of the most common formats for business data exchange.

---

## `.xlsx` vs `.csv`

Excel data is commonly stored in two formats.

### `.xlsx` (Excel Workbook)
- Modern Excel format
- Supports multiple sheets
- Supports formatting (colors, fonts, charts)
- Supports formulas and calculations

### `.csv` (Comma-Separated Values)
- Plain text format
- Stores data in simple rows and columns
- No formatting or formulas
- Faster to process in data analysis

### Key Difference Table

| Feature | `.xlsx` | `.csv` |
|---|---|---|
| Sheets | Multiple | Single |
| Formatting | Yes | No |
| Formulas | Yes | No |
| File Type | Binary | Plain text |

---

## Worksheets and Workbooks

### Workbook
A workbook is the complete Excel file.

It can contain:
- one or more sheets
- full dataset collection

Example:
- `sales_data.xlsx`

---

### Worksheet
A worksheet is a single sheet inside a workbook.

Example:
- January sheet
- February sheet
- March sheet

Each worksheet contains tabular data.

---

### Structure View

| Level | Description |
|---|---|
| Workbook | Entire Excel file |
| Worksheet | Single sheet inside file |

---

## Rows, Columns, and Cells

Excel data is arranged in a grid structure.

### Columns
- Vertical structure
- Labeled as A, B, C, etc.
- Represent features/variables

### Rows
- Horizontal structure
- Numbered as 1, 2, 3, etc.
- Represent records/observations

### Cells
- Intersection of row and column
- Example: A1, B3, C5
- Holds a single value

---

### Example Table Structure

| A (Name) | B (Age) | C (Salary) |
|---|---|---|
| Ali | 25 | 50000 |
| Sara | 28 | 70000 |

---

## Why Excel is Heavily Used in Business

Excel is widely used because it is simple, flexible, and powerful for non-programmers.

### Main Reasons

- Easy to learn and use
- Works well for structured data
- Supports calculations and formulas
- Suitable for reporting and dashboards

# 2. Reading Excel Files Using Pandas

In this topic, we will learn how to professionally read Excel workbooks using Pandas. First we have to create a real multi-sheet workbook (`company_data.xlsx`), we will use it like a real data analyst or freelancer.

This is one of the most important skills in data analysis because almost every real project starts with reading data correctly.

We will focus on:
- loading Excel files into Pandas
- working with multiple sheets
- selecting specific data
- handling missing values during import
- controlling how data is read

---



```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Hina"],
    "Department": ["IT", "HR", "Finance", "IT", "Marketing", "HR"],
    "Age": [28, 32, None, 26, 29, 35],
    "Salary": [70000, 65000, 80000, 60000, None, 72000],
    "Experience_Years": [5, 7, 8, 3, 4, 10]
})
```

```python
performance = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Performance_Score": [8, 9, 7, None, 6, 10],
    "Projects_Completed": [12, 15, 10, 8, 9, 20],
    "Overtime_Hours": [5, 2, 7, 3, None, 1]
})
```

```python
departments = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "Marketing"],
    "Head": ["Bilal", "Hina", "Zara", "Fahad"],
    "Budget": [500000, 300000, 400000, 350000],
    "Employees_Count": [2, 2, 1, 1]
})
```

```python
employees, performance, departments
```

```python
with pd.ExcelWriter("company_data.xlsx") as writer:
    employees.to_excel(writer, sheet_name="Employees", index=False)
    performance.to_excel(writer, sheet_name="Performance", index=False)
    departments.to_excel(writer, sheet_name="Departments", index=False)
```

# 1. Reading an Excel File (Basic)

The simplest way to load an Excel file is using:

## `pd.read_excel()`

By default:
- Pandas reads the FIRST sheet only
- loads everything into a DataFrame

---



```python
df = pd.read_excel('company_data.xlsx')
df
```

# Understanding What Happens Here

When you run `pd.read_excel()`:

- The Excel file is converted into a DataFrame
- Only the first sheet is loaded automatically
- Column headers are taken from the first row
- Missing values are converted to NaN automatically

---

# Important Insight

If your file has multiple sheets (like ours):
- Employees
- Performance
- Departments

Pandas will NOT load all of them by default

# 2. Reading a Specific Sheet

Excel workbooks often contain multiple sheets.

In our case:
- Employees
- Performance
- Departments

You can choose a sheet using `sheet_name`.

---

## Why This Is Important

In real projects:
- HR data is in one sheet
- Sales data in another
- Financial data in another

So selecting correct sheet is critical.

```python
df = pd.read_excel('company_data.xlsx', sheet_name='Employees')
df
```

# Alternative: Using Sheet Index

Instead of name, you can use index:

| Sheet | Index |
|---|---|
| Employees | 0 |
| Performance | 1 |
| Departments | 2 |

---

## Why index is useful

- Faster when sheet names are unknown
- Useful for automation scripts

```python
df = pd.read_excel('company_data.xlsx', sheet_name= 2)
df
```

# 3. Viewing All Sheet Names

Before working with a file, professionals first check sheet names.

This helps you:
- understand file structure
- avoid mistakes
- choose correct data source

---

## Method: ExcelFile object

```python
excel_file = pd.ExcelFile('company_data.xlsx')
excel_file.sheet_names
```

# 4. Selecting Specific Columns (usecols)

Real datasets are large, but you often only need specific columns.

---

## Why this is important

- reduces memory usage
- speeds up processing
- avoids unnecessary data
- very common in freelancing tasks

```python
df = pd.read_excel('company_data.xlsx', usecols= ['Head', 'Budget'], sheet_name = 2)
df
```

# 5. Handling Headers

By default:
- first row is treated as header

But real files may have:
- extra titles
- merged rows
- metadata at top

So you may need to control header position.

```python
df_header = pd.read_excel('company_data.xlsx', header= 0, sheet_name= 'Employees')
df_header
```

# 6. Skipping Rows (skiprows)

Sometimes Excel files contain:
- titles
- empty rows
- notes
- company logos data rows

So we skip unnecessary rows.

```python
df_skip = pd.read_excel(
    "company_data.xlsx",
    sheet_name="Employees",
    skiprows = 2,
    header = 0
)

df_skip
```

# 7. Handling Missing Values During Reading

In real datasets, missing values are very common.

Pandas automatically converts:
- empty cells → NaN

But sometimes data contains:
- "NA"
- "Missing"
- "--"

We must define them manually.

```python
df_missing = pd.read_excel(
    "company_data.xlsx",
    sheet_name="Employees",
    na_values=["NA", "Missing", "--"]
)

df_missing
```

# 8. Reading All Sheets at Once

Sometimes you need complete workbook data.

Instead of reading one sheet at a time, you can load all sheets.

## Output Type

This returns a dictionary:
- key = sheet name
- value = DataFrame

```python
all_sheets = pd.read_excel('company_data.xlsx', sheet_name= None)
all_sheets
```

# How to Access Individual Sheets from Dictionary

After loading all sheets:

You can access them like this:

- Employees
- Performance
- Departments

This is very useful in automation tasks.

```python
all_sheets['Employees']
```

```python
all_sheets['Performance']
```

# 1. Handling Missing Values

Missing values are one of the most common problems in datasets.

Examples:
- empty salary
- missing age
- missing department
- incomplete dates

Pandas represents missing values as:

## `NaN`

---

# Why Missing Values Matter

Missing values can:
- break calculations
- reduce model accuracy
- create wrong summaries
- cause analysis errors

```python
import pandas as pd

df = pd.read_excel("company_data.xlsx", sheet_name="Employees")

df
```

# Detecting Missing Values

Use:

## `isnull()`

to identify missing data.

```python
df.isnull()
```

# Counting Missing Values

To see total missing values in each column:

```python
df.isnull().sum()
```

# Removing Missing Values

Use:

## `dropna()`

This removes rows containing missing values.

---

# Important Note

This is useful when:
- missing data is small
- incomplete rows are not important

But excessive removal can lose valuable data.

```python
df_clean = df.dropna()

df_clean
```

# Filling Missing Values

Instead of removing rows, we often fill missing data.

This is more professional in many cases.

Use:

## `fillna()`

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df
```

# Professional Insight

Common filling strategies:

| Column Type | Common Fill Method |
|---|---|
| Numeric | mean / median |
| Text | mode / Unknown |
| Dates | previous date or estimated |

---

# Example

- missing salary → average salary
- missing department → "Unknown"

# 2. Handling Duplicate Rows

Duplicate data is another common real-world issue.

Duplicates happen because:
- repeated entries
- system errors
- manual mistakes
- merged files

---

# Why Duplicates Are Dangerous

Duplicates can:
- inflate reports
- distort averages
- create wrong statistics
- affect ML models

```python
df_duplicates = pd.concat([df, df])

df_duplicates
```

# Detecting Duplicate Rows

Use:

## `duplicated()`

```python
df_duplicates.duplicated()
```

# Removing Duplicate Rows

Use:

## `drop_duplicates()`

```python
df_no_duplicates = df_duplicates.drop_duplicates()

df_no_duplicates
```

# 3. Handling Extra Spaces

Extra spaces are extremely common in Excel files.

Example:
- `" IT"`
- `"HR "`
- `" Finance "`

These spaces create analysis problems because:
- `"IT"` and `" IT"` are treated differently

This causes:
- grouping issues
- filtering problems
- incorrect counts

```python
df["Department"] = df["Department"].str.strip()

df
```

# Understanding `str.strip()`

`strip()` removes:
- leading spaces
- trailing spaces

Example:

| Before | After |
|---|---|
| `" IT"` | `"IT"` |
| `"HR "` | `"HR"` |

---

# Professional Insight

Always clean text columns before:
- grouping
- filtering
- merging datasets

# 4. Handling Inconsistent Column Names

Real datasets often contain bad column names like:

- `" Employee Name "`
- `"salary"`
- `"Joining Date"`
- `"DEPARTMENT"`

Inconsistent columns create coding confusion.

Professionals standardize column names immediately.

```python
df.columns
```

# Standardizing Column Names

A professional workflow is:
- lowercase names
- remove spaces
- replace spaces with underscores

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

df.columns
```

# Result Example

| Before | After |
|---|---|
| Employee ID | employee_id |
| Joining Date | joining_date |

---

# Why This Matters

Consistent columns:
- improve readability
- reduce errors
- make automation easier
- improve team collaboration

# 5. Handling Broken Formatting Issues

Real Excel files may contain:
- mixed data types
- wrong date formats
- numbers stored as text
- inconsistent capitalization

Example:
- `"it"`
- `"IT"`
- `"It"`

All represent same department but are treated differently.

```python
df["department"] = df["department"].str.upper()

df
```

# Why Standardization Is Important

Consistent formatting ensures:
- correct grouping
- correct filtering
- accurate reports
- reliable ML preprocessing

---

# Common Standardization Methods

| Method | Purpose |
|---|---|
| `str.upper()` | Convert to uppercase |
| `str.lower()` | Convert to lowercase |
| `str.title()` | Capitalize words |
