# 3. File Modes in Python

File modes define **how a file is opened and how Python is allowed to interact with it**.

Choosing the correct mode is critical in real-world work because:
- Wrong mode can delete data
- Wrong mode can cause errors
- Correct mode ensures safe and controlled file operations

# 3.1 Read Mode (`r`)

## Concept
Read mode is used when you only want to **read data from an existing file**.

### Key Rules:
- File must already exist
- You cannot modify data
- Cursor starts at the beginning of file

## Use Cases:
- Reading datasets (CSV, TXT, JSON)
- Loading logs
- Viewing stored data

````python
with open('example.txt', 'r') as file:
  content = file.read()

print(content)
````

# 3.2 Write Mode (`w`)

## Concept
Write mode is used to **create or overwrite a file**.

### Key Rules:
- Creates file if it does not exist
- Deletes existing content if file exists
- Starts writing from beginning

## Use Cases:
- Saving reports
- Exporting processed data
- Creating output files

````python
with open('output.txt', 'w') as file:
  file.write('This is new data written in the file')
````

````python
with open('output.txt', 'r') as file:
  content = file.read()
print(content)
````

# 3.3 Append Mode (`a`)

## Concept
Append mode is used to **add new data at the end of an existing file**.

### Key Rules:
- Does NOT delete existing data
- Adds new content at the end
- Creates file if it does not exist

## Use Cases:
- Logging systems
- Tracking events
- Adding new records continuously

````python
with open('output.txt', 'a') as file:
  file.write('\n I append this data')
````

````python
with open('output.txt', 'r') as file:
  content = file.read()
print(content)
````

# 3.4 Exclusive Creation Mode (`x`)

## Concept
Exclusive creation mode is used to **create a new file only if it does NOT already exist**.

### Key Rules:
- Creates a new file
- Fails if file already exists
- Used for safety against overwriting

## Use Cases:
- Creating unique files
- Preventing accidental overwrites
- Safe file initialization

````python
with open("unique_file.txt", "x") as file:
    file.write("This file is created only if it does not exist.")
````

Reading files is one of the MOST important skills in data analysis and freelancing.

Python provides multiple ways to read files depending on the use case.

# 4.1 `read()` Method

## Concept

`read()` is used to read the **entire file at once** as a single string.

## When to use:
- small files
- quick inspection
- simple text processing

````python
with open("demo.txt", "r") as file:
    content = file.read()

print(content)
````

## Important Behavior of `read()`

- Reads full file in one go
- Returns a single string
- Cursor moves to end of file after reading

## Problem with large files:
- Can consume high memory
- Not suitable for big datasets (CSV, logs)

---

## Real-world use:
- configuration files
- small text files
- quick debugging

# 4.2 `readline()` Method

## Concept

`readline()` reads **one line at a time** from the file.

## When to use:
- structured text files
- line-by-line processing
- logs

````python
with open('demo.txt', 'w') as file:
  file.write('I was looking for you\n')
  file.write('Now you are looking for me\n')
  file.write('Is not it is funny!')
  file.write(' Checked')
````

````python
with open('demo.txt', 'r') as file:
  line1 = file.readline()
  line2 = file.readline()
  line3 = file.readline()
  line4 = file.readline()
print(line1)
print(line2)
print(line3)
print(line4)
````

## Important Behavior of `readline()`

- Reads only one line per call
- Keeps cursor at next line
- Useful for step-by-step processing

## Real-world use:
- log file analysis
- reading large files safely
- streaming data processing

# 4.3 `readlines()` Method

## Concept

`readlines()` reads **all lines and returns them as a list**.

Each line becomes one element in the list.

````python
with open('demo.txt', 'r') as file:
  lines = file.readlines()

print(lines)
````

## Important Behavior of `readlines()`

- Returns a list of strings
- Each line includes `\n`
- Loads entire file into memory

## Example Output:
```
["Hello\n", "World\n"]
```

## Real-world use:
- small dataset processing
- splitting structured text

# 4.4 Iterating Through Files (MOST IMPORTANT METHOD)

## Concept

Instead of reading everything at once, Python allows you to loop through a file line by line.

This is the MOST efficient and professional method.

````python
with open('demo.txt', 'r') as file:
  for line in file:
    print(line)
````

## Why this method is powerful

- Memory efficient
- Works for large files
- Industry standard approach

## Used in:
- big data processing
- log analysis
- CSV preprocessing (before pandas)
- automation scripts

# 4.5 File Pointer Behavior (VERY IMPORTANT)

Understanding cursor movement is key to mastering file reading.

## Rules:
- `read()` → moves cursor to end
- `readline()` → moves line by line
- `readlines()` → moves to end
- loop iteration → moves line by line automatically

````python
with open("demo.txt", "r") as file:
    print(file.read())
    print(file.read())  # nothing (cursor already at end)
````

````python
with open("demo.txt", "r") as file:
    print(file.readline()) #Cursor shift to next line
    print(file.read())
````

# 4.6 Resetting File Pointer (`seek`)

Sometimes you need to read again.

You must reset cursor using `seek(0)`.

````python
with open("demo.txt", "r") as file:
    print(file.readline()) #Cursor shift to next line
    file.seek(0)
    print(file.read())
````

# 4.7 Best Practices for Reading Files

## Professional rules:

### 1. Prefer iteration for large files
```python
for line in file:
```

### 2. Avoid `read()` for big datasets

### 3. Use `with` always

### 4. Reset pointer only when necessary (`seek`)

# Summary of All Methods

| Method | Use Case | Memory |
|------|--------|--------|
| `read()` | small files | high usage |
| `readline()` | line-by-line control | medium |
| `readlines()` | list of lines | high usage |
| `for line in file` | best practice | low usage |

# 5.3 `writelines()` Method

## Concept

`writelines()` writes multiple lines at once using:
- lists
- tuples
- iterables

## Syntax

```python
file.writelines(iterable)
```

````python
lines = ["Python\n", "SQL\n", "Pandas", 'Numpy']

with open("skills.txt", "w") as file:
    file.writelines(lines)
````

````python
with open('skills.txt', 'r') as file:
  print(file.readlines())
````

# Important Behavior of `writelines()`

## Very Important:
`writelines()` does NOT add newlines automatically.

You must include:
```python
\n
```

yourself.

````python
lines = ["Python", "SQL", "Pandas"]

with open("skills.txt", "w") as file:
    file.writelines(lines)
````

````python
with open('skills.txt', 'r') as file:
  print(file.read())
````

# Output Problem

Output becomes:
```text

```

because newline characters were missing.

# Correct Professional Way

````python
lines = ["Python\n", "SQL\n", "Pandas\n"]

with open("skills.txt", "w") as file:
    file.writelines(lines)
````

# 5.5 File Pointer After Writing

After writing:
- cursor moves forward
- reading immediately may return nothing

This is a very important professional concept.

````python
with open('demo.txt', 'w+') as file:
  file.write("Hello")

  print(file.read())
````

# Why This Happens

After writing:
- cursor is at end of file

So Python has nothing left to read.

Solution:
```python
seek(0)
```
moves cursor back to start.

````python
with open("demo.txt", "w+") as file:
    file.write("Hello")

    file.seek(0)

    print(file.read())
````

# 5.6 Flushing Data (Professional Concept)

Python temporarily stores written data in memory before saving to disk.

This is called buffering.

Sometimes you may force immediate writing using:
```python
file.flush()
```

Usually `with` handles this automatically.

````python
with open("demo.txt", "w") as file:
    file.write("Buffered data")

    file.flush()
````

````python
with open("demo.txt", "r") as file:
    print(file.read())
````

Very common in real projects.

# 6.1 Problem Without `with`

Before understanding `with`, you must understand the problem it solves.

Normally we open and close files manually:

````python
file = open('demo.txt', 'r')
content = file.read()

print(content)

file.close()
````

# Problems with Manual Closing

## 1. Forgetting `close()`
This can leave files open.

## 2. Memory/resource leaks
Too many open files can waste system resources.

## 3. Errors before `close()`
If program crashes before `close()`, the file remains open.

````python
file = open("demo.txt", "r")

print(file.read())

#10 / 0 raise error

file.close()
````

# What Happened?

The program crashed before:
```python
file.close()
```

So the file was never properly closed.

This is exactly the problem `with` solves.

# 6.2 The `with` Statement

## Concept

The `with` statement automatically manages resources.

For files:
- opens file
- performs operations
- automatically closes file

even if errors occur.

````python
with open("demo.txt", "r") as file:
    content = file.read()

print(content)
````

# Syntax of `with`

```python
with open("file.txt", "mode") as variable:
```

## Components

| Part | Meaning |
|------|--------|
| `with` | starts context manager |
| `open()` | opens file |
| `as file` | creates file object |
| indented block | operations on file |

# 6.3 Automatic File Closing

This is the MOST important feature of `with`.

When block ends:
- Python automatically closes file

Even if:
- error occurs

## Concept

`tell()` returns the current cursor position inside the file.

## Syntax

```python
file.tell()
```

Returns:

````python
with open('demo.txt', 'r') as file:
  print(file.tell())

  print(file.read(5)) #Reading 5 characters

  print(file.tell())

  print(file.read())
````

because cursor moved forward.

# 9.3 `seek()` Method

## Concept

`seek()` moves the cursor to a specific position inside the file.

## Syntax

```python
file.seek(position)
```

````python
with open("demo.txt", "r") as file:
    file.seek(0)

    print(file.read())
````

# What `seek(0)` Means

```python
seek(0)
```

moves cursor to:
- beginning of file

# Example: Reading Again After `seek()`

````python
with open('demo.txt', 'r') as file:
  print(file.read())

  print(file.seek(2))

  print(file.read())
````

# Why This Works

Normally:
- first `read()` moves cursor to end

But:
```python
seek(2)
```

resets cursor, allowing rereading.
