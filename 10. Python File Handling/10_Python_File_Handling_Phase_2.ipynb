# 2.7 Delimiters in `read_csv()`

Not all CSV files use commas.

Some datasets use:
- semicolon `;`
- tab `\t`
- pipe `|`

# Example — Semicolon Delimiter

````python
data = """name;age;city
Ali;22;Lahore
Sara;25;Karachi
"""

with open("semicolon_data.csv", "w") as file:
    file.write(data)
````

````python
with open("semicolon_data.csv", "r") as file:
    print(file.read())
````

# Problem

If we read normally:
```python
pd.read_csv()
```

pandas assumes comma delimiter.

````python
df = pd.read_csv("semicolon_data.csv")

print(df)
````

# Correct Solution

Use:
```python
sep=";"
```

````python
df = pd.read_csv("semicolon_data.csv", sep = ';')
df
````

# Why This is VERY Important

Wrong delimiter is one of the MOST common real-world CSV issues.

# 4.2 Creating a Sample CSV File

We will create a file to practice reading and writing.

````python
data = """name,age,city
Ali,22,Lahore
Sara,25,Karachi
Ahmed,21,Islamabad
"""

with open("people.csv", "w") as file:
    file.write(data)
````

# 4.3 `csv.reader`

## Concept

`csv.reader` reads CSV files row by row.

Each row is returned as a list.

````python
with open("people.csv", "r") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
````

# Understanding Output

Each row becomes a list:

```python
['name', 'age', 'city']
['Ali', '22', 'Lahore']
['Sara', '25', 'Karachi']
```

So:
- each row = list of values
- everything is treated as text

# Why csv.reader is Useful

It is useful when:
- you want low-level control
- you are processing files manually
- you are building simple scripts

# 4.4 Skipping Header Row

Sometimes we don’t want the header row.

````python
with open('people.csv', 'r') as file:
  reader = csv.reader(file)

  next(reader) #skip row

  for row in reader:
    print(row)
````

# 4.5 `csv.writer`

## Concept

`csv.writer` is used to write data into CSV files.

````python
data = [
    ["name", "age", "city"],
    ["Ali", 22, "Lahore"],
    ["Sara", 25, "Karachi"],
    ["Ahmed", 21, "Islamabad"]
]

print(data)
````

````python
with open("output.csv", "w+", newline="") as file:
    writer = csv.writer(file)

    writer.writerows(data)
    file.seek(0)

    print(file.read())
````

# Understanding writerows()

```python
writer.writerows(data)
```

writes multiple rows at once.

# Single Row Writing

If we want to write one row:

````python
with open('output2.csv', 'w+', newline="") as file:
  writer = csv.writer(file)

  writer.writerow(['name','age','city'])
  writer.writerow(['Ali',22,'Lahore'])
  writer.writerow(['Sara',25,'Karachi'])
  file.seek(0)
  print(file.read())
````

# Why newline="" is Important

Without:
```python
newline=""
```

extra blank lines may appear in CSV files.

This is a common beginner mistake.

# 4.6 `csv.DictReader`

## Concept

`DictReader` reads CSV files as dictionaries instead of lists.

````python
with open("people.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
````

# Understanding Output

Each row becomes:

```python
{'name': 'Ali', 'age': '22', 'city': 'Lahore'}
```

So:
- keys = column names
- values = row data

# Why DictReader is Powerful

Advantages:
- more readable
- easier data access
- avoids index confusion

Example:
```python
row["name"]
```

instead of:
```python
row[0]
```

# 4.7 Practical Use Case

Example: printing only names

````python
with open("people.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row['name'])
````
