# 3. Writing JSON Files (`json.dump()`)

One of the most important operations is saving Python data into a JSON file.

We use:

## `json.dump()`

---

# Syntax

```python
json.dump(python_object, file_object)
```

---

# What Happens Here?

- Python dictionary → converted into JSON
- Data gets stored inside a `.json` file

---

# Example Workflow

```plaintext
Python Dictionary
        ↓
json.dump()
        ↓
JSON File
```

````python
employee = {
    "name": "Ali",
    "age": 25,
    "department": "IT",
    "salary": 70000
}

with open('employees.json', 'w') as file:
  json.dump(employee, file)
````

# Understanding the Code

| Part | Purpose |
|---|---|
| `"employee.json"` | JSON filename |
| `"w"` | Write mode |
| `json.dump()` | Save dictionary as JSON |

---

# Output File Content

The generated JSON file will contain:

```json
{
    "name": "Ali",
    "age": 25,
    "department": "IT",
    "salary": 70000
}
```

# 4. Making JSON More Readable (`indent`)

# Using `indent`

The `indent` parameter formats JSON neatly.

````python
import json

employee = {
    "name": "Sara",
    "age": 28,
    "department": "HR"
}

with open("formatted_employee.json", "w") as file:
    json.dump(employee, file, indent=4)
````

# 3. Converting JSON to Python Objects

One of the most important concepts in JSON handling is understanding how JSON data becomes normal Python objects after loading.

This topic is extremely important because:
- almost all JSON processing depends on Python structures
- APIs return JSON that becomes Python objects
- data extraction depends on understanding object types

If you do not understand Python object mapping properly, nested JSON becomes very confusing.

# Main Goal of This Topic

# 1. JSON Object → Python Dictionary

The MOST important mapping:

| JSON | Python |
|---|---|
| Object `{}` | Dictionary `dict` |

---

# Example JSON Object

```json
{
    "name": "Ali",
    "age": 25
}
```

After loading into Python:

```python
{
    "name": "Ali",
    "age": 25
}
```

becomes a Python dictionary.

````python
json_data = '''
{
    "name": "Ali",
    "age": 25
}
'''

data = json.loads(json_data)

print(data)
print(type(data))
````

# Understanding the Code

## `json.loads()`

Converts:
- JSON string
→ Python object

---

# Output Type

```python
<class 'dict'>
```

This means:
- the JSON object became a Python dictionary

---

# Important Note

There are two similar functions:

| Function | Purpose |
|---|---|
| `json.load()` | Read JSON file |
| `json.loads()` | Read JSON string |

We use `loads()` when JSON is already inside a Python string.

# 2. Accessing Dictionary Data

Since JSON objects become dictionaries:

We use:
- dictionary keys

to access values.

````python
print(data["name"])
print(data["age"])
````

# 3. JSON Array → Python List

Another very important mapping:

| JSON | Python |
|---|---|
| Array `[]` | List `list` |

---

# Example JSON Array

```json
[
    "Python",
    "SQL",
    "Pandas"
]
```

becomes:

```python
["Python", "SQL", "Pandas"]
```

````python
json_array = '''
[
  "Python",
  "SQL",
  "Pandas"
]
'''

skills = json.loads(json_array)
print(skills)
print(type(skills))
````

# Understanding the Output

```python
<class 'list'>
```

This means:
- JSON array became Python list

---

# Important Concept

After conversion:
- lists behave like normal Python lists
- you can use indexing
- loops
- list methods

````python
print(skills[0])
print(skills[1])
````

# 4. Nested JSON → Nested Python Objects

Real-world JSON usually contains combinations of:
- dictionaries
- lists
- nested dictionaries

After loading:
- JSON becomes nested Python structures

````python
nested_json = '''
{
    "employee": {
        "name": "Sara",
        "skills": ["Python", "SQL"],
        "address": {
            "city": "Lahore",
            "country": "Pakistan"
        }
    }
}
'''

employee_data = json.loads(nested_json)

print(employee_data)
````

# 4. Working with Nested JSON in Pandas

Real-world JSON is usually nested.

Example:
- user details inside objects
- address inside user
- lists inside dictionaries

Nested JSON must often be flattened before analysis.

````python
nested_data = {
    "employees": [
        {
            "name": "Ali",
            "department": "IT",
            "address": {
                "city": "Lahore",
                "country": "Pakistan"
            }
        },
        {
            "name": "Sara",
            "department": "HR",
            "address": {
                "city": "Karachi",
                "country": "Pakistan"
            }
        }
    ]
}

nested_data
````

# Understanding the Structure

```plaintext
employees
   ├── employee 1
   │      ├── name
   │      ├── department
   │      └── address
   │             ├── city
   │             └── country
   │
   └── employee 2
```

---

# Problem

Nested dictionaries are difficult to analyze directly in DataFrames.

We need to flatten them.

# 5. Flattening Nested JSON Using `json_normalize()`

Pandas provides:

## `pd.json_normalize()`

This converts nested JSON into flat table structure.

This is an extremely important professional skill.

````python
flat_df = pd.json_normalize(nested_data['employees'])
flat_df
````
