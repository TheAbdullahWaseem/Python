# What is a Path?

A path simply means:

```plaintext
The location of a file or folder
```

Example:

```plaintext
/content/data.csv
```

This path tells Python:
- where the file exists

---

# Real-World Analogy

Think of a path like:
- a home address



# What Does `Path()` Do?

When we write:

```python
Path("/content")
```

we are creating a:

# Path Object

This object represents:
- a file location
or
- a folder location

---

# Important Concept

`Path()` DOES NOT:
- create a folder
- create a file
- move anything

It only REPRESENTS a path.

---

# Example

```python
folder = Path("/content")
```

means:

```plaintext
"This variable represents the /content folder"
```

---

# Why Not Just Use Strings?

Old style:

```python
"/content/data.csv"
```

This is only a normal string.

It has NO special file methods.

---

# But `Path()` Creates Smart Objects

Example:

```python
file = Path("data.csv")
```

Now this object can:
- check existence
- iterate folders
- search files
- join paths
- get filename
- get extension

This is why `pathlib` is powerful.

---

# Visual Difference

# 4. Saving Cleaned Files Automatically

Now we save cleaned datasets.

Professional workflows rarely overwrite original files.

Usually:

```plaintext

        ↓
```

````python
clean_folder = Path('Cleaned_Files')

if not clean_folder.exists():
  clean_folder.mkdir()

for file in Path('/content').glob('*.csv'):
  df = pd.read_csv(file)
  df.columns=df.columns.str.strip()

  df["product"]=df["product"].str.strip()

  df.dropna(inplace=True)

  output_file = clean_folder/f"cleaned_{file.name}"

  df.to_csv(output_file, index= False)
````
