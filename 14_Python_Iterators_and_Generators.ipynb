## Imagine There Were No Iterators or Generators

Let's imagine that iterators and generators never existed.

Suppose you are working as a Data Analyst for an online shopping company. Every night, the company generates a sales report containing millions of transactions.

Your task is to calculate the total sales, identify the highest-selling products, and prepare a summary report.

How would your program process this data?

A straightforward approach would be:

1. Read the entire file.
2. Store every record in memory.
3. Process the data.

For a small dataset, this approach works perfectly.

However, software is rarely built only for small datasets. Professional applications must continue to work even as the amount of data grows from thousands of records to millions or even billions of records.

As datasets grow, this approach becomes increasingly inefficient because the program attempts to keep all the data in memory at the same time.

````python
sales = [1250, 980, 1500, 760, 890]

total_sales = 0

for sale in sales:
  total_sales += sale

print(f"Total Sales: {total_sales}")
````

For five numbers, there is obviously no problem.

Even for hundreds or thousands of values, modern computers can usually handle the data without difficulty.

The challenge begins when the amount of data becomes extremely large.

Consider some real-world examples:

- A CSV file containing 50 million customer records.
- A machine learning dataset with millions of images.
- A server log generated every second.
- Weather sensors continuously sending measurements.
- A financial application receiving live stock prices.
- A social media platform processing millions of new posts every day.

These datasets are often too large to load into memory all at once.

In some cases, the data is **continuous**, meaning it never truly ends.

If Python insisted on loading all of the data before starting any processing, many real-world applications would either become extremely slow or run out of available memory.

## A Better Way to Think About Data

Instead of asking:

> "Can I have all of the data right now?"

Python asks a different question:

> "Can I have just the next piece of data whenever I need it?"

This small change completely changes how large datasets can be processed.

Rather than waiting until every piece of data has been collected, Python can begin processing immediately.

The program receives one value, processes it, discards it if necessary, and then requests the next value.

This process continues until there is no more data to process.

By working with only one item at a time, Python dramatically reduces memory usage while still accomplishing the same task.

````python
sales = [1250, 980, 1500, 760, 890]

for sale in sales:
    print(f"Processing sale: {sale}")
````

Although this example uses only five values, notice **how** the `for` loop behaves.

It processes one sale...

then the next...

then the next...

At no point does the loop need to process every value simultaneously.

This "one item at a time" approach is one of the most important ideas in Python.

Later in this notebook, you will discover that this is exactly how Python's iteration model works internally.

## Why Is This Important?

Many professional applications work with data that is far too large to fit comfortably into memory.

For example:

**Data Science**

A dataset may contain hundreds of millions of rows. Processing one row at a time is often much more efficient than loading the entire dataset.

**Machine Learning**

Training data is commonly supplied to a model in small batches instead of loading every sample into memory simultaneously.

**ETL Pipelines**

Data is extracted, transformed, and loaded record by record as it flows through the pipeline.

**Web Scraping**

A scraper may continuously receive web pages one after another instead of downloading everything before processing.

**Streaming APIs**

Applications process each incoming response as soon as it arrives.

**Automation**

Large folders containing thousands of files can be processed one file at a time instead of keeping information about every file in memory.

In each of these situations, the fundamental idea is the same:

> **Process only what you need, when you need it.**

# Topic 2 — Eager Evaluation vs Lazy Evaluation

In the previous topic, we learned that Python was designed to process data efficiently, especially when working with large or continuous datasets.

This naturally raises an important question:

> **When should a program create or compute data?**

Should it create **everything immediately**, even if some of the data may never be used?

Or should it create data **only when it is actually needed**?

These two different approaches are known as:

- **Eager Evaluation**
- **Lazy Evaluation**

Understanding these concepts is essential because they form the foundation of Python's iterator and generator model. Almost every iterator and generator in Python relies on **lazy evaluation** to process data efficiently while minimizing memory usage.

Throughout this topic, we will learn not only what these approaches are, but also why Python uses them, how they affect memory and performance, and when each approach is the better choice.

## What Is Evaluation?

Before discussing *eager* and *lazy* evaluation, we first need to understand what the word **evaluation** means.

In programming, **evaluation** simply means:

> **Computing, creating, or producing a value.**

Whenever Python performs an operation to produce a result, it is evaluating an expression.

For example, when Python calculates `5 + 3`, it evaluates the expression and produces the value `8`.

Similarly, when Python creates a list, calculates a mathematical expression, reads data, or executes a function, it is evaluating something.

The difference between **eager** and **lazy** evaluation is **when** that evaluation happens.

````python
# Python evaluates this expression immediately.

result = 5 + 3

print(result)
````

In the previous example, Python did not wait until the value was needed later.

As soon as it encountered the expression, it calculated the result and stored it in memory.

This is an example of **immediate evaluation**, which leads us to our first concept: **Eager Evaluation**.

## What Is Eager Evaluation?

**Eager Evaluation** is an evaluation strategy in which Python computes or creates all required values **immediately**, before they are actually used.

In other words, the work is completed **as early as possible**.

Think of it as preparing everything in advance.

Imagine a teacher who prints all examination papers for every student in the university before knowing how many students will actually attend the exam.

Some papers may never be used, but the work has already been done.

Similarly, with eager evaluation, Python performs the computation immediately—even if some of the computed values may never be needed.

Many common Python objects use eager evaluation, including:

- Lists
- Tuples
- Dictionaries
- Sets

These objects store all of their elements in memory as soon as they are created.

````python
numbers = [1, 2, 3, 4, 5]

print(numbers)
````

- When Python executes the previous statement, **every element of the list is created immediately**.

- Even if your program only uses the first element later, Python has already created and stored all five elements in memory.

- For a list of five values, this is perfectly acceptable.

- However, imagine creating a list containing **100 million values**.

- Python would still attempt to create **every value immediately**, consuming a large amount of memory before your program even begins processing the data.

- This is one of the limitations of eager evaluation when working with very large datasets.

## What Is Lazy Evaluation?

- **Lazy Evaluation** is an evaluation strategy in which values are **not created immediately**.

- Instead, Python waits until a value is actually needed.

- Only then does it compute or produce that value.

- Think of it as preparing things **on demand** rather than preparing everything in advance.

- Imagine ordering food at a restaurant.

- The chef does not cook every item on the menu every morning.

- Instead, the food is prepared only after a customer places an order.

- This saves ingredients, reduces waste, and improves efficiency.

- Python follows the same philosophy when using lazy evaluation.

- Instead of creating every value at once, it produces one value only when requested.

````python
numbers = range(1, 6)

print(numbers)
````

- Notice something interesting.

- Unlike a list, Python did **not** display all the values immediately.

- Instead, it displayed a `range` object.

- This is because `range()` follows a lazy approach.

- It knows **how** to produce the numbers, but it does not create all of them immediately.

- The numbers are produced only when your program requests them, such as inside a `for` loop.

- Later in this notebook, you will learn that this behavior is closely related to iterators.

````python
numbers = range(1, 6)

for number in numbers:
    print(number)
````

The numbers appear one by one as the loop requests them.

Although the final output looks similar to a list, the internal behavior is completely different.

A list already contains every value.

A `range` object generates values only when they are needed.

This "produce only when requested" behavior is the essence of **lazy evaluation**.

## Comparing Eager Evaluation and Lazy Evaluation

The following table summarizes the key differences between the two approaches.

| Feature | Eager Evaluation | Lazy Evaluation |
|---------|------------------|-----------------|
| When are values created? | Immediately | Only when needed |
| Memory usage | Usually higher | Usually lower |
| Initial execution | More work upfront | Less work upfront |
| Suitable for | Small to medium datasets | Large or continuous datasets |
| Common examples | `list`, `tuple`, `dict`, `set` | `range`, iterators, generators |
| Can avoid unnecessary work? | No | Yes |

## Memory Efficiency

One of the biggest advantages of lazy evaluation is **memory efficiency**.

Imagine you need one million numbers.

A list attempts to create and store every number immediately.

A lazy object, such as `range()`, stores only the information required to produce the next value when requested.

As a result, lazy evaluation generally requires significantly less memory.

This is one of the primary reasons Python uses iterators and generators extensively for processing large datasets.

````python
numbers_list = list(range(1, 11))

numbers_range = range(1, 11)

print(numbers_list)
print(numbers_range)
````

Although both objects represent the same sequence of numbers, they are fundamentally different.

- `numbers_list` already contains every value.
- `numbers_range` contains the logic needed to produce those values when required.

The difference becomes much more important when the sequence contains millions of elements instead of just ten.

## Performance Considerations

Choosing between eager and lazy evaluation is not simply a matter of memory usage.

Performance depends on the problem you are trying to solve.

### Eager Evaluation

Advantages:

- Immediate access to all values.
- Faster when the entire dataset will definitely be used multiple times.
- Simpler to modify because the data already exists in memory.

Disadvantages:

- Higher memory consumption.
- May perform unnecessary work by creating values that are never used.

### Lazy Evaluation

Advantages:

- Lower memory usage.
- Begins processing immediately.
- Avoids creating unused values.
- Ideal for streaming and large datasets.

Disadvantages:

- Individual values are computed when requested, which may introduce a small amount of overhead.
- Some lazy objects can only be traversed once, as we will learn later with iterators.

## Real-World Examples

Lazy evaluation is widely used in professional software development.

### Data Science

Processing datasets containing millions of rows without loading the entire dataset into memory.

### Machine Learning

Supplying training data in batches rather than loading every sample simultaneously.

### ETL Pipelines

Reading, transforming, and writing one record at a time.

### Web Scraping

Processing each downloaded web page immediately instead of storing thousands of pages first.

### APIs

Handling streaming responses as they arrive.

### Log Analysis

Reading server log files line by line instead of loading the entire file into memory.

## When Should You Use Each Approach?

There is no universally "better" approach.

The right choice depends on your problem.

Use **eager evaluation** when:

- The dataset is relatively small.
- You need frequent or repeated access to the data.
- Fast random access is important.
- Modifying the data is required.

Use **lazy evaluation** when:

- The dataset is very large.
- Data arrives continuously.
- Memory efficiency is important.
- Values may never all be used.
- You want to process data as it becomes available.

## Key Takeaways

- **Evaluation** means computing or producing values.
- **Eager Evaluation** creates values immediately.
- **Lazy Evaluation** creates values only when they are needed.
- Lazy evaluation generally reduces memory usage because it avoids storing unnecessary data.
- Python's iterators and generators are built around the concept of lazy evaluation.
- Understanding lazy evaluation is essential for working efficiently with large datasets, streaming data, ETL pipelines, machine learning workflows, and many professional Python applications.

# Topic 3 — Iterable vs Iterator

In the previous topic, we learned that Python often processes data **one item at a time** instead of creating everything at once.

This naturally raises another question:

> **If Python processes data one item at a time, where does that data come from?**

Think about a simple `for` loop.

```python
for number in numbers:
    print(number)
```

This loop works with many different types of objects:

- Lists
- Tuples
- Strings
- Dictionaries
- Sets
- Files
- `range()` objects

Although these objects are completely different internally, Python can loop over all of them using exactly the same syntax.

How is that possible?

The answer lies in two closely related concepts:

- **Iterable**
- **Iterator**

These are among the most fundamental concepts in Python because almost every loop, every generator, and many built-in functions depend on them.

In this topic, we will understand what these two terms mean, how they are related, why Python treats them as different concepts, and why understanding this distinction is essential before learning the iterator protocol.

## A Simple Observation

Before learning any definitions, let's observe something interesting.

Look at the following examples.

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

````python
name = "Python"

for letter in name:
    print(letter)
````

````python
scores = {
    "Ali": 85,
    "Sara": 92,
    "John": 78
}

for student in scores:
    print(student)
````

Even though these objects are completely different:

- A list stores values.
- A string stores characters.
- A dictionary stores key-value pairs.

Python's `for` loop works perfectly with all of them.

This tells us something important:

> The `for` loop does **not** care whether it receives a list, a string, or a dictionary.

It only cares whether the object can provide values **one at a time**.

Objects that can do this belong to a special category in Python called **iterables**.

## What Is an Iterable?

An **iterable** is any object that can provide its elements **one at a time**.

In simple words,

> **An iterable is an object that Python can loop over.**

Whenever you write a `for` loop, the object after the keyword `in` must be an iterable.

For example,

```python
for item in something:
```

Here, `something` must be an iterable.

If it is not, Python does not know how to retrieve values from it, and the loop cannot execute.

At this stage, don't worry about *how* an iterable provides those values.

We'll learn that in the next topic when we study the iterator protocol.

For now, simply remember:

> **An iterable knows how to provide its elements one by one.**

````python
cities = [
    "Islamabad",
    "Lahore",
    "Karachi",
    "Peshawar"
]

for city in cities:
    print(city)
````

The list contains four cities.

Notice that the `for` loop does not receive all four values at once.

Instead, it receives:

- Islamabad
- Lahore
- Karachi
- Peshawar

one after another until no values remain.

This is exactly what makes a list an iterable.

## Common Built-in Iterables

Many built-in Python objects are iterable.

Some of the most commonly used ones are:

| Object | Iterable? | Common Usage |
|---------|-----------|--------------|
| `list` | ✅ Yes | Collections of values |
| `tuple` | ✅ Yes | Immutable collections |
| `str` | ✅ Yes | Processing text character by character |
| `dict` | ✅ Yes | Processing keys, values, or items |
| `set` | ✅ Yes | Unique collections |
| `range()` | ✅ Yes | Number sequences |
| File objects | ✅ Yes | Reading files line by line |
| Generators | ✅ Yes | Producing values lazily |

These are called **iterables** because Python can retrieve their elements one at a time.

````python
print("List:")

for value in [1, 2, 3]:
    print(value)

print("\nTuple:")

for value in (10, 20, 30):
    print(value)

print("\nString:")

for character in "Python":
    print(character)

print("\nRange:")

for value in range(1, 4):
    print(value)
````

## What Is an Iterator?

Now that we understand what an iterable is, we can introduce another important concept:

An **iterator** is an object that retrieves values from an iterable **one at a time**.

Notice the difference.

An iterable **contains or represents** a collection of values.

An iterator **travels through** that collection and returns one value whenever Python asks for it.

A useful analogy is a book.

- The **book** is the iterable.
- The **person reading one page at a time** is the iterator.

The book contains all the pages.

The reader decides which page to read next.

Similarly,

- the iterable contains the data,
- while the iterator retrieves each value one at a time.

Later, we'll see that Python creates an iterator automatically whenever a `for` loop begins.

## Relationship Between Iterables and Iterators

An iterable and an iterator work together.

Their relationship can be summarized as follows:

```
Iterable
     │
     │ creates
     ▼
Iterator
     │
     │ returns one value at a time
     ▼
for loop
```

Whenever Python encounters a `for` loop,

it does **not** read values directly from the iterable.

Instead,

1. Python creates an iterator from the iterable.
2. The iterator returns one value.
3. The loop processes that value.
4. The iterator returns the next value.
5. This continues until no values remain.

We'll learn exactly how Python creates that iterator in the next topic using `iter()` and `next()`.

## Iterable vs Iterator

Although these terms are closely related, they describe different objects.

| Iterable | Iterator |
|-----------|----------|
| Represents a collection of data | Retrieves data from an iterable |
| Can create an iterator | Created from an iterable |
| Does not keep track of the current position | Keeps track of the current position |
| Can usually create multiple new iterators | Moves forward one element at a time |
| Examples: `list`, `tuple`, `str`, `dict`, `set`, `range()` | Examples: `list_iterator`, `tuple_iterator`, generator objects |

A common beginner mistake is thinking that an iterable and an iterator are the same thing.

They are **related**, but they serve different purposes.

## Why Python Separates These Concepts?

You might wonder,

> **Why didn't Python simply let the iterable do everything? Why introduce a separate iterator?**

Separating these responsibilities makes Python much more flexible.

The iterable is responsible for **representing the collection**.

The iterator is responsible for **keeping track of the current position** while traversing that collection.

Because these responsibilities are separate:

- Multiple independent iterators can traverse the same iterable.
- The original iterable remains unchanged.
- Python can use the same iteration mechanism for lists, tuples, strings, files, generators, and many other objects.

This separation is one of the reasons Python's iteration model is both elegant and highly reusable.

## Understanding the Big Picture

At this point, you should think about iteration like this:

        ▼
    Iterable

        │
        ▼
    Iterator
        │

        ▼
     for loop

Everything we study from this point onward—including:

- `iter()`
- `next()`
- `__iter__()`
- `__next__()`
- generators
- `yield`
- the `itertools` module

is built upon this simple relationship between **iterables** and **iterators**.

Understanding this relationship is the key to mastering Python's iteration model.

## Key Takeaways

- An **iterable** is an object that Python can loop over.
- An **iterator** retrieves values from an iterable one at a time.
- Every iterator is associated with an iterable.
- Python's `for` loop works by using an iterator behind the scenes.
- Separating iterables and iterators makes Python's iteration model flexible, reusable, and memory-efficient.

# Lesson 4.1 — Introduction to the Iterator Protocol

So far, we have learned three important ideas:

- Python prefers processing data one item at a time.
- Objects that can be looped over are called **iterables**.
- Python uses **iterators** to retrieve values from those iterables.

However, an important question still remains unanswered.

> **How does Python know how to retrieve the next value from completely different objects such as a list, a string, a dictionary, or a file?**

Consider the following examples.

```python
for number in numbers:
    ...

for character in text:
    ...

for line in file:
    ...
```

Although these objects are completely different internally, Python uses exactly the same `for` loop syntax for all of them.

How is this possible?

The answer is that **every iterable follows a common set of rules**.

Python calls these rules the **Iterator Protocol**.

The Iterator Protocol defines **how an object should behave** so that Python can retrieve its elements one at a time in a consistent way.

Because every iterable follows the same protocol, Python only needs to learn one way of iterating over objects.

This is one of the reasons why Python is simple, consistent, and extensible.

## What Is a Protocol?

The word **protocol** is used in many areas of computer science.

In general, a protocol is simply:

> **A set of rules that different systems agree to follow so they can work together correctly.**

For example:

- Two computers communicate using networking protocols.
- Web browsers communicate with web servers using the HTTP protocol.
- Email applications communicate using SMTP.

Python follows the same idea.

Instead of creating different looping mechanisms for lists, tuples, strings, files, and many other objects, Python defines one common protocol for iteration.

If an object follows this protocol, Python automatically knows how to iterate over it.

This design makes Python both flexible and highly extensible because even your own custom objects can participate in `for` loops by following the same rules.

## The Two Rules of the Iterator Protocol

Although the Iterator Protocol is very powerful, it is based on only **two operations**.

1. **Create an iterator from an iterable.**
2. **Retrieve the next value from that iterator.**

Python provides two built-in functions to perform these operations:

| Function | Purpose |
|----------|---------|
| `iter()` | Creates an iterator from an iterable |
| `next()` | Retrieves the next value from an iterator |

These two functions are the foundation of Python's entire iteration mechanism.

Everything else—including `for` loops, generators, file iteration, and many functions from the `itertools` module—is built on top of these two operations.

In the following sections, we will study each function individually.

## The `iter()` Function

The first step in iteration is creating an iterator.

Python does this using the built-in `iter()` function.

### Syntax

```python
iterator = iter(iterable)
```

The `iter()` function accepts any iterable object and returns an **iterator object**.

Notice something important.

The `iter()` function **does not return the values** stored inside the iterable.

Instead, it returns an object whose job is to produce those values one at a time.

Think of it like borrowing a bookmark for a book.

The book (iterable) already contains all the pages.

The bookmark (iterator) remembers where you are currently reading.

````python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)

print(iterator)
print(type(iterator))
````

- Notice that Python did **not** print the values stored in the list.

- Instead, it returned a **`list_iterator` object**.

- This object is responsible for retrieving values from the original list one at a time.

- The list itself has not changed.

- Only a new iterator has been created.

## The `next()` Function

Once an iterator has been created, Python retrieves values using the built-in `next()` function.

### Syntax

```python
next(iterator)
```

Each time `next()` is called:

- the iterator returns the next available value,
- remembers its current position,
- and moves forward.

This process continues until there are no values remaining.

Unlike indexing, the iterator always moves **forward**.

It never automatically moves backward or restarts from the beginning.

````python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
````

- Notice the behavior carefully.

- The iterator remembers its current position after every call to `next()`.

- This means the second call does **not** start from the beginning again.

- Instead, it continues exactly where the previous call stopped.

- This ability to maintain state is one of the defining characteristics of an iterator.

# Lesson 4.2 — Understanding `__iter__()` and `__next__()`

In the previous lesson, we learned about two important built-in functions:

- `iter()`
- `next()`

These functions allow Python to iterate over objects in a consistent way.

However, an important question still remains:

> **How do `iter()` and `next()` know what to do?**

Consider the following examples.

```python
numbers = [10, 20, 30]

iterator = iter(numbers)
print(next(iterator))
```

The `iter()` function works perfectly with a list.

Now replace the list with a tuple.

```python
coordinates = (5, 10, 15)

iterator = iter(coordinates)
```

It still works.

Now replace the tuple with a string.

```python
text = "Python"

iterator = iter(text)
```

Again, it works.

Lists, tuples, strings, dictionaries, sets, files, generators, and many other objects are completely different internally.

Yet Python's `iter()` function works with all of them.

How?

The answer is simple.

Every iterable object agrees to provide two special methods that follow Python's **Iterator Protocol**.

These methods are:

- `__iter__()`
- `__next__()`

The built-in functions `iter()` and `next()` simply use these methods internally.

## Special (Magic) Methods

Methods such as:

- `__iter__()`
- `__next__()`
- `__len__()`
- `__str__()`

are called **special methods** (also commonly known as **magic methods** or **dunder methods**, because they begin and end with double underscores).

You have already used some special methods indirectly.

For example,

```python
len(my_list)
```

internally calls

```python
my_list.__len__()
```

Similarly,

```python
str(obj)
```

internally calls

```python
obj.__str__()
```

The same idea applies here.

```python
iter(obj)
```

internally calls

```python
obj.__iter__()
```

and

```python
next(iterator)
```

internally calls

```python
iterator.__next__()
```

The built-in functions simply provide a clean and readable interface for these special methods.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(iterator)
````

Although we wrote:

```python
iterator = iter(numbers)
```

Python actually performs something conceptually similar to:

```python
iterator = numbers.__iter__()
```

Both statements produce the same iterator object.

However, as Python programmers, we should almost always use the built-in `iter()` function because it is the standard and recommended approach.

## What Is `__iter__()`?

The `__iter__()` method is responsible for **creating and returning an iterator**.

Think back to our earlier analogy.

Imagine you own a book.

The book contains all the pages.

Before you can start reading, you need someone who knows **where to begin reading**.

The book itself doesn't read its pages.

Instead, it provides a reader.

Similarly,

- the iterable stores the data,
- while `__iter__()` provides an iterator that knows how to traverse that data.

Therefore,

> **The responsibility of `__iter__()` is to create (or return) an iterator.**

It does **not** return the actual values.

It returns the object that will later provide those values one by one.

````python
numbers = [100, 200, 300]

iterator = numbers.__iter__()

print(iterator)
print(type(iterator))
````

Notice that `__iter__()` did **not** return:

```python
100
200
300
```

Instead, it returned a **list iterator object**.

This iterator is now responsible for producing those values one at a time.

## What Is `__next__()`?

Once an iterator has been created, someone must be responsible for returning each value.

That responsibility belongs to `__next__()`.

Every time Python needs another value, it calls:

```python
iterator.__next__()
```

The iterator then:

1. Returns the current value.
2. Moves to the next position.
3. Waits until another value is requested.

Notice something very important.

The iterator always remembers **where it stopped**.

This ability to remember its current position is called **state**.

Maintaining state is one of the defining characteristics of an iterator.

````python
numbers = [100, 200, 300]

iterator = numbers.__iter__()

print(iterator.__next__())
print(iterator.__next__())
print(iterator.__next__())
````

````python
numbers = [100, 200, 300]

iterator = iter(numbers)

print(next(iterator))
print(iterator.__next__())
print(iterator.__next__())
````

This code produces exactly the same output as:

```python
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

The only difference is that the first version calls the special method directly, while the second uses Python's built-in function.

# Lesson 4.3 — `StopIteration` and How `for` Loops Work Internally

So far, we have learned that:

1. An iterable can produce an iterator using `__iter__()`.
2. An iterator returns one value at a time using `__next__()`.

However, one important question still remains.

> **How does Python know when there are no more values to retrieve?**

Consider the following code.

```python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

After retrieving all three values, what should happen if Python calls `next()` one more time?

Should it:

- return `None`?
- return `0`?
- start again from the beginning?
- keep returning the last value?

None of these choices are correct because they would create ambiguity.

Instead, Python follows a much cleaner approach.

When an iterator has no more values to produce, it raises a special exception called **`StopIteration`**.

This exception is the iterator's way of saying:

> **"There are no more values available."**

Understanding this exception is essential because it is how every iterator tells Python that iteration has finished.

## What Is `StopIteration`?

`StopIteration` is a built-in Python exception.

Its purpose is very specific:

> **It signals that an iterator has been exhausted, meaning there are no more values left to return.**

Unlike exceptions such as `ValueError` or `TypeError`, which usually indicate an error in your program, `StopIteration` is **not an error** in the normal sense.

It is simply a signal used internally by Python's iteration mechanism.

Every iterator eventually reaches a point where it has returned all available values.

At that moment, it raises `StopIteration` to inform Python that iteration is complete.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))

# One more request
print(next(iterator))
````

The first three calls to `next()` succeed because the iterator still has values to return.

The fourth call fails because the iterator has already reached the end of the list.

Instead of returning an incorrect value, Python raises the `StopIteration` exception.

This makes the end of iteration explicit and unambiguous.

## Handling `StopIteration`

Since `StopIteration` is an exception, it can be handled using a `try`-`except` block.

Although this is rarely needed when using `for` loops, it is useful when manually working with iterators.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

while True:
  try:
    print(next(iterator))

  except StopIteration:
    print("No more values.")
    break
````

In this example:

1. `next()` retrieves one value at a time.
2. When all values have been returned, the iterator raises `StopIteration`.
3. The `except` block catches the exception.
4. The loop terminates gracefully.

This pattern demonstrates exactly how iteration can be controlled manually.

## How Does a `for` Loop Work Internally?

One of the most beautiful aspects of Python is that the familiar `for` loop is built entirely on top of the Iterator Protocol.

When you write:

```python
for item in iterable:
    print(item)
```

Python does **not** directly access the iterable.

Instead, it performs several internal steps automatically.

### Step 1 — Create an Iterator

Python first creates an iterator from the iterable.

Conceptually:

```python
iterator = iter(iterable)
```

### Step 2 — Request the First Value

Python retrieves the first element.

```python
item = next(iterator)
```

### Step 3 — Execute the Loop Body

Python executes the statements inside the loop using the retrieved value.

### Step 4 — Request the Next Value

Python asks the iterator for another value.

```python
item = next(iterator)
```

If another value exists, the loop continues.

Otherwise, the iterator raises `StopIteration`.

### Step 5 — Finish the Loop

When `StopIteration` is raised, Python catches it automatically and exits the loop.

This entire process happens silently.

That is why you almost never see `StopIteration` while using a normal `for` loop.

## Internal Workflow of a `for` Loop

```
                Iterable
                    │
                iter()
                    │
                    ▼
               Iterator
                    │
             next() → Value
                    │
          Execute Loop Body
                    │
             next() → Value
                    │
          Execute Loop Body
                    │
             next() → Value
                    │
          Execute Loop Body
                    │
             next()
                    │
          StopIteration Raised
                    │
                    ▼
            Loop Terminates
```

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

while True:
    try:
        number = next(iterator)
        print(number)
    except StopIteration:
        break
````

Both programs produce exactly the same output.

The first version is the clean Python syntax that we write every day.

The second version reveals what Python is doing behind the scenes.

This demonstrates that a `for` loop is essentially a convenient wrapper around:

- `iter()`
- `next()`
- `StopIteration`

# Lesson 4.4 — Creating Custom Iterators

So far, we have used iterators created automatically by Python.

For example,

- A `list` can create a list iterator.
- A `tuple` can create a tuple iterator.
- A `str` can create a string iterator.
- A file object can create a file iterator.

These built-in iterators are sufficient for many tasks.

However, professional software often needs to iterate over data that does not naturally exist as a list or tuple.

For example,

- Reading a huge CSV file one record at a time.
- Streaming data from an API.
- Processing sensor readings from an IoT device.
- Loading Machine Learning training batches.
- Reading millions of log records.
- Generating invoice numbers.
- Producing Fibonacci numbers.

In situations like these, creating a complete list first may be impossible or highly inefficient.

Instead, we create **custom iterators** that generate values one at a time whenever Python requests them.

By implementing the Iterator Protocol ourselves, we can make our own objects work naturally with Python's `for` loop and all functions that expect an iterable.

## How Do We Create a Custom Iterator?

To create a custom iterator, a class must follow the **Iterator Protocol**.

That means implementing two special methods:

| Method | Responsibility |
|---------|----------------|
| `__iter__()` | Return the iterator object |
| `__next__()` | Return the next value |

Together, these methods allow Python to retrieve values one at a time until `StopIteration` is raised.

The workflow is exactly the same as the one used by built-in iterators.

## Example 1 — Counting Numbers

Let's build our first custom iterator.

Our goal is simple.

We want an object that returns:

```

1
2
3
4
5

```

one value at a time.

Instead of storing these numbers in a list, the iterator will generate them whenever Python asks for the next value.

This example focuses on understanding the Iterator Protocol rather than solving a real-world problem.

````python
class Counter:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.end:
            raise StopIteration

        number = self.current
        self.current += 1
        return number


counter = Counter(1, 5)

for number in counter:
    print(number)
````

## Understanding the Code

Let's understand what each method does.

### `__init__()`

Stores the starting and ending values.

```python
self.current = start
self.end = end
```

---

### `__iter__()`

```python
return self
```

Since the object itself already knows how to produce the next value, it simply returns itself as the iterator.

---

### `__next__()`

This method performs four steps every time Python requests another value.

1. Check whether iteration has finished.
2. Raise `StopIteration` if no values remain.
3. Return the current value.
4. Move to the next value.

Every call to `next()` repeats this process.

This is exactly how custom iterators participate in Python's iteration system.

      ▼

      │
      ▼
Iterator

next() → 1
next() → 2
next() → 3
next() → 4
next() → 5
next() → StopIteration
```


## Example 2 — Employee ID Generator

Suppose a company assigns employee IDs sequentially.

Instead of storing thousands of IDs in a list, we can generate them only when needed.

Although this is a simplified example, it demonstrates a common business pattern where values are produced on demand.

````python
class EmployeeIDGenerator:
    def __init__(self, start_id, end_id):
        self.current_id = start_id
        self.end_id = end_id

    def __iter__(self):
        return self

    def __next__(self):
        if self.current_id > self.end_id:
            raise StopIteration

        employee_id = f"EMP-{self.current_id}"
        self.current_id += 1
        return employee_id


employee_ids = EmployeeIDGenerator(1001, 1005)

for employee_id in employee_ids:
    print(employee_id)
````

## Example 3 — Processing Sales Records

Imagine that sales records are received one at a time from another system.

Rather than storing all records first, our iterator supplies one record whenever Python requests it.

This mirrors the way many ETL pipelines process incoming data.

````python
class SalesIterator:
    def __init__(self, sales):
        self.sales = sales
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.sales):
            raise StopIteration

        record = self.sales[self.index]
        self.index += 1
        return record


sales_data = [
    {"customer": "Ali", "amount": 1200},
    {"customer": "Sara", "amount": 950},
    {"customer": "John", "amount": 1800},
]

sales_iterator = SalesIterator(sales_data)

for record in sales_iterator:
    print(record)
````

## Example 4 — Mini-Batch Iterator

Training Machine Learning models on millions of samples at once is usually impossible because of memory limitations.

Instead, datasets are divided into **mini-batches**.

Each batch is processed independently.

The following example demonstrates this idea using a custom iterator.

````python
class BatchIterator:
    def __init__(self, data, batch_size):
        self.data = data
        self.batch_size = batch_size
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration

        batch = self.data[self.index:self.index + self.batch_size]
        self.index += self.batch_size
        return batch


dataset = list(range(1, 11))

batches = BatchIterator(dataset, batch_size=3)

for batch in batches:
    print(batch)
````

This example demonstrates one of the most common ideas in Machine Learning.

Instead of loading and processing the entire dataset at once, the model receives a small batch of samples.

Although real frameworks such as **TensorFlow** and **PyTorch** use far more sophisticated data loaders, the underlying idea is the same: retrieve the next batch only when it is needed.

## Best Practices

When creating custom iterators:

- Implement both `__iter__()` and `__next__()`.
- Always raise `StopIteration` when iteration finishes.
- Keep `__next__()` focused on producing the next value.
- Avoid placing unrelated business logic inside `__next__()`.
- Use meaningful attribute names such as `current_index` or `current_record`.
- Prefer generators (which we'll study next) for simple sequential iteration because they are shorter, cleaner, and easier to maintain.

In the previous topic, we successfully built our own custom iterators.

That is a lot of code simply to produce values one at a time.

Now imagine writing dozens of such iterators in a Data Science project or an ETL pipeline.

The code would become repetitive and difficult to maintain.

Python's designers recognized this problem and introduced **generators**.

Generators allow us to create iterators without manually implementing the Iterator Protocol.

Instead of writing the machinery for iteration ourselves, we simply describe **what values should be produced**, and Python automatically creates the iterator behind the scenes.

## A Reminder: Our Custom Iterator

In the previous topic, we created a counter iterator.

Although it was not difficult, it required an entire class.

The iterator's primary responsibility was simply to produce the numbers:

```

1
2
3
4
5

```

one at a time.

Let's briefly revisit that implementation.

````python
class Counter:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.end:
            raise StopIteration

        number = self.current
        self.current += 1
        return number


counter = Counter(1, 5)

for number in counter:
    print(number)
````

Even though the iterator's job was simple, we had to write an entire class.

Wouldn't it be nice if we could write something much simpler, such as:

```python

yield 3

yield 5
```

That's exactly what generators allow us to do.

Instead of implementing the Iterator Protocol manually, Python does it automatically for us.

## What Is a Generator Function?

A **generator function** is a special type of function that produces values **one at a time** instead of returning them all at once.

Unlike a regular function, a generator function uses the `yield` keyword instead of the `return` keyword to produce values.

When Python encounters a function containing `yield`, it no longer treats it as a normal function.

Instead, it creates a **generator function**.

Calling that function does **not** immediately execute its body.

Instead, Python returns a **generator object**, which can produce values one at a time whenever requested.

This behavior is what makes generators both **lazy** and **memory-efficient**.

````python
def count_to_five():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5


numbers = count_to_five()

print(numbers)
print(type(numbers))
````

Notice something surprising.

Calling the function did **not** print any numbers.

Instead, Python returned a **generator object**.

This tells us something very important.

A generator function does **not** execute immediately when it is called.

Instead, it creates a generator object that is capable of producing values later.

In other words,

calling the function **creates the generator**,

while retrieving values **executes the generator**.

## Retrieving Values from a Generator

A generator is an iterator.

Therefore, it follows exactly the same Iterator Protocol that we studied earlier.

That means we can retrieve values using either:

- the `next()` function, or
- a `for` loop.

````python
def count_to_five():
  yield 1
  yield 2
  yield 3
  yield 4
  yield 5

generator = count_to_five()
print(next(generator))
print(next(generator))
print(next(generator))
````

Each call to `next()` retrieves exactly one value from the generator.

Notice that the generator remembers where it stopped.

The next call continues from the following `yield` statement rather than starting from the beginning.

This behavior should already feel familiar because generators are simply another type of iterator.

````python
def count_to_five():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5


for number in count_to_five():
    print(number)
````

Using a `for` loop is the most common way to consume values from a generator.

Just like with lists, tuples, and custom iterators, the `for` loop automatically:

1. Creates an iterator.
2. Retrieves each value.
3. Stops when `StopIteration` is raised.

The difference is that Python generated the iterator for us automatically.

# Lesson 5.2 — The `yield` Keyword

In the previous lesson, we learned that a function containing the `yield` keyword becomes a **generator function**.

However, we did not discuss **what `yield` actually does**.

This lesson answers that question.

Understanding `yield` is essential because it is the feature that makes generators possible.

By the end of this lesson, you will understand:

- What `yield` is.
- How it works.
- Why it pauses execution instead of ending the function.
- How it differs from `return`.
- Why it enables lazy evaluation.

## What Is `yield`?

The `yield` keyword is used inside a generator function to **produce one value at a time**.

Unlike `return`, which immediately ends the function, `yield` does two things:

1. Returns a value to the caller.
2. **Pauses** the function at that exact point.

The function does **not** start over the next time it is called.

Instead, it resumes execution from the line immediately after the previous `yield`.

This ability to pause and later continue execution is what makes generators unique.

````python
def numbers():
    yield 10
    yield 20
    yield 30


generator = numbers()

print(next(generator))
print(next(generator))
print(next(generator))
````

Although three values were produced, notice that they were **not** returned all at once.

Each call to `next()` caused the generator to continue execution until it reached the next `yield` statement.

After producing the value, the generator paused again.

This process repeated until no `yield` statements remained.

## How Does `yield` Work?

Let's follow the execution of a generator step by step.

Consider the following generator.

````python
def demo():
    print("Start")

    yield 1

    print("Middle")

    yield 2

    print("End")

    yield 3


generator = demo()
````

Notice that we created the generator.

```python
generator = demo()
```

But nothing was printed.

Why?

Because **calling a generator function does not execute it**.

It only creates a generator object.

Execution begins only when Python requests the first value.

````python
print(next(generator))
````

Let's examine what happened.

1. The generator started executing.
2. It printed `"Start"`.
3. It reached:

```python
yield 1
```

4. It returned `1`.
5. It **paused immediately**.

Everything after `yield 1` remained unexecuted.

````python
print(next(generator))
````

Notice something remarkable.

The generator did **not** print `"Start"` again.

Instead, it continued exactly where it had paused.

Execution resumed after:

```python
yield 1
```

Then it printed `"Middle"`.

Next, it reached:

```python
yield 2
```

returned the value,

and paused once more.

````python
print(next(generator))
````

Again, the generator resumed from the previous pause.

It printed `"End"`,

reached:

```python
yield 3
```

returned the value,

and paused.

````python
print(next(generator))
````

There were no more `yield` statements to execute.

The generator therefore raised `StopIteration`, exactly like every other iterator.

## `yield` vs `return`

Although both keywords can produce values, they behave very differently.

| Feature | `yield` | `return` |
|---------|----------|----------|
| Returns a value | ✅ Yes | ✅ Yes |
| Pauses the function | ✅ Yes | ❌ No |
| Ends the function | ❌ No | ✅ Yes |
| Can continue execution later | ✅ Yes | ❌ No |
| Used in | Generator functions | Regular functions |
| Produces multiple values | ✅ Yes | ❌ No (single return per call) |
| Supports lazy evaluation | ✅ Yes | ❌ No |

This is the single most important table in the Generators topic.

Almost everything about generators follows from these differences.

## Example — Using `return`

A regular function stops executing as soon as it reaches a `return` statement.

````python
def calculate():
    print("Start")

    return 100

    print("End")


print(calculate())
````

Notice that `"End"` was never printed.

Once Python encountered `return`, the function terminated immediately.

Everything after `return` became unreachable.

## Example — Using `yield`

Now compare this with a generator.

````python
def calculated():
  print("Start")

  yield 100

  print("End")

generator = calculated()

print(next(generator))
````

````python
print(next(generator))
````

The important difference is that the function did **not** terminate after producing `100`.

Instead, it paused.

When another value was requested, execution continued from exactly where it had paused.

## Why Does `yield` Support Lazy Evaluation?

Recall what we learned in Topic 2.

Lazy evaluation means:

> **Produce values only when they are needed.**

That is exactly what `yield` does.

Instead of computing every value immediately,

the generator waits until `next()` requests another value.

Only then does it continue execution.

This means:

- unnecessary work is avoided,
- memory usage stays low,
- large datasets can be processed efficiently.

This is why generators are ideal for streaming data and handling large collections.

## Data Analysis Example

Imagine sales records arriving one by one from a reporting system.

Instead of creating one huge list, a generator can produce each record only when the analysis code requests it.

````python
def sales_records():
    yield {"customer": "Ali", "amount": 1200}
    yield {"customer": "Sara", "amount": 980}
    yield {"customer": "John", "amount": 1450}


for record in sales_records():
    print(record)
````

Each record is produced only when the `for` loop requests it.

If the loop stops early, the remaining records are never generated.

This is an important advantage when processing very large datasets.

# Lesson 5.3 — Generator State, Pausing, and Resuming Execution

In the previous lesson, we learned that the `yield` keyword pauses a generator instead of terminating it.

However, that raises an important question.

Suppose a generator pauses after producing a value.

When we request the next value later, how does Python know:

- which line to execute next?
- what the local variables were?
- what the loop counter was?
- where the function had paused?

The answer is that Python automatically saves the **state** of the generator.

When execution resumes, Python restores that state exactly as it was before the pause.

This ability to preserve execution state is what makes generators fundamentally different from regular functions.

## What Is Generator State?

The **state** of a generator is all the information Python needs to continue execution from where the generator last paused.

This includes:

- The current execution position (the next line to execute).
- The values of local variables.
- The values of function parameters.
- The progress of loops.
- Any intermediate calculations that have already been performed.

You can think of generator state as a **snapshot** of the function at the moment it reaches a `yield` statement.

Instead of discarding this snapshot, Python stores it in memory.

When `next()` is called again, Python restores the snapshot and continues execution as if the function had never stopped.

## Regular Functions Forget Everything

A regular function begins execution from the first line every time it is called.

Once it finishes, all of its local variables are destroyed.

For example:

````python
def counter():
    number = 1
    print(number)


counter()
counter()
counter()
````

Each call starts from the beginning.

The variable `number` is created, used, and then destroyed.

The function remembers nothing from previous calls.

## Generators Preserve Their State

Now compare that behavior with a generator.

````python
def counter():
  number = 1

  yield number

  number += 1

  yield number

  number += 1

  yield number

generator = counter()

print(next(generator))
print(next(generator))
print(next(generator))
````

Notice something remarkable.

The variable `number` was created only once.

After the first `yield`, the generator paused.

The value of `number` remained stored inside the generator.

When execution resumed, Python restored the value of `number`, increased it by one, and continued.

The variable was **not recreated**.

It simply continued from its previous state.

        ▼

        │
        ▼

------------

Next line:
number += 1

        │
 next()
        ▼

number += 1

yield number
        │
        ▼
Pause

State Saved
------------
number = 2
Next line:
number += 1

        │
 next()
        ▼

number += 1

yield number

```

## Using Multiple `yield` Statements

A generator may contain any number of `yield` statements.

Each `yield` represents another point where execution can pause.

Every time `next()` is called:

1. Execution resumes.
2. The generator runs until the next `yield`.
3. A value is returned.
4. Execution pauses again.

````python
def weekdays():
    yield "Monday"
    yield "Tuesday"
    yield "Wednesday"
    yield "Thursday"
    yield "Friday"


for day in weekdays():
    print(day)
````

Although the generator contains five `yield` statements, only one value exists at a time.

After producing a value, the generator immediately pauses.

This behavior allows generators to produce extremely long sequences without storing them all in memory.

## State Is Preserved Inside Loops

One of the most powerful features of generators is that they remember the progress of loops.

Let's see an example.

````python
def squares(limit):
  for number in range(1, limit + 1):
    yield number ** 2

generator = squares(5)

print(next(generator))
print(next(generator))
print(next(generator))
````

Notice that the generator did not restart the loop each time.

Instead, it remembered:

- the current value of `number`,
- the current loop iteration,
- the next line to execute.

When execution resumed, the loop continued naturally from where it had paused.

This is one reason generators work so well for processing large datasets.

## Data Analysis Example — Processing Records Sequentially

Imagine processing customer transactions.

Instead of loading all transactions into memory at once, a generator produces one record at a time.

````python
def sales_records():
    records = [
        {"customer": "Ali", "amount": 1200},
        {"customer": "Sara", "amount": 980},
        {"customer": "John", "amount": 1450},
    ]

    for record in records:
        yield record


generator = sales_records()

print(next(generator))
print(next(generator))
````

After producing the first record, the generator remembers:

- which record was processed,
- where the loop stopped,
- which record should be produced next.

This makes generators ideal for ETL pipelines and large-scale data processing.

## Machine Learning Example — Mini-Batch Processing

Machine Learning models rarely load an entire dataset into memory.

Instead, they process one batch at a time.

A generator naturally remembers which batch has already been processed.

````python
def batches():
    dataset = list(range(1, 11))

    batch_size = 3

    for start in range(0, len(dataset), batch_size):
        yield dataset[start:start + batch_size]


for batch in batches():
    print(batch)
````

# Understanding the `batches()` Generator

- The purpose of this generator is to divide a dataset into **smaller groups (called batches)** and produce one batch at a time.

- Instead of returning the entire dataset at once, it lazily generates one batch whenever it is requested.

- This is a common technique in Machine Learning because models are usually trained on **mini-batches** rather than the entire dataset.

After yielding each batch, the generator remembers the current position in the dataset.

When the training loop requests another batch, execution resumes from the next position instead of starting over.

This is exactly the principle used by many Machine Learning data loaders.

## Why Generator State Matters

Generator state is one of the reasons generators are so powerful.

Without generator state, we would need to manually store:

- loop counters,
- current positions,
- intermediate calculations,
- remaining work.

The generator automatically preserves all of this information for us.

As a result, code becomes:

- shorter,
- cleaner,
- easier to maintain,
- more memory-efficient.

## Best Practices

- Let generators maintain their own state instead of using global variables.
- Use local variables freely inside generators—Python preserves them automatically.
- Use loops with `yield` when producing sequences of values.
- Prefer generators over manually tracking positions with custom iterator classes unless you need additional control.

# Lesson 5.4 — Sending Data into Generators

So far, every generator we have written has behaved in the same way.

It produced values, and the outside world consumed them.

For example,

```python
def numbers():
    yield 10
    yield 20
    yield 30
```

The communication was always **one-way**.

```
Generator  ─────────►  Caller
```

The generator sent values outward.

However, Python generators are capable of something much more interesting.

Instead of only **sending data out**, they can also **receive data back** while they are paused.

This is possible through the `.send()` method.

With `.send()`, communication becomes **two-way**.

```
          value
Generator ─────────► Caller
     ▲                 │
     │                 ▼
     ◄────── send() ───┘
```

Although this feature is not required for most Data Science or Machine Learning projects, understanding it gives you a much deeper understanding of how generators work internally.


## Roadmap

In this lesson, we will cover:

1. Why `.send()` exists.
2. How `.send()` works.
3. Why the first call must be different.
4. Two-way communication between the caller and the generator.
5. Practical examples.
6. Real-world applications.

## Why Was `.send()` Introduced?

Until now, our generators could only produce values.

For example,

```
Generator

↓

yield 10

↓

yield 20

↓

yield 30
```

The caller could receive these values, but it had **no way to provide new information back to the generator**.

Suppose a generator asks:

> "What should I process next?"

or

> "What discount should I apply?"

or

> "Should I continue?"

Without `.send()`, the generator cannot receive these answers while it is running.

The `.send()` method solves this problem by allowing the caller to send a value directly into the paused generator.

## Our First `.send()` Example

Let's begin with the smallest possible example.

Don't worry if it looks unusual at first—we'll explain every line.

````python
def simple_generator():
    value = yield
    print(f"Received: {value}")


generator = simple_generator()
````

Notice that we have written:

```python
value = yield
```

This is different from:

```python
yield 10
```

Here, the generator is **waiting to receive a value**.

That value will later be assigned to the variable `value`.

## Why Can't We Call `.send()` Immediately?

Let's try this.

````python
generator.send("Hello")
````

This surprises many beginners.

Why does Python raise an error?

The reason is simple.

When the generator is first created, execution has **not yet reached the first `yield` statement**.

There is nowhere for `"Hello"` to go.

Python must first move the generator to its first pause point.

Only then can it receive data.

## Priming a Generator

Before sending real data, we must **start** (or **prime**) the generator.

There are two common ways to do this.

Using `next()`:

```python
next(generator)
```

or

```python
generator.send(None)
```

Both methods advance execution until the first `yield`.

After that, the generator pauses and is ready to receive values.

````python
def simple_generator():
  value = yield
  print(f"Received {value}")

generator = simple_generator()

next(generator)

generator.send("Hello")
````

Let's examine what happened.

### Step 1

```python
generator = simple_generator()
```

A generator object is created.

Nothing inside the function executes yet.

---

### Step 2

```python
next(generator)
```

Execution starts.

The generator reaches

```python
yield
```

and pauses.

It is now waiting to receive a value.

---

### Step 3

```python
generator.send("Hello")
```

The value `"Hello"` is sent into the paused generator.

The statement

```python
value = yield
```

becomes

```python
value = "Hello"
```

Execution continues.

Finally,

```python
print(value)
```

prints

```
Received: Hello
```

The function reaches its end, so Python raises `StopIteration`.

## Execution Flow

```
Generator Created

↓

(next())

↓

yield
▲
│
Paused

↓

.send("Hello")

↓

value = "Hello"

↓

print(value)

↓

StopIteration
```

## A Generator Can Both Send and Receive Values

The real strength of `.send()` becomes clear when the generator both **yields** values and **receives** values.

Communication now works in both directions.

````python
def conversation():
    name = yield "What is your name?"
    yield f"Hello, {name}!"


generator = conversation()

print(next(generator))

print(generator.send("Abdullah"))
````

Notice the sequence of events.

The generator first sends a message:

```
"What is your name?"
```

The caller responds:

```
.send("Abdullah")
```

The generator receives that value, stores it in `name`, and then produces another message.

This is genuine **two-way communication**.

## Practical Example — Applying a Discount

Suppose a billing system asks another part of the program for the discount percentage.

Instead of hardcoding the value, the generator waits until the caller provides it.

````python
def calculate_price(price):
  discount = yield "Enter discount percentage"

  final_price = price - (price * discount / 100)

  yield final_price

generator = calculate_price(5000)

print(next(generator))

print(generator.send(10))
````

The generator first requests the discount.

The caller supplies the value using `.send()`.

The generator resumes execution, performs the calculation, and yields the final price.

This demonstrates how generators can receive input during execution instead of requiring all information beforehand.

## Where Is `.send()` Used?

The `.send()` method is much less common than ordinary generators.

However, it is useful in specialized situations such as:

- Event-driven systems.
- Interactive pipelines.
- Stateful data processing.
- Coroutine-based programming (before Python introduced `async` and `await`).
- Some advanced networking and asynchronous frameworks.

For most Data Science, Machine Learning, and ETL tasks, you will use ordinary generators far more frequently than `.send()`.

- `.send()` is an advanced feature that is used less frequently than ordinary generators but is valuable for understanding Python's generator model and certain advanced programming techniques.

## What Is Generator Chaining?

Generator chaining means connecting multiple generators together so that the output of one generator becomes the input of another.

Instead of one generator doing everything, each generator is responsible for only one task.

Think of it like an assembly line in a factory.

```

Raw Material

↓

Machine 1

↓

Machine 2

↓

Machine 3

Each machine performs only one operation.

Generators work the same way.

Each generator receives data, processes it, and yields the result to the next generator.

Suppose we want to process employee names.

We have two separate tasks:

1. Convert names to uppercase.
2. Add a greeting.

Instead of writing one large function, we'll create two generators.

````python
def uppercase(names):
  for name in names:
    yield name.upper()

def greeting(names):
  for name in names:
    yield f"Welcome {name}"

employees = ["Ali", "Sara", "John"]

result = greeting(uppercase(employees))

for message in result:
  print(message)
````

Let's understand what happened.

John

```text

Welcome JOHN
```

```
Employees

↓

↓

ALI

↓

greeting()

↓

Welcome ALI

↓

Print
```

The same process repeats for every employee.

## Why Data Pipelines?

Data rarely arrives in a format that is immediately useful.

For example, a CSV file might contain:

- extra spaces,
- missing values,
- incorrect capitalization,
- unwanted records.

Instead of performing every operation in one large loop, professionals split the work into small processing stages.

Each stage performs one specific task.

This makes the code easier to understand, test, and maintain.

````python
def remove_spaces(records):
  for record in records:
    yield record.strip()

def remove_empty(records):
  for record in records:
    if record:
      yield record

def uppercase(records):
  for record in records:
    yield record.upper()

data = [
    " ali ",
    "",
    " sara ",
    " john "
]

pipeline = uppercase(
    remove_empty(
        remove_spaces(data)
    )
)

for item in pipeline:
  print(item)
````

Each generator performs only one responsibility.

```
Original Data

↓

remove_spaces()

↓

remove_empty()

↓

uppercase()

↓

Output
```

This design follows an important software engineering principle:

> **One function should have one responsibility.**

This principle makes software easier to extend and maintain.

## Streaming Data

One of the biggest strengths of generators is that they naturally support **streaming**.

Streaming means data is processed as it arrives instead of waiting for the entire dataset.

This is extremely important when working with:

- large CSV files,
- server logs,
- sensor data,
- API responses,
- live financial data,
- IoT devices."

````python
def log_stream():
    logs = [
        "INFO User logged in",
        "ERROR Database connection failed",
        "INFO Report generated",
        "ERROR Invalid password",
    ]

    for log in logs:
        yield log


def error_logs(log_generator):
    for log in log_generator:
        if "ERROR" in log:
            yield log


for error in error_logs(log_stream()):
    print(error)
````

Notice that the `error_logs()` generator never waits for all logs.

Each log entry is processed immediately after it is produced.

This allows applications to monitor systems continuously instead of waiting until all logs have been collected.

## Machine Learning Pipeline

Machine Learning data usually goes through several stages before training.

For example:

```

↓

Normalize

↓

Create Batches

↓

Model Training

```

Each stage performs one transformation.

Generators are ideal for this type of workflow because each sample is processed only when the model requests it.

````python
def images():
    for image in ["img1", "img2", "img3"]:
        yield image


def normalize(images):
    for image in images:
        yield f"Normalized({image})"


def batches(images):
    batch = []

    for image in images:
        batch.append(image)

        if len(batch) == 2:
            yield batch
            batch = []

    if batch:
        yield batch


pipeline = batches(normalize(images()))

for batch in pipeline:
    print(batch)
````

Although our dataset contains only three images, the same approach scales to millions of training samples.

Each stage performs one specific task, and data flows through the pipeline one item at a time.

# Lesson 6.1 — Generator Expressions

In the previous topic, we learned how to create generators using **generator functions** and the `yield` keyword.

Generator functions are powerful because they allow us to produce values one at a time while keeping memory usage low.

However, consider a very simple task.

Suppose we want to generate the squares of the numbers from 1 to 5.

Using a generator function, we would write:

```python
def squares():
    for number in range(1, 6):
        yield number ** 2
```

This is clear and readable.

But for such a simple operation, writing an entire function can feel unnecessary.

Python provides a shorter and more concise way to create simple generators.

This feature is called a **generator expression**.

Generator expressions allow us to create generators in a single line of code while still preserving all the benefits of generators, including lazy evaluation and memory efficiency.

## Why Do Generator Expressions Exist?

Imagine you need a generator only once.

For example:

- Squaring numbers
- Converting text to uppercase
- Filtering valid records
- Processing a sequence before passing it to another function

Writing a complete generator function every time would introduce unnecessary boilerplate code.

Generator expressions were introduced to solve this problem.

They provide a concise, readable, and Pythonic way to create simple generators without writing a separate function.

They are to **generator functions** what **list comprehensions** are to traditional `for` loops:

- They reduce the amount of code.
- They improve readability.
- They preserve lazy evaluation.

Before introducing generator expressions, let's revisit a generator function that produces the squares of numbers.

````python
def squares():
    for number in range(1, 6):
        yield number ** 2


for value in squares():
    print(value)
````

This generator works perfectly.

However, notice that we had to:

- define a function,
- write a `for` loop,
- use the `yield` keyword.

For a simple transformation, this is more code than necessary.

## What Is a Generator Expression?

A **generator expression** is a compact way to create a generator without writing a generator function.

Instead of using the `yield` keyword inside a function, we write an expression inside **parentheses**.

When Python evaluates the expression, it creates a **generator object**.

Like generator functions, generator expressions:

- produce values one at a time,
- are lazily evaluated,
- implement the Iterator Protocol automatically,
- are memory-efficient.

The main difference is that generator expressions are intended for **simple, single-expression generators**, while generator functions are better suited for more complex logic.

## Syntax of a Generator Expression

The general syntax is:

```python
(expression for item in iterable)
```

If a condition is needed:

```python
(expression for item in iterable if condition)
```

Notice the **parentheses**.

They distinguish a generator expression from a list comprehension, which uses square brackets.

Let's rewrite our previous example using a generator expression.

````python
squares = (x ** 2 for x in range(1, 6))

print(squares)
print(type(squares))
````

Notice something familiar.

Just like calling a generator function, creating a generator expression **does not immediately produce values**.

Instead, Python creates a **generator object**.

The values will only be produced when they are requested.

Since a generator expression creates a generator object, we use it exactly like any other generator.

````python
for number in squares:
  print(number)
````

````python
squares = (number ** 2 for number in range(1, 6))

print(next(squares))
print(next(squares))
print(next(squares))
````

The behavior should look familiar.

Each call to `next()` requests the next value from the generator expression.

After producing the value, the generator pauses until another value is requested.

This confirms that generator expressions follow the same Iterator Protocol as generator functions.

## Relationship with Generator Functions

Generator expressions and generator functions solve the same problem:

> **Producing values lazily, one at a time.**

The difference lies in **how they are written**.

| Generator Function | Generator Expression |
|--------------------|----------------------|
| Uses the `def` keyword | Does not use `def` |
| Uses the `yield` keyword | Uses an expression inside parentheses |
| Suitable for complex logic | Suitable for simple transformations |
| Can contain multiple statements | Limited to a single expression |
| Easier to extend with complex processing | More concise for simple tasks |

Although their syntax differs, both create **generator objects**, both are **lazy**, and both implement the **Iterator Protocol**.

## Professional Example — Processing Sales Data

Suppose we have a list of sales amounts and want to calculate tax for each sale.

Instead of creating a new list immediately, we can generate each taxed value only when it is needed.

````python
sales = [1200, 850, 1500, 980]

taxed_sales = (sale * 1.18 for sale in sales)

for amount in taxed_sales:
    print(amount)
````

In this example, each taxed value is calculated only when the `for` loop requests it.

If the loop stops early, the remaining calculations are never performed.

This is another example of lazy evaluation, making generator expressions useful for processing large datasets efficiently.

# Lesson 6.2 — Generator Expressions vs List Comprehensions

Generator expressions and list comprehensions have almost identical syntax.

For example:

```python
# List comprehension
[number ** 2 for number in range(5)]

# Generator expression
(number ** 2 for number in range(5))
```

The only visible difference is the brackets.

However, this small syntactic difference completely changes how Python evaluates and stores the data.

Understanding this distinction is essential because it affects:

- memory usage,
- execution strategy,
- performance,
- and the situations where each approach should be used.

In this lesson, we'll compare both approaches from a practical and professional perspective.

## Similar Syntax

Let's begin with two pieces of code that look almost identical.

````python
numbers = [number ** 2 for number in range(1, 6)]

print(numbers)
print(type(numbers))
````

````python
numbers = (number ** 2 for number in range(1, 6))

print(numbers)
print(type(numbers))
````

Although both pieces of code describe the same mathematical operation, the results are completely different.

The first creates a **list**.

The second creates a **generator object**.

This difference affects how and when the values are produced.

## List Comprehensions Use Eager Evaluation

Recall what we learned in **Topic 2 — Eager vs Lazy Evaluation**.

A list comprehension uses **eager evaluation**.

This means Python computes **every value immediately** before the list becomes available.

Consider the following example.

````python
numbers = [number ** 2 for number in range(1, 6)]

print(numbers)
````

Before `print()` is executed, Python has already calculated:

```
1² = 1
2² = 4
3² = 9
4² = 16
5² = 25
```

and stored all of them in memory.

The entire list exists before your program uses the first element.

```
range(1, 6)

↓

Calculate Every Square

↓

[1, 4, 9, 16, 25]

↓

Store Entire List in Memory

↓

Program Uses the List
```

## Generator Expressions Use Lazy Evaluation

A generator expression behaves differently.

When Python executes the following code:

```python
numbers = (number ** 2 for number in range(1, 6))
```

it **does not calculate any squares immediately**.

Instead, Python creates a generator object and waits until a value is requested.

Each value is computed only when `next()` or a `for` loop asks for it.

````python
numbers = (number ** 2 for number in range(1, 6))

print(next(numbers))
print(next(numbers))
print(next(numbers))
````

```
Generator Created

↓

No Values Calculated Yet

↓

next()

↓

Calculate 1²

↓

Return 1

↓

Pause

↓

next()

↓

Calculate 2²

↓

Return 4

↓

Pause
```

## List Comprehension vs Generator Expression

| Feature | List Comprehension | Generator Expression |
|---------|--------------------|----------------------|
| Syntax | `[]` | `()` |
| Returns | `list` | `generator` |
| Evaluation | Eager | Lazy |
| Memory Usage | Stores all values | Produces one value at a time |
| Supports Indexing | ✅ Yes | ❌ No |
| Can Be Iterated Multiple Times | ✅ Yes | ❌ No (must recreate after exhaustion) |
| Suitable for Large Data | ❌ Usually No | ✅ Yes |
| Suitable for Small Data | ✅ Yes | ✅ Yes |

## Do Generators Eventually Produce All Values?

A common question is:

> "If I iterate through the entire generator, won't all values eventually be produced anyway?"

Yes, they will.

The important difference is **when** they are produced.

A list comprehension computes **everything first**.

A generator expression computes **one value only when it is needed**.

For small datasets, this difference is usually insignificant.

For very large datasets or streaming data, the difference becomes extremely important because generators avoid storing all values in memory at once.

## Processing Customer Records

Suppose we have customer names and want to convert them to uppercase.

A list comprehension creates all transformed names immediately.

A generator expression creates each transformed name only when it is requested.

````python
customers = ["Ali", "Sara", "John", "Ayesha"]

upper_names = (customer.upper() for customer in customers)

for name in upper_names:
    print(name)
````

If the loop stops after the first two names, the remaining names are never converted.

This saves unnecessary work and memory when processing large datasets.

## Choosing Between Them

Choose a **list comprehension** when:

- You need all values immediately.
- You need indexing or slicing.
- You will iterate over the data multiple times.
- The dataset is reasonably small.

Choose a **generator expression** when:

- You're processing a large dataset.
- Values can be produced one at a time.
- Memory efficiency is important.
- You're building data pipelines or streaming data.

## Memory Usage

The biggest advantage of generator expressions is that they **do not store all values in memory at once**.

Instead, they produce one value only when it is requested.

Consider the following examples.

````python
# List comprehension
numbers = [number for number in range(1_000_000)]
````

````python
# Generator expression
numbers = (number for number in range(1_000_000))
````

Although both pieces of code describe one million numbers, they behave very differently.

### List Comprehension

Python immediately creates:

```
[0, 1, 2, 3, 4, ..., 999999]
```

Every number is stored in memory.

---

### Generator Expression

Python creates only a generator object.

```
<generator object>
```

No numbers are produced yet.

The first number is created only when the program requests it using `next()` or a `for` loop.

List Comprehension

Generate All Values
        │
        ▼
Store All Values
        │

Use the List

Generator Expression

## Why Memory Matters

Imagine processing:

- 20 million customer records
- 50 GB log file
- Millions of transactions
- Streaming sensor data
- Live API responses

Creating a list containing every record may consume a huge amount of memory or even exhaust the available RAM.

A generator avoids this problem because only one record (or a small number of records) exists in memory at any given moment.

This is one of the main reasons generators are widely used in ETL pipelines, data engineering, and large-scale data processing.

## Does Lower Memory Mean Better Performance?

Not necessarily.

Memory usage and execution speed are related, but they are **not the same thing**.

A list comprehension calculates every value once and stores the results.

After that, accessing an element is very fast because the value already exists in memory.

A generator expression calculates each value every time it is requested.

This saves memory but introduces a small amount of overhead because the generator must pause and resume execution repeatedly.

Therefore, generators are usually **more memory-efficient**, but they are **not always faster**.

## When a List Can Be Better

Suppose we need to iterate over the same data several times.

A list comprehension computes the values once.

The list can then be reused as many times as needed.

````python
squares = [number ** 2 for number in range(1, 6)]

print(sum(squares))
print(max(squares))
print(min(squares))
````

## Generators Produce Values Only Once

A generator does not store all of its values.

Once a value has been produced, it is considered consumed.

Let's see what happens.

````python
squares = (number ** 2 for number in range(1, 6))

print(sum(squares))

print(sum(squares))
````

Why did the second result become `0`?

The first call to `sum()` consumed every value produced by the generator.

By the time the second `sum()` runs, the generator has already reached the end of the sequence.

There are no values left to produce.

If you need the values again, you must create a new generator.

## Data Science Example

Suppose we are processing sales data stored in a large CSV file.

We want to calculate tax for each sale.

A generator expression allows us to process one record at a time.

````python
sales = [1200, 850, 1500, 980]

taxed_sales = (sale * 1.18 for sale in sales)

for amount in taxed_sales:
    print(amount)
````

In a real project, the `sales` values could come from a CSV file containing millions of rows.

The generator expression would calculate each taxed value only when it is needed, keeping memory usage low throughout the processing pipeline.

## Machine Learning Example

Training datasets can contain millions of samples.

Loading every transformed sample into memory at once may be impractical.

Instead, data is often generated as the training process requests it.

This is one reason generators are commonly used in data loading pipelines.

````python
images = ["img1", "img2", "img3"]

normalized_images = (
    f"Normalized({image})"
    for image in images
)

for image in normalized_images:
    print(image)
````

Although our example is small, the same approach scales naturally to much larger datasets.

Only one transformed sample needs to exist in memory at a time.

# `itertools.chain()`

## The Problem

Suppose a company stores employees from different departments separately.

```text
Sales
------
Ali
Sara

IT
----
Ahmed
Fatima

HR
----
John
Ayesha
```

Now you need to process **every employee**.

Without `itertools.chain()`, you have two common options.

### Option 1

Write multiple `for` loops.

### Option 2

Combine all lists into one large list.

Both approaches work.

However, neither is ideal when working with very large datasets.

The first approach becomes repetitive.

The second approach creates another list in memory.

Python asked a better question:

> **Why create another list when we only need to iterate through the values?**

The answer is `itertools.chain()`.

## What Is `itertools.chain()`?

`itertools.chain()` creates a single iterator that visits multiple iterables one after another.

It does **not** merge or copy them.

Instead, it simply keeps track of which iterable it is currently reading from.

Once one iterable is exhausted, it automatically continues with the next one.

Because it returns an iterator, values are produced lazily, one at a time.

## Syntax

```python
itertools.chain(iterable1, iterable2, iterable3, ...)
```

````python
import itertools

sales = ["Ali", "Sara"]
it_department = ["Ahmed", "Fatima"]
hr = ["John", "Ayesha"]

employees = itertools.chain(
    sales,
    it_department,
    hr
)

for employee in employees:
  print(employee)
````

Notice that we never created a new list containing all employees.

Instead, `itertools.chain()` created an iterator.

The iterator first reads every element from `sales`.

After that list is exhausted, it automatically continues with `it_department`, and finally with `hr`.

This is both memory-efficient and elegant.

```
Sales
 │
 ▼
Ali
Sara
 │
 ▼
IT
 │
 ▼
Ahmed
Fatima
 │
 ▼
HR
 │
 ▼
John
Ayesha
```

`itertools.chain()` simply walks through each iterable in order.

Imagine customer data is stored in multiple CSV files.

Each file has already been read into a separate list.

Instead of merging all datasets before processing, we can process them as one continuous stream.

````python
import itertools

january = [
    "Ali",
    "Sara"
]

february = [
    "Ahmed",
    "Fatima"
]

march = [
    "John"
]

customers = itertools.chain(
    january,
    february,
    march
)

for customer in customers:
   print(customer)
````

In real ETL pipelines, each iterable could represent thousands or even millions of records.

`itertools.chain()` allows the pipeline to process them sequentially without allocating memory for one massive combined list.

# `itertools.islice()`

## The Problem

You've already used list slicing:

```python
numbers[2:8]
```

This works perfectly for lists.

But what if the object is **not a list**?

Suppose it's:

- a generator,
- a file object,
- or another iterator.

These objects cannot be sliced using normal indexing.

Python needed a way to lazily retrieve only a portion of an iterator.

The solution is `itertools.islice()`.

## What Is `itertools.islice()`?

`itertools.islice()` works like normal slicing, but for **any iterator**.

Instead of returning a list, it returns another iterator.

Like the rest of `itertools`, it is lazy.

## Syntax

```python
itertools.islice(iterable, start, stop)

itertools.islice(iterable, stop)

itertools.islice(iterable, start, stop, step)
```

````python
import itertools

numbers = itertools.count(1)

first_five = itertools.islice(numbers, 5)

for number in first_five:
  print(number)
````

Notice something remarkable.

`itertools.count()` creates an **infinite iterator**.

A normal slice cannot work because there is no end.

`itertools.islice()` solves this by requesting only the first five values.

Without it, the loop would continue forever.

````python
import itertools

numbers = iter(range(20))

selected = itertools.islice(
    numbers,
    5,
    10
)

for number in selected:
    print(number)
````

Suppose a generator produces training samples.

During testing, you only want the first 100 samples.

Instead of loading the entire dataset, `itertools.islice()` can retrieve only the required portion.

````python
import itertools

samples = (
    f"Sample {index}"
    for index in range(1000)
)

test_samples = itertools.islice(
    samples,
    5
)

for sample in test_samples:
  print(sample)
````

````python
import itertools

numbers = itertools.count(6)

first_five = itertools.islice(numbers, 5, 20, 2)

for number in first_five:
    print(number)
````

# `itertools.starmap()`

## The Problem

Suppose each employee record stores two numbers.

```python
(5, 3)

(10, 2)

(7, 4)
```

Now suppose you have a function:

```python
power(base, exponent)
```

You want to call:

```python
power(5, 3)

power(10, 2)

power(7, 4)
```

Writing a loop yourself is straightforward.

However, this pattern occurs frequently when processing:

- CSV rows
- Database records
- API responses
- Coordinate pairs
- Mathematical operations

Python provides a cleaner solution:

`itertools.starmap()`.

## What Is `itertools.starmap()`?

`itertools.starmap()` applies a function to every element in an iterable.

The difference from the built-in `map()` function is that each element is automatically **unpacked** before calling the function.

If an element is:

```python
(5, 3)
```

then

```python
power(5, 3)
```

is called automatically.

````python
import itertools

def power(base, exponent):
  return base ** exponent

values = [
    (2, 3),
    (5, 2),
    (10, 3)
]

results = itertools.starmap(
    power,
    values
)

for result in results:
  print(result)
````

Notice that we never manually unpacked the tuples.

`itertools.starmap()` performed the unpacking automatically before calling the function.

This makes code cleaner and especially useful when working with structured data where each record naturally contains multiple fields.

````python
import itertools


def calculate_total(price, quantity):
    return price * quantity


orders = [
    (100, 3),
    (250, 2),
    (500, 4)
]

totals = itertools.starmap(
    calculate_total,
    orders
)

for total in totals:
    print(total)
````

In a real Data Analysis project, each tuple could represent a row from a CSV file or database.

`itertools.starmap()` allows you to pass the fields of each row directly to a function without manually unpacking them.

# `itertools.count()`

## The Problem

Suppose you're assigning invoice numbers.

```
Invoice 1001
Invoice 1002
Invoice 1003
...
```

You don't know how many invoices will be created.

You simply need the next number whenever a new invoice arrives.

One solution is:

```python
number = 1001

while True:
    print(number)
    number += 1
```

This works.

However, Python already provides an iterator specifically designed for this task.

That iterator is `itertools.count()`.

## What Is `itertools.count()`?

`itertools.count()` returns an iterator that generates consecutive numbers forever.

Each time the iterator is asked for another value, it returns the next number in the sequence.

It does **not** create an infinite list.

Instead, it calculates one number at a time using lazy evaluation.

## Syntax

```python
itertools.count(start=0, step=1)
```

### Parameters

- `start` — the first number in the sequence.
- `step` — the amount added after each value.

Both parameters are optional.

````python
import itertools

counter = itertools.count(start = 1)

for _ in range(5):
  print(next(counter))
````

Why Are We Calling ``next()`` Ourselves Here?

Because ``itertools.count()`` is infinite.

If we write:

```
for number in counter:
    print(number)
```
it will never stop.

So instead, we control how many values we want by writing:
```
for _ in range(5):
    print(next(counter))
```
This says:

"Ask the iterator for the next value exactly five times."

The underscore (_) is a convention that means:

"I don't care about this value. I'm only repeating something."

It is still a normal variable. Python doesn't treat _ specially in a for loop. We simply use it to make our intention clear to anyone reading the code.

Notice something important.

The iterator itself never stops.

We stopped after five numbers only because our `for` loop executed five iterations.

If we kept calling `next(counter)`, the sequence would continue indefinitely.

````python
import itertools

counter = itertools.count(start=10, step=5)

for _ in range(6):
    print(next(counter))
````

Imagine an automation script that creates ticket numbers.

Every new ticket should receive the next available ID automatically.

Instead of manually updating a variable, `itertools.count()` provides a clean and reusable solution.

````python
import itertools

ticket_numbers = itertools.count(start=1001)

for customer in ["Ali", "Sara", "Ahmad"]:
  print(f"Ticket {next(ticket_numbers)} ->{customer}")
````

# `itertools.cycle()`

## The Problem

Suppose three employees work on rotating shifts.

```
Ali
Sara
Ahmed
```

After Ahmed finishes, the schedule starts again with Ali.

```
Ali
Sara
Ahmed
Ali
Sara
Ahmed
...
```

Instead of manually resetting the loop every time, Python provides `itertools.cycle()`.

## What Is `itertools.cycle()`?

`itertools.cycle()` repeatedly iterates over an iterable.

After reaching the last element, it automatically starts again from the beginning.

This process continues indefinitely.

## Syntax

```python
itertools.cycle(iterable)
```

````python
import itertools

employees = ["Ali", "Sara", "Ahmed"]

rotation = itertools.cycle(employees)

for _ in range(9):
  print(next(rotation))
````

The original list contains only three names.

However, the iterator continues producing values because it keeps cycling through the iterable.

A common use case is assigning incoming requests to servers in a round-robin fashion.

Each request is sent to the next available server.

After the last server is used, assignment begins again from the first server.

````python
import itertools

servers = ["Server-1", "Server-2", "Server-3"]

assignment = itertools.cycle(servers)

for _ in range(1, 8):
  print(next(assignment))
````

# `itertools.accumulate()`

## The Problem

Imagine a company records its daily sales.

| Day | Sales |
|------|------:|
| Monday | 100 |
| Tuesday | 150 |
| Wednesday | 120 |
| Thursday | 200 |

Suppose your manager asks:

> **"What is the total sales after each day?"**

Notice that they are **not** asking only for the final total.

They want:

| Day | Running Total |
|------|--------------:|
| Monday | 100 |
| Tuesday | 250 |
| Wednesday | 370 |
| Thursday | 570 |

This is called a **running total** (also known as a cumulative total).

Without `itertools.accumulate()`, we would need to maintain a variable manually and update it during every iteration.

Since this is such a common requirement, Python provides `itertools.accumulate()`.

## What Is `itertools.accumulate()`?

`itertools.accumulate()` returns an iterator that produces the cumulative result of applying an operation to the values in an iterable.

By default, that operation is **addition**.

Unlike `sum()`, which returns only the final result,

`itertools.accumulate()` returns **every intermediate result**.

This makes it ideal for tracking progress over time.

## Syntax

```python
itertools.accumulate(iterable)

itertools.accumulate(iterable, function)
```

If no function is provided, addition is used.

````python
import itertools

sales = [100, 150, 120, 200]

running_total = itertools.accumulate(sales)

for total in running_total:
  print(total)
````

Let's see what happened internally.

| Current Value | Running Total |
|--------------:|--------------:|
| 100 | 100 |
| 150 | 250 |
| 120 | 370 |
| 200 | 570 |

Each new value is added to the previous total.

Notice that `itertools.accumulate()` did **not** return only `570`.

Instead, it produced every cumulative total, one at a time.

Suppose you're analyzing monthly profits.

Rather than looking only at the total profit at the end of the year, you may want to know how profits have grown month by month.

Running totals are commonly used in:

- Power BI
- Tableau
- Excel dashboards
- Financial reporting
- Business Intelligence

````python
import itertools

monthly_profit = [
    1200,
    1500,
    1000,
    1800
]

for total in itertools.accumulate(monthly_profit):
  print(total)
````

One of the strengths of `itertools.accumulate()` is that it is **not limited to addition**.

You can provide another function to define how values should be accumulated.

For example, we can calculate cumulative multiplication instead of cumulative addition.

````python
import itertools
import operator

numbers = [2,3,4]

result = itertools.accumulate(
    numbers,
    operator.mul
)

for value in result:
  print(value)
````

Internally, Python performs:

```
2

2 × 3 = 6

6 × 4 = 24
```

Instead of adding values, it multiplies them.

This flexibility allows `itertools.accumulate()` to perform many different cumulative calculations.

# `itertools.groupby()`

## The Problem

Suppose we have employee records.

```text
IT
IT
IT
HR
HR
Sales
Sales
Sales
```

Imagine you need to process employees department by department.

Without `itertools.groupby()`, you would have to manually detect when the department changes.

Since this pattern is very common, Python provides `itertools.groupby()`.

## What Is `itertools.groupby()`?

`itertools.groupby()` groups **consecutive** elements that have the same key.

The keyword here is:

> **Consecutive**

It does **not** search the entire iterable looking for matching values.

Instead, it starts a new group every time the key changes.

This is one of the most important characteristics of `groupby()` and the source of many beginner mistakes.

## Syntax

```python
itertools.groupby(iterable)

itertools.groupby(iterable, key=function)
```

````python
import itertools

departments = [
    "IT",
    "IT",
    "HR",
    "HR",
    "Sales",
    "Sales"
]

groups = itertools.groupby(departments)

for department, employees in groups:
  print(department, list(employees))
````

Notice that each group contains only **consecutive** matching values.

`groupby()` creates a new group whenever the value changes.

````python
import itertools

departments = [
    "IT",
    "HR",
    "IT",
    "Sales",
    "HR"
]

for department, group in itertools.groupby(departments):
    print(department, list(group))
````

Many beginners expect all `"IT"` values to be grouped together.

That does **not** happen.

Why?

Because `itertools.groupby()` only groups **adjacent (consecutive)** elements.

Think of it like reading a file from top to bottom.

Whenever the value changes, a new group begins.

It never goes back to earlier records.

Suppose an ETL pipeline has already sorted transaction records by customer.

Now we want to process one customer's transactions at a time.

Because the data is sorted, all records for the same customer are consecutive.

This makes `itertools.groupby()` an excellent choice.

````python
import itertools

transactions = [
    ("Ali", 100),
    ("Ali", 200),
    ("Sara", 300),
    ("Sara", 150),
    ("John", 400)
]

for customer, records in itertools.groupby(
    transactions,
    key = lambda record: record[0]
    ):
  print(customer)

  for record in records:
    print(" ", record)
````

This is a common pattern in ETL pipelines.

Instead of loading all of a customer's transactions into memory manually, `groupby()` processes one customer at a time as the records are read.

This is efficient for large, sorted datasets.

## Building Iterator Pipelines

One of the greatest strengths of the `itertools` module is that its functions all return iterators.

This means the output of one function can be passed directly to another without creating intermediate lists.

This style of programming is called an **iterator pipeline**.

Each function performs one small task, and together they create a complete data-processing workflow.

Because every stage is lazy, memory usage remains low even when processing very large datasets.

Suppose sales data arrives from three different branches.

We want to:

1. Treat all branches as one continuous stream.
2. Process only the first five records.

We can combine `itertools.chain()` and `itertools.islice()`.

````python
import itertools

branch_1 = [120, 150]
branch_2 = [200, 180]
branch_3 = [300, 250, 400]

sales = itertools.chain(
    branch_1,
    branch_2,
    branch_3
)

first_five = itertools.islice(sales, 5)

for sale in first_five:
  print(sale)
````

Instead of first storing all values and then calculating cumulative totals, we can directly pass one iterator into another.

This demonstrates how iterator pipelines naturally compose.

````python
import itertools

branch_1 = [100, 150]
branch_2 = [200, 250]

sales = itertools.chain(
    branch_1,
    branch_2
)

running_total = itertools.accumulate(sales)

for total in running_total:
  print(total)
````

Imagine a simple log-processing pipeline.

The logs come from two servers. We combine them, inspect only the first few records, and process them one by one.

Although this example is simple, the same pattern is used in ETL pipelines and streaming systems.

````python
import itertools

server_1 = [
    "INFO User logged in",
    "ERROR Database connection failed"
]

server_2 = [
    "INFO Backup completed",
    "ERROR Invalid password"
]

logs = itertools.chain(server_1, server_2)

recent_logs = itertools.islice(logs, 3)

for log in recent_logs:
  print(log)
````

## Mistake 1 — Converting Everything to a List

Many beginners immediately write:

```python
list(iterator)
```

Doing this removes one of the biggest advantages of iterators—**lazy evaluation**.

Only convert an iterator to a list when you genuinely need random access or need to store all values.

## Mistake 2 — Forgetting That Iterators Are Consumed

Once an iterator has produced its values, those values are gone.

```python
numbers = iter([1, 2, 3])

list(numbers)
list(numbers)
```

The second call returns an empty list because the iterator has already been exhausted.

## Mistake 3 — Misunderstanding `groupby()`

`itertools.groupby()` groups **consecutive** values, not all matching values.

If records are not grouped together, you should usually sort them first.

Functions like `itertools.count()` and `itertools.cycle()` never stop.

Always control how many values you consume using techniques such as:

- `range()`
- `itertools.islice()`
